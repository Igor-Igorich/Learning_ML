# Конспект. Модуль 11 Интерпретируемость — Feature Importance и SHAP

## 1. Зачем это нужно и как это связано с предыдущими модулями

Модули 1–10 научили нас строить точные, хорошо регуляризованные, оптимально настроенные модели. Но для банковского ML этого недостаточно если модель отклоняет кредитную заявку или блокирует транзакцию как мошенническую, регулятор, служба поддержки и сам клиент вправе спросить «почему именно это решение» — просто число `fraud_probability=0.87` на этот вопрос не отвечает. Этот модуль даёт инструментарий для ответа на два разных, но связанных вопроса

- Глобальный какие признаки модель в целом считает важными (Feature Importance)
- Локальный почему модель приняла конкретное решение по конкретному объекту (SHAP) — именно это нужно для вашего будущего эндпоинта `explain` в FraudGuard.

## 2. Три вида Feature Importance — и почему они не совпадают

### 2.1. Определения

Вспомните структуру дерева из Модуля 6 каждый узел — это разбиение по конкретному признаку, с посчитанным Gain (выигрышем) и статистикой `H` (суммой гессианов объектов, попавших в узел). На основе этой информации, накопленной по всем узлам всех деревьев ансамбля, где использовался конкретный признак `f`, строятся три разных метрики важности

- Split (в XGBoost называется `weight`) — сколько раз признак `f` вообще был выбран для разбиения, суммарно по всем деревьям. Простой счётчик использований.
- Gain — суммарный (или средний — см. важную оговорку раздела 2.3) выигрыш качества (Gain из формулы Модуля 6) от всех разбиений, где использовался признак `f`.
- Cover — среднее число объектов, «затронутых» разбиениями по признаку `f`. Важная деталь, которую стоит связать с уже пройденным материалом в точном определении XGBoost `cover` для конкретного разбиения — это буквально `H_j`, та же самая сумма гессианов узла, которую мы вывели в Модуле 6 и обсуждали применительно к `min_child_weight` в Модуле 9! Для MSE это сводится к простому числу объектов в узле; для LogLoss — к взвешенной по `p(1-p)` величине.

### 2.2. Почему они дают разные топы — численный пример

Возьмём ансамбль из нескольких деревьев (обученных на антифрод-данных) и сравним статистику по двум признакам `hour` (непрерывный, много возможных порогов) и `is_new_device` (бинарный, сильный, но редко переиспользуемый сигнал — устройство либо новое, либо нет, после одного разбиения по нему больше почти нечего уточнять).

 Метрика  `hour`  `is_new_device` 
 Split (число использований по всему ансамблю)  12  3 
 Gain (сумма по всем разбиениям)  32.0  113.0 
 Gain (в среднем на одно разбиение)  2.67  37.7 
 Cover (среднее число объектов на разбиение)  ~800  ~150 

Ранжирование по каждой метрике

 Ранг  По Split  По Gain  По Cover 
 #1  `hour`  `is_new_device`  `hour` 
 #2  `is_new_device`  `hour`  `is_new_device` 

Что произошло содержательно `hour` используется часто (12 раз против 3), но каждое отдельное разбиение по нему даёт лишь скромный выигрыш (в среднем 2.67) — это типичное поведение признака, который помогает немного уточнить решение во многих разных узлах, но никогда не является решающим фактором сам по себе. `is_new_device`, напротив, используется редко (только там, где он реально критичен), но при использовании даёт огромный выигрыш (в среднем 37.7 — более чем в 14 раз выше, чем у `hour`) — это признак с концентрированной, сильной предсказательной силой в узком контексте.

Это прямой, конкретный ответ на первый чек-поинт вопрос модуля признак может быть #1 по Split (используется чаще всех остальных), но при этом не попасть даже в топ-5 по Gain, если каждое отдельное его использование даёт лишь незначительный прирост качества — частота использования и «сила» каждого отдельного использования измеряют принципиально разные вещи, и метрика Gain обычно считается более честной оценкой реальной важности признака именно потому, что она отражает величину вклада, а не просто частоту обращения к нему.

### 2.3. Важная практическая ловушка — расхождение соглашений между библиотеками

Стоит знать заранее (это реальный источник путаницы на практике) в XGBoost метрика `importance_type='gain'` по умолчанию вычисляет среднее значение Gain по разбиениям (то, что мы обозначили как «Gain в среднем на одно разбиение» в таблице выше), а для суммарного значения нужно явно указывать `importance_type='total_gain'`. В LightGBM, напротив, `importance_type='gain'` по умолчанию возвращает суммарное значение Gain, без отдельного параметра для «среднего». То же расхождение существует и для `cover``total_cover` в XGBoost. Практический вывод если вы напрямую сравниваете графики важности признаков из XGBoost и LightGBM «в лоб», убедитесь, что оба используют один и тот же тип агрегации (сумма или среднее) — иначе легко сделать неверный вывод о том, что один признак «важнее» другого только из-за разницы в соглашениях библиотек, а не из-за реального различия в данных. (Уточняйте точное поведение по документации конкретной используемой версии библиотеки — детали API могут со временем меняться.)

## 3. От глобального к локальному зачем нужен SHAP

Feature Importance (раздел 2) отвечает на вопрос «какие признаки модель считает важными в среднем по всей обучающей выборке» — это глобальная характеристика модели целиком. Она ничего не говорит о конкретной транзакции даже если `amt_log` — самый важный признак «в среднем», для данной конкретной подозрительной транзакции решающим фактором мог оказаться совершенно другой признак (например, именно `is_new_device`, если сумма была обычной, а устройство — новым). Для эндпоинта `explain`, который должен объяснять каждое отдельное решение модели, нужен инструмент локальной интерпретации — это и есть SHAP (SHapley Additive exPlanations).

## 4. Значения Шепли интуиция из теории кооперативных игр

### 4.1. Постановка задачи

Представим признаки модели как игроков в кооперативной игре, а предсказание модели — как выигрыш, который получает коалиция (подмножество) признаков, если бы модель делала предсказание, зная только значения признаков этой коалиции (для признаков вне коалиции — их влияние усредняется по обучающей выборке, «маргинализуется»). Обозначим `v(S)` — предсказание модели (например, лог-оддс или вероятность), если известны только признаки из множества `S`.

Вопрос, на который отвечает значение Шепли признака `i` «Усреднённо по всем возможным порядкам, в которых признаки могли бы „вступать в коалицию один за другим, каков средний предельный (маргинальный) вклад признака `i` в изменение предсказания в момент его добавления»

### 4.2. Формула

```python
φ_i = Σ(S ⊆ F{i})  [ S!·(F-S-1)!  F! ]  ·  [ v(S∪{i}) - v(S) ]
```

где `F` — полное множество признаков, сумма берётся по всем подмножествам `S`, не содержащим `i`. Весовой коэффициент — это в точности вероятность того, что при равновероятном случайном порядке добавления всех `F` признаков множество `S` окажется ровно тем, что уже «добавлено» непосредственно перед `i`. Это можно понимать эквивалентно двумя способами как взвешенную сумму по всем подмножествам с этими весами, или как простое среднее предельного вклада `i` по всем `F!` возможным полным перестановкам признаков (мы используем именно этот, более наглядный способ в численном примере ниже — они математически эквивалентны).

### 4.3. Численный пример 3 признака, полный ручной расчёт

Возьмём одну конкретную (гипотетическую) подозрительную транзакцию с тремя признаками `A = amt_log`, `B = hour_night` (бинарный ночная транзакция или нет), `C = is_new_device`. Зададим (иллюстративно, как если бы это были уже посчитанные через маргинализацию значения `v(S)` обученной модели) значения для всех `2³=8` возможных подмножеств

```python
v(∅)        = 0.10   (базовая вероятность без учёта каких-либо признаков)
v({A})      = 0.20
v({B})      = 0.15
v({C})      = 0.45
v({A,B})    = 0.25
v({A,C})    = 0.55
v({B,C})    = 0.50
v({A,B,C})  = 0.60   (итоговое предсказание модели для этой транзакции)
```

Перебираем все `3!=6` порядков добавления признаков и считаем предельный вклад каждого признака в момент его добавления к уже накопленной коалиции

 Порядок  Вклад A  Вклад B  Вклад C 
 A->B->C  `v(A)-v(∅)=0.10`  `v(AB)-v(A)=0.05`  `v(ABC)-v(AB)=0.35` 
 A->C->B  `v(A)-v(∅)=0.10`  `v(ABC)-v(AC)=0.05`  `v(AC)-v(A)=0.35` 
 B->A->C  `v(AB)-v(B)=0.10`  `v(B)-v(∅)=0.05`  `v(ABC)-v(AB)=0.35` 
 B->C->A  `v(ABC)-v(BC)=0.10`  `v(B)-v(∅)=0.05`  `v(BC)-v(B)=0.35` 
 C->A->B  `v(AC)-v(C)=0.10`  `v(ABC)-v(AC)=0.05`  `v(C)-v(∅)=0.35` 
 C->B->A  `v(ABC)-v(BC)=0.10`  `v(BC)-v(C)=0.05`  `v(C)-v(∅)=0.35` 

Значение Шепли — среднее по столбцу

```python
φ_A = 0.10   φ_B = 0.05   φ_C = 0.35
```

Проверка аддитивности (раздел 5) `φ_A+φ_B+φ_C = 0.10+0.05+0.35 = 0.50`, а `v({A,B,C}) - v(∅) = 0.60-0.10 = 0.50` — точное совпадение.

Важная честная оговорка про этот пример. Обратите внимание, что вклад каждого признака оказался одинаковым в любом порядке добавления (например, `A` всегда даёт `+0.10`, независимо от того, что уже было добавлено до него). Это не случайность вычислений, а следствие того, что заданные значения `v(S)` в этом примере полностью аддитивны — между признаками нет никакого «взаимодействия» (синергии или гашения эффектов). В такой ситуации усреднение по всем порядкам буквально не требовалось бы — любой порядок дал бы тот же ответ. Реальные модели так почти никогда себя не ведут — признаки обычно взаимодействуют.

### 4.4. Модифицированный пример когда усреднение по порядкам реально необходимо

Изменим только одно значение пусть `v({A,C}) = 0.65` (было `0.55`) — это моделирует синергию крупная сумма транзакции (`A`) в сочетании с новым устройством (`C`) вместе создают заметно более сильный сигнал риска, чем можно было бы ожидать, просто сложив их отдельные эффекты (совершенно реалистичный сценарий для антифрода — именно комбинация факторов часто говорит о захвате аккаунта сильнее, чем каждый фактор по отдельности).

Пересчитаем вклад `A` в тех порядках, где он идёт сразу после `C`

- Порядок `C->A->B` вклад `A` = `v(A,C)-v(C) = 0.65-0.45 = 0.20` (было `0.10`!)
- Остальные 5 порядков не затрагивают напрямую пару `(A,C)` без промежуточного `B`, их вклад `A` остаётся `0.10` каждый.

Новое среднее `φ_A = (0.10×5 + 0.20)  6 = 0.706 ≈ 0.117`.

Вот почему усреднение по всем порядкам принципиально важно предельный вклад `A` зависит от того, был ли уже «добавлен» `C` до него — в порядках, где это так, синергия проявляется и вклад `A` больше; в остальных порядках синергия не проявлена (потому что мы ещё не «узнали» про `C`). Единственный статистически честный способ получить одно итоговое число, справедливо отражающее вклад `A` с учётом всех возможных взаимодействий — усреднить предельный вклад по всем возможным порядкам «раскрытия» информации о признаках, что и делает формула Шепли.

## 5. Аддитивность SHAP — ключевое свойство

Формальное утверждение, которое мы уже проверили численно в разделе 4.3

```python
F(x) = base_value + Σ(i=1..M) φ_i(x)
```

где `base_value = E[F(X)]` — среднее предсказание модели по всей обучающей выборке (концептуальный аналог `F_0` из Модуля 3–4 — «предсказание без какой-либо конкретной информации об объекте»), а `φ_i(x)` — SHAP-значение признака `i` именно для данного объекта `x`.

Что означает «аддитивность» простыми словами (прямой ответ на второй чек-поинт вопрос модуля) сумма вкладов всех признаков для конкретного объекта, плюс базовое (среднее по выборке) предсказание, в точности (не приближённо!) равна итоговому предсказанию модели для этого объекта. Это не просто удобное свойство «для наглядности» — это математическая гарантия, прямо следующая из формулы Шепли (в частности, из того, что сумма всех весов `S!(F-S-1)!F!` по всем `S` в точности равна `1` — вы можете убедиться в этом на численном примере раздела 4.3, где мы явно проверили точное равенство). Именно эта гарантия делает возможным водопадные диаграммы (waterfall plots) каждый признак честно добавляет свой кусочек, и сумма всех кусочков плюс базовая линия обязана дать ровно итоговое предсказание — ничего не «теряется» и не «додумывается».

## 6. TreeSHAP точный расчёт за полиномиальное время

### 6.1. Проблема общего (model-agnostic) подхода

Прямой расчёт формулы Шепли требует перебора всех `2^M` подмножеств признаков — при `M=150` признаках (как в вашем FraudGuard) это `2^150` — вычислительно абсолютно невозможно. KernelSHAP — универсальный, работающий для любой модели (не только деревьев) алгоритм — обходит эту проблему, приближённо оценивая значения Шепли через взвешенную линейную регрессию на случайной выборке подмножеств (а не на всех `2^M`). Это даёт приближённый результат с некоторой статистической погрешностью (зависящей от числа сэмплированных подмножеств) и требует много обращений к самой модели (каждая оценка `v(S)` — это фактический прогон модели) — может быть медленным, особенно при большом числе признаков.

### 6.2. Почему для деревьев можно точно и быстро

TreeSHAP (Lundberg et al., 2018) использует то, что структура дерева уже явно кодирует, как признак влияет на маршрут объекта к листу. Ключевая идея если признак известен (входит в `S`), объект однозначно идёт по одной из двух ветвей узла (как обычно при инференсе). Если признак неизвестен (не входит в `S`), нужно усреднить результат по обеим ветвям, взвесив их по тому, сколько обучающих объектов прошло через каждую ветвь при обучении — а это в точности та же статистика Cover (`H_j`, сумма гессианов узла), которую мы уже обсуждали в разделе 2.1 этого же модуля! TreeSHAP использует рекурсивный, динамический обход дерева (по сути, аналог динамического программирования — каждое поддерево пересчитывается один раз, а не многократно для каждого отдельного подмножества `S`), что позволяет получить точный (не приближённый!) ответ, эквивалентный полному перебору всех `2^M` подмножеств, но за полиномиальное время

```python
Сложность TreeSHAP  O(T · L · D²)     вместо наивных     O(T · L · 2^M)
```

где `T` — число деревьев, `L` — максимальное число листьев в дереве, `D` — максимальная глубина, `M` — число признаков. Обратите внимание сложность TreeSHAP вообще не зависит экспоненциально от `M` — только от размера и глубины самих деревьев, что и делает точный расчёт SHAP на моделях с сотнями признаков практически осуществимым за разумное время.

## 7. Глобальная vs локальная интерпретация — два разных графика

- `summary_plot` — глобальный взгляд для каждого объекта обучающей (или тестовой) выборки и каждого признака откладывается его SHAP-значение — получается «облако точек», по которому видно не только какие признаки в среднем важнее (аналог Feature Importance, но теперь честно выведенный из аддитивных вкладов, а не из SplitGainCover), но и направление эффекта (высокие значения признака систематически повышают или понижают предсказание).
- `waterfall_plot` — локальный взгляд для одного конкретного объекта начинается с `base_value`, затем каждый признак «добавляет» свой кусочек (положительный или отрицательный), и сумма всех кусочков приводит ровно к итоговому предсказанию для этого объекта (раздел 5) — это буквально то, что должен возвращать ваш эндпоинт `explain` для конкретной подозрительной транзакции показать, какие именно признаки и насколько подняли или опустили итоговый скор относительно среднего.

## 8. Практика код

### 8.1. Базовое использование `TreeExplainer`

```python
import shap
import lightgbm as lgb
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(
    n_samples=20000, n_features=15, n_informative=8,
    weights=[0.92, 0.08], random_state=42
)
feature_names = [f"feature_{i}" for i in range(15)]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
model = lgb.LGBMClassifier(
    num_leaves=31, n_estimators=200,
    learning_rate=0.05, random_state=42, verbose=-1
)
model.fit(X_train, y_train)
```

```python
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)
```

```python
# Для бинарной классификации shap_values может быть списком [для класса 0, для класса 1]
# или одним массивом с логит-вкладами - поведение зависит от версии библиотеки shap.
print("Форма массива SHAP-значений", np.array(shap_values).shape)
print("base_value", explainer.expected_value)
```

### 8.2. Глобальная картина — `summary_plot`

```python
shap.summary_plot(shap_values, X_test, feature_names=feature_names)
```

Что искать признаки отсортированы по важности сверху вниз (это тот же принцип, что Gain-важность, но выведенный строго через аддитивные вклады). Цвет точки обычно показывает значение признака (красный — высокое, синий — низкое) для конкретного объекта, а положение по горизонтали — его SHAP-вклад для этого объекта. Если для важного признака красные точки систематически справа (положительный вклад), а синие — слева, значит высокие значения признака повышают предсказание модели — прямая, наглядная интерпретация направления эффекта, которую Feature Importance (раздел 2) в принципе дать не может (она даёт только силу, но не направление влияния).

### 8.3. Локальное объяснение — `waterfall_plot` для одной транзакции

```python
# Найдём объект, для которого модель предсказала высокую вероятность мошенничества
proba = model.predict_proba(X_test)[:, 1]
suspicious_idx = proba.argmax()
```

```python
explanation = shap.Explanation(
    values=shap_values[1][suspicious_idx] if isinstance(shap_values, list) else shap_values[suspicious_idx],
    base_values=explainer.expected_value[1] if isinstance(explainer.expected_value, (list, np.ndarray)) else explainer.expected_value,
    data=X_test[suspicious_idx],
    feature_names=feature_names,
)
shap.plots.waterfall(explanation)

print(f"Итоговая вероятность {proba[suspicious_idx]:.4f}")
print(f"База (среднее по выборке) {explainer.expected_value}")
```

Это — прямая заготовка для вашего эндпоинта `explain` из спецификации FraudGuard вместо вызова `.plots.waterfall()` (который рисует картинку) вы бы вернули топ-N признаков с наибольшими по модулю SHAP-значениями в виде JSON — именно так устроено большинство реальных production-эндпоинтов объяснимости.

## 9. Частые вопросы на собеседовании

Вопрос  На что обратить внимание в ответе 
 Почему признак может быть #1 по Split, но не топ-5 по Gain  Split измеряет частоту использования, Gain — суммарную величину выигрыша; признак может использоваться часто, но давать лишь небольшой выигрыш при каждом использовании, тогда как другой признак — использоваться редко, но каждый раз крайне результативно 
 Что означает «аддитивность» SHAP-значений  Сумма SHAP-вкладов всех признаков для конкретного объекта плюс базовое значение (среднее предсказание по выборке) в точности, без приближений, равна итоговому предсказанию модели для этого объекта 
 Почему TreeSHAP даёт точный, а не приближённый результат, в отличие от KernelSHAP  TreeSHAP использует явную структуру дерева для эффективного (полиномиального) точного вычисления через рекурсивный обход, тогда как KernelSHAP — модель-агностичный метод, приближающий значения Шепли через регрессию на случайной выборке подмножеств признаков 
 Зачем вообще усреднять предельный вклад признака по всем порядкам, если можно просто посмотреть на один «естественный» порядок  При наличии взаимодействий между признаками (синергии или взаимного гашения эффектов) предельный вклад признака зависит от того, какие признаки уже «известны» — единственный статистически честный способ получить одно справедливое число — усреднить по всем возможным порядкам раскрытия информации 
 В чём разница между `cover`-важностью признака и его буквальным «числом объектов в узле»  В точном определении cover — это сумма гессианов (`H_j`), а не просто счётчик объектов; для MSE это совпадает с простым счётчиком (гессиан=1), но для LogLoss и других функций потерь — это взвешенная по локальной кривизне функции потерь величина (та же логика, что у `min_child_weight`, Модуль 9)

summaryРаскрыть после того, как попробуете ответить самиsummary

1. Split измеряет частоту — сколько раз признак вообще участвовал в разбиении по всему ансамблю. Gain измеряет величину — суммарный выигрыш качества от всех этих разбиений. Признак, который часто используется как «второстепенный уточнитель» (небольшой, но стабильный выигрыш при каждом использовании), может набрать высокий Split, но скромный суммарный Gain; признак, который используется редко, но исключительно результативно (большой выигрыш при каждом редком использовании), — наоборот. Численный пример раздела 2.2 демонстрирует это напрямую `hour` (12 использований, средний Gain 2.67) против `is_new_device` (3 использования, средний Gain 37.7).

2. Сумма SHAP-вкладов всех признаков объекта плюс базовое (среднее по выборке) предсказание модели в точности, без приближений, восстанавливает итоговое предсказание модели для этого конкретного объекта — ни один «кусочек» вклада не теряется и не додумывается, все признаки честно «складываются» в финальный результат.

3. Потому что в исходном примере значения `v(S)` были заданы полностью аддитивными — не было никакого взаимодействия (синергии) между признаками, поэтому предельный вклад каждого признака оказывался одним и тем же независимо от того, в каком порядке и после каких других признаков он добавлялся в коалицию — усреднение по всем порядкам в этом частном случае просто возвращало бы то же самое число, что и любой отдельный порядок. В модифицированном примере введена синергия между `A` и `C` (`v({A,C})` стало непропорционально больше суммы отдельных эффектов) — теперь предельный вклад `A` зависит от того, был ли уже «добавлен» `C` до него, и только усреднение по всем возможным порядкам даёт единственный статистически честный, учитывающий это взаимодействие ответ.

4. KernelSHAP — модель-агностичный (работает для любой модели, включая нейросети, линейные модели, что угодно) приближённый метод он оценивает значения Шепли через взвешенную линейную регрессию на случайно сэмплированной выборке подмножеств признаков, поскольку не имеет доступа к внутренней структуре модели и вынужден многократно «спрашивать» модель как чёрный ящик. TreeSHAP применим только к моделям на основе деревьев именно потому, что он использует явную, известную заранее структуру дерева (какие признаки и пороги используются в каждом узле, сколько обучающих объектов прошло по каждой ветви) для точного, эффективного рекурсивного вычисления — для моделей без такой явной древовидной структуры (например, нейросети) этот конкретный алгоритм неприменим, там используются другие методы (например, DeepSHAP).

5. Столбчатая диаграмма Feature Importance даёт только одно число на признак, усреднённое по всей выборке — она не показывает, как именно значение признака влияет на предсказание (повышает или понижает его) и не показывает разброс этого влияния между разными объектами (может ли один и тот же признак у разных объектов давать очень разный по знаку эффект). `summary_plot`, показывая SHAP-значение каждого объекта по каждому признаку одновременно с цветовой кодировкой значения самого признака, даёт возможность увидеть и силу, и направление эффекта, и его вариативность по всей выборке — гораздо более богатую картину поведения модели.
