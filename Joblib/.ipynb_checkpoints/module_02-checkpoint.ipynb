# Модуль 2. Экономим диск и память: сжатие и memory mapping

## Зачем этот модуль

В Модуле 1 мы научились сохранять модели. Но в реальной жизни обученные модели могут весить **сотни мегабайт или даже гигабайты**. Например:
- `RandomForest` с 500 деревьями на 200 признаках — легко 50–100 МБ.
- Матрица весов нейросети (хотя joblib чаще для классических ML) — гигабайты.
- Предобученные эмбеддинги — десятки гигабайт.

Если вы сохраняете модель «как есть», вы:
- тратите много места на диске;
- долго передаёте файл по сети (например, в S3 или на сервер);
- долго загружаете в память при старте backend-приложения.

В этом модуле мы научимся **сжимать** файлы joblib и **загружать их частично** — без выгрузки всего содержимого в оперативную память сразу.

## 1. Проблема размера: наглядный пример

Давайте создадим модель и посмотрим, сколько она весит без сжатия.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
from joblib import dump
import os
```

```python
# Создаём большой датасет
X, y = make_classification(
    n_samples=50_000,      # 50 тысяч строк
    n_features=100,        # 100 признаков
    random_state=42
)

# Обучаем лес из 300 деревьев
model = RandomForestClassifier(n_estimators=300, random_state=42)
model.fit(X, y)

# Сохраняем без сжатия
dump(model, 'model_uncompressed.joblib')

size_mb = os.path.getsize('model_uncompressed.joblib') / (1024 * 1024)
print(f"Размер файла: {size_mb:.2f} МБ")
```

**Типичный результат:** 50–150 МБ.

Если у вас 10 моделей — уже гигабайт. Если вы деплоите в Docker-контейнер с лимитом 500 МБ — модель может не влезть. Нужно сжимать.

## 2. Параметр `compress`: как он работает

### Синтаксис

```python
dump(obj, filename, compress=...)
```

Параметр `compress` говорит `joblib`: «перед записью на диск сожми данные».

### Виды значений `compress`

| Значение | Что происходит |
|----------|---------------|
| `0` или `False` | Без сжатия (по умолчанию) |
| `1`–`9` | Сжатие `zlib` с указанным уровнем. Число больше — сильнее сжатие, но медленнее |
| `('zlib', 3)` | Явное указание алгоритма и уровня |
| `('gzip', 3)` | Алгоритм gzip, уровень 3 |
| `('bz2', 3)` | Алгоритм bz2, уровень 3 |
| `('lzma', 3)` | Алгоритм lzma, уровень 3 |
| `('xz', 3)` | Алгоритм xz, уровень 3 |

### Простой пример

```python
from joblib import dump

# Уровень 3 — хороший баланс
dump(model, 'model_compressed.joblib', compress=3)
size_mb = os.path.getsize('model_uncompressed.joblib') / (1024 * 1024)
print(f"Размер файла: {size_mb:.2f} МБ")
```

### Как это работает «под капотом»

Когда вы указываете `compress`, `joblib`:
1. берёт ваш объект;
2. сериализует его;
3. пропускает через алгоритм сжатия;
4. записывает сжатые байты в файл.

При загрузке (`load`) происходит обратный процесс: файл распаковывается автоматически, и вы получаете исходный объект. **Вам не нужно ничего делать дополнительно** — `load` сам понимает, что файл сжат.

## 3. Алгоритмы сжатия: что выбрать

### Сравнение алгоритмов

| Алгоритм | Скорость сжатия | Степень сжатия | Когда использовать |
|----------|----------------|----------------|-------------------|
| **zlib** | Быстрая | Средняя | **По умолчанию**. Лучший баланс для ML-моделей |
| **gzip** | Быстрая | Средняя | Совместимость с Unix-утилитами (`gunzip`) |
| **bz2** | Медленная | Хорошая | Если нужно сильнее сжать, чем zlib |
| **lzma / xz** | Очень медленная | Очень сильная | Для архивного хранения, когда размер критичен |
| **None** | Мгновенно | Нет | Для скорости загрузки, если места много |

### Правило большого пальца

- **`compress=3`** — золотая середина. Файл уменьшится на 30–60%, скорость записи/чтения останется приемлемой.
- **`compress=('lzma', 6)`** — если вы отправляете модель в облако один раз, а читать будете редко. Максимальное сжатие.
- **Без сжатия** — если модель маленькая (< 5 МБ) или если она загружается очень часто (например, на каждый запрос в API).

### Наглядное сравнение

```python
import time
import os
from joblib import dump, load
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=20_000, n_features=50, random_state=42)
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

configs = [
    (0, 'Без сжатия'),
    (3, 'zlib=3'),
    (('zlib', 6), 'zlib=6'),
    (('gzip', 3), 'gzip=3'),
    (('bz2', 3), 'bz2=3'),
    (('lzma', 3), 'lzma=3'),
]

print(f"{'Вариант':<15} {'Размер (МБ)':<12} {'Время записи (с)':<18}")
print("-" * 50)

for compress, name in configs:
    filename = f'model_{name.replace("=", "_").replace(" ", "_")}.joblib'
    
    t0 = time.time()
    dump(model, filename, compress=compress)
    t_write = time.time() - t0
    
    size = os.path.getsize(filename) / (1024 * 1024)
    print(f"{name:<15} {size:<12.2f} {t_write:<18.3f}")
```

**Типичный результат:**

```python
Вариант         Размер (МБ)  Время записи (с)  
--------------------------------------------------
Без сжатия      25.40        0.120             
zlib=3          8.50         0.180             
zlib=6          7.20         0.350             
gzip=3          8.55         0.190             
bz2=3           6.80         0.600             
lzma=3          5.10         2.100             
```

**Вывод:** `zlib=3` даёт уменьшение размера в 3 раза при небольшом замедлении. `lzma` сжимает сильнее, но в 10 раз медленнее.

## 4. Memory mapping: загрузка без полного чтения в память

### Проблема: модель не влезает в RAM

Представьте: у вас сервер с 4 ГБ оперативной памяти, а модель весит 2 ГБ. Если 10 пользователей одновременно загрузят модель — память закончится.

Но что, если модель использует только **часть** своих данных для каждого предсказания? Или если вы хотите загрузить массив, но работать с кусками?

### Что такое memory mapping

**Memory mapping** (отображение в память) — это механизм операционной системы, который говорит: «не загружай весь файл в RAM сразу. Загружай только те куски, к которым программа обратилась».

Аналогия: вместо того чтобы фотокопировать всю энциклопедию, вы оставляете её на полке и открываете только нужные страницы.

### Как использовать в joblib

```python
from joblib import load

# 'r' = read-only memory mapping
data = load('big_array.joblib', mmap_mode='r')

print(type(data))  # numpy array
print(data.shape)  # форма сохранена
```

### Режимы `mmap_mode`

| Режим | Значение |
|-------|----------|
| `'r'` | Только чтение. Файл не изменится. Самый безопасный. |
| `'r+'` | Чтение и запись. Изменения сохранятся в файл. |
| `'w+'` | Чтение и запись, но файл должен уже существовать. |

> **Backend-совет:** всегда используйте `mmap_mode='r'` для production-моделей. Это защищает от случайного изменения файла и позволяет нескольким процессам читать один файл без дублирования в памяти.

### Ограничение memory mapping

`mmap_mode` работает только для объектов, содержащих **numpy-массивы**. Если вы сохранили словарь с моделью sklearn, `mmap_mode` не даст эффекта — потому что модель sklearn состоит из множества Python-объектов, а не одного большого массива.

Но если вы сохраняете, например, матрицу эмбеддингов или большой датасет — `mmap_mode` спасает.

### Пример с большим массивом

```python
import numpy as np
from joblib import dump, load
import os

# Создаём массив 100 млн чисел (~762 МБ в памяти)
big_array = np.random.rand(100_000_000)

# Сохраняем
dump(big_array, 'big_array.joblib')

# --- Обычная загрузка ---
import psutil, os
process = psutil.Process(os.getpid())
mem_before = process.memory_info().rss / 1024 / 1024

arr_normal = load('big_array.joblib')
mem_after = process.memory_info().rss / 1024 / 1024
print(f"Обычная загрузка: +{mem_after - mem_before:.0f} МБ RAM")

# --- Memory-mapped загрузка ---
mem_before = process.memory_info().rss / 1024 / 1024
arr_mmap = load('big_array.joblib', mmap_mode='r')
mem_after = process.memory_info().rss / 1024 / 1024
print(f"MMAP загрузка: +{mem_after - mem_before:.0f} МБ RAM")

# Читаем только первые 1000 элементов — подгрузится только этот кусок
print(arr_mmap[:1000].mean())
```

**ВНИМАНИЕ:** работает только на несжатых файлах

## 5. Комбинация: сжатие + memory mapping

Важный нюанс: **memory mapping НЕ работает со сжатыми файлами**. Если файл сжат (`compress > 0`), его нельзя отобразить в память напрямую — сначала нужно распаковать весь файл.

Поэтому выбирайте:

| Сценарий | compress | mmap_mode |
|----------|----------|-----------|
| Модель sklearn, загружается при старте сервера | `3` | не используем |
| Большой numpy-массив, читаем кусками | `0` | `'r'` |
| Архивное хранение модели | `('lzma', 6)` | не используем |
| Частая загрузка, критична скорость | `0` | не используем |

## 6. Backend-сценарии

### Сценарий A: Деплой в Docker

```python
FROM python:3.11-slim

COPY requirements.txt .
RUN pip install -r requirements.txt

# Модель сжата — образ меньше
COPY model_compressed.joblib /app/model.joblib

CMD ["uvicorn", "main:app", "--host", "0.0.0.0"]
```

Сжатая модель уменьшает размер Docker-образа и ускоряет `docker push/pull`.

### Сценарий B: Загрузка модели в FastAPI с lazy loading

```python
from fastapi import FastAPI
from joblib import load

app = FastAPI()

_model = None

def get_model():
    global _model
    if _model is None:
        # Загружаем один раз при первом запросе
        _model = load('model.joblib')  # compress здесь не нужен, load сам поймёт
    return _model

@app.post("/predict")
def predict(features: list[float]):
    model = get_model()
    return {"prediction": model.predict([features]).tolist()}
```

> Примечание: `compress` указывается при `dump`, при `load` он определяется автоматически по заголовку файла.

### Сценарий C: Хранение в S3

```python
import boto3
from joblib import dump, load
import io

# Сохраняем в байтовый поток с сжатием
buffer = io.BytesIO()
dump(model, buffer, compress=3)
buffer.seek(0)

# Загружаем в S3
s3 = boto3.client('s3')
s3.put_object(Bucket='my-models', Key='v1/model.joblib', Body=buffer.getvalue())

# Скачиваем обратно
obj = s3.get_object(Bucket='my-models', Key='v1/model.joblib')
model = load(io.BytesIO(obj['Body'].read()))
```

Сжатие критично: передача 8 МБ вместо 25 МБ по сети экономит время и деньги.

## 8. Эталонное решение

### Решение 2.1 (фрагмент)

```python
import time
import os
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
from joblib import dump, load

X, y = make_classification(n_samples=10_000, n_features=50, random_state=42)
model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X, y)

configs = [
    (0, 'none'),
    (3, 'zlib3'),
    (('lzma', 3), 'lzma3'),
    (('bz2', 3), 'bz2_3'),
]

results = []
for compress, label in configs:
    fname = f'model_{label}.joblib'
    
    t0 = time.time()
    dump(model, fname, compress=compress)
    t_write = time.time() - t0
    
    t0 = time.time()
    _ = load(fname)
    t_read = time.time() - t0
    
    size = os.path.getsize(fname) / (1024 * 1024)
    results.append((label, size, t_write, t_read))

print(f"{'Вариант':<10} {'Размер, МБ':<12} {'Запись, с':<12} {'Чтение, с':<12}")
for r in results:
    print(f"{r[0]:<10} {r[1]:<12.2f} {r[2]:<12.3f} {r[3]:<12.3f}")
```

## 9. Вопросы для самопроверки

1. **Что делает параметр `compress=3` в `joblib.dump()`?**  
   *(Ответ: сжимает файл алгоритмом zlib с уровнем 3 перед записью на диск.)*

2. **Почему `lzma` даёт меньший размер файла, чем `zlib`, но работает медленнее?**  
   *(Ответ: lzma использует более сложный алгоритм сжатия, который лучше убирает избыточность, но требует больше вычислений.)*

3. **Можно ли использовать `mmap_mode='r'` на сжатом файле?**  
   *(Ответ: нет. Для memory mapping файл должен быть несжатым, потому что ОС не может отобразить сжатые данные напрямую в память.)*

4. **Какой уровень сжатия рекомендуется для ежедневного использования?**  
   *(Ответ: `compress=3` — хороший баланс между размером и скоростью.)*

5. **Почему сжатие важно при деплое в Docker?**  
   *(Ответ: уменьшает размер образа, ускоряет сборку, push и pull контейнера.)*

## Итоги модуля

- **Сжатие** через `compress` уменьшает размер файлов `.joblib` в 2–5 раз без потери данных.
- Алгоритм `zlib` с уровнем 3 — оптимальный выбор для большинства ML-моделей.
- **Memory mapping** (`mmap_mode='r'`) позволяет работать с большими массивами, не загружая их целиком в RAM.
- Сжатие и memory mapping **несовместимы**: выбирайте один из подходов в зависимости от задачи.
- В production сжатие экономит место в хранилище и время передачи по сети.

**В следующем модуле** мы изучим второй инструмент `joblib` — **`Memory`**, который позволяет кешировать результаты функций и не пересчитывать их дважды.
