### 1.1. Типы признаков в ML: зачем это важно

Модель машинного обучения «видит» данные не так, как человек. Для неё всё — это числа. Но **как** мы превратим текст, дату или категорию в числа — зависит от типа признака. Если перепутать тип, модель выучит бессмыслицу.

#### Числовые признаки (Numerical)

**Непрерывные** — могут быть любыми числами в диапазоне.

| Пример | Значения | Почему непрерывный |
|--------|----------|-------------------|
| Рост человека | 175.3 см, 160.0 см | Между 175 и 176 есть бесконечное число значений |
| Температура | 36.6°, 37.2° | Может быть 36.65°, 36.651°... |
| Цена товара | 999.50 ₽, 15000.00 ₽ | Любое положительное число |

**Дискретные** — только отдельные, «цельные» значения (обычно целые числа).

| Пример | Значения | Почему дискретный |
|--------|----------|-------------------|
| Количество детей | 0, 1, 2, 3 | Не бывает 2.5 детей |
| Этаж дома | 1, 2, 3... | Между 5 и 6 этажом ничего нет |
| Число покупок | 0, 1, 2... | Целое число |

> **Важно:** иногда дискретный признак выглядит как непрерывный. Например, «возраст в годах» — дискретный (34 года, 35 лет), но мы часто считаем его непрерывным для простоты.

#### Категориальные признаки (Categorical)

**Номинальные** — группы без порядка. Нельзя сказать, что одно «больше» другого.

```
# Примеры номинальных признаков
город = ['Москва', 'СПб', 'Казань', 'Новосибирск']
цвет_глаз = ['голубые', 'карие', 'зелёные', 'серые']
марка_авто = ['Toyota', 'BMW', 'Lada']
```

**Порядковые** — группы с естественным порядком. Можно сравнивать.

```
# Примеры порядковых признаков
образование = ['начальное', 'среднее', 'высшее']  # высшее > среднее
категория_отеля = [1, 2, 3, 4, 5]  # 5★ лучше 3★
уровень_удовлетворенности = ['низкий', 'средний', 'высокий']
```

> **Ловушка:** модель не знает, что «высшее» лучше «начального». Если закодировать образование числами 1, 2, 3 — модель подумает, что 3 в 3 раза больше 1. Для порядковых признаков нужен **Ordinal Encoding**, для номинальных — **One-Hot Encoding** (подробно в Модуле 8).

#### Дата и время (Datetime)

Выглядят как текст, но содержат богатую информацию:

```
дата = '2024-03-15 14:30:00'
# Можно извлечь:
# - год: 2024
# - месяц: 3 (март)
# - день: 15
# - день недели: пятница (5)
# - час: 14
# - сезон: весна
# - выходной? да/нет
```

Даты требуют специального формата. В pandas это `datetime64`.

#### Текст (Text)

Свободные строки, которые человек пишет сам:

```
отзыв = "Отличный товар, быстрая доставка, рекомендую!"
описание = "Квартира в центре, 2 комнаты, балкон, вид на реку"
```

Текст нельзя напрямую скормить модели. Нужно извлекать признаки (длина, ключевые слова) или использовать специальные методы (TF-IDF, эмбеддинги).

#### Геоданные (Geospatial)

```
координаты = (55.7558, 37.6173)  # широта, долгота — Москва, Красная площадь
```

Из координат можно создать признаки: расстояние до центра, район, плотность застройки.

### 1.2. Источники данных

#### CSV (Comma-Separated Values)
Самый популярный формат. Таблица в виде текста, где значения разделены запятыми.

```
имя,возраст,город
Анна,25,Москва
Борис,30,СПб
```

Плюсы: простой, универсальный, мало весит.
Минусы: нет форматов ячеек, проблемы с кодировкой, разделитель может быть точкой с запятой (`;`) в русской локали.

#### Excel (.xlsx, .xls)
Формат Microsoft Excel. Может содержать несколько листов, формулы, форматирование.

Плюсы: удобно для людей, несколько листов.
Минусы: медленно читается, тяжёлые файлы, может содержать скрытые данные.

#### JSON (JavaScript Object Notation)
Формат для передачи данных в интернете. Похож на Python-словари.

```
{
  "имя": "Анна",
  "возраст": 25,
  "город": "Москва"
}
```

Плюсы: удобен для API, поддерживает вложенность.
Минусы: для таблиц требует «разворачивания» (flattening).

#### SQL-базы данных
Данные хранятся в базах: PostgreSQL, MySQL, SQLite. Доступ через язык запросов SQL.

```
SELECT имя, возраст, город FROM клиенты WHERE возраст > 18;
```

Плюсы: быстрая фильтрация, большие объёмы.
Минусы: нужно знать SQL.

#### API (Application Programming Interface)
Программный интерфейс для получения данных из интернета. Например, погода, курсы валют, данные о фильмах.

Пример (псевдокод):

```
import requests
response = requests.get('https://api.example.com/weather?city=Moscow')
data = response.json()
```

#### Парсинг веб-страниц (Web Scraping)
Когда данных нет в удобном формате, но они есть на сайте. Используют библиотеки `requests` + `BeautifulSoup` или `Selenium`.

> **Важно:** всегда проверяйте лицензию и `robots.txt` сайта перед парсингом.

### 1.7. Работа с памятью: оптимизация типов данных

#### Почему это важно
Если у вас 10 миллионов строк, таблица может занимать гигабайты оперативной памяти. Но часто данные можно хранить компактнее **без потери информации**.

#### Типы данных в pandas

| Тип | Что хранит | Диапазон | Память |
|-----|-----------|----------|--------|
| `int64` | Целые числа | -9×10¹⁸ до 9×10¹⁸ | 8 байт |
| `int32` | Целые числа | -2×10⁹ до 2×10⁹ | 4 байта |
| `int16` | Целые числа | -32768 до 32767 | 2 байта |
| `int8` | Целые числа | -128 до 127 | 1 байт |
| `float64` | Числа с дробью | Очень большой | 8 байт |
| `float32` | Числа с дробью | Большой | 4 байта |
| `object` | Текст | Любой | Много (16+ байт на символ) |
| `category` | Категории | Ограниченный набор | 1–8 байт |

#### Пример оптимизации

```
# До оптимизации
df.info()
# memory usage: 800.0 MB

# 1. Оптимизируем целые числа
df['возраст'] = df['возраст'].astype('int8')  # возраст 0–120 влезает в int8

# 2. Оптимизируем числа с плавающей точкой
df['доход'] = df['доход'].astype('float32')   # точности float32 хватит для денег

# 3. Категориальные столбцы -> category
df['город'] = df['город'].astype('category')

# После оптимизации
df.info()
# memory usage: 150.0 MB  <- в 5 раз меньше!
```

```
def optimize_memory(df):
    """Автоматически уменьшает размер DataFrame"""
    start_mem = df.memory_usage().sum() / 1024**2
    print(f'Начальный размер: {start_mem:.2f} MB')
    
    for col in df.columns:
        col_type = df[col].dtype
        
        if col_type != object:
            c_min = df[col].min()
            c_max = df[col].max()
            
            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
            else:
                if c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
        else:
            df[col] = df[col].astype('category')
    
    end_mem = df.memory_usage().sum() / 1024**2
    print(f'Конечный размер: {end_mem:.2f} MB')
    print(f'Уменьшено на {(start_mem - end_mem) / start_mem * 100:.1f}%')
    
    return df

# Применяем
df = optimize_memory(df)
```
