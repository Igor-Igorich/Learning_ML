### 3.1. Зачем визуализировать данные

Статистика из Модуля 2 даёт точные числа, но человеческий мозг плохо умеет «видеть» таблицу из 10 000 строк. Визуализация переводит числа в картинки.

**Классический пример:** набор данных Энскомба (Anscombe's quartet). Четыре разных набора данных имеют **одинаковые** среднее, дисперсию и корреляцию. Но если нарисовать — они совершенно разные.

| Набор | Среднее X | Среднее Y | Корреляция | Что видно на графике |
|-------|-----------|-----------|------------|---------------------|
| I | 9.0 | 7.5 | 0.816 | Линейная зависимость |
| II | 9.0 | 7.5 | 0.816 | Парабола |
| III | 9.0 | 7.5 | 0.816 | Линия с одним выбросом |
| IV | 9.0 | 7.5 | 0.816 | Вертикальная линия с одним отклонением |

**Вывод:** без графиков вы можете принять параболу за прямую линию.

### 3.2. Основы matplotlib

`matplotlib` — фундаментальная библиотека для визуализации в Python. Все остальные (Seaborn, Plotly) строятся поверх неё или используют её идеи.

#### Архитектура: Figure и Axes

```
import matplotlib.pyplot as plt

# Создаём «холст» (Figure) размером 10x6 дюймов
fig = plt.figure(figsize=(10, 6))

# Добавляем «оси» (Axes) — область, где будет график
ax = fig.add_subplot(1, 1, 1)  # 1 строка, 1 столбец, 1-й график

# Рисуем линию
ax.plot([1, 2, 3, 4], [1, 4, 2, 3])

# Подписываем
ax.set_title('Мой первый график')
ax.set_xlabel('Ось X')
ax.set_ylabel('Ось Y')

plt.show()
```

**Термины:**
- **Figure** — весь холст. Может содержать несколько графиков.
- **Axes** — одна система координат (один график). На Figure может быть много Axes.
- **Axis** — ось X или Y внутри Axes.

#### Быстрый способ (pyplot)

Для простых случаев можно короче:

```
plt.figure(figsize=(10, 6))
plt.plot([1, 2, 3, 4], [1, 4, 2, 3])
plt.title('Мой первый график')
plt.xlabel('Ось X')
plt.ylabel('Ось Y')
plt.show()
```

#### Несколько графиков на одном холсте

```
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
# axes — это массив 2x2: axes[0,0], axes[0,1], axes[1,0], axes[1,1]

axes[0, 0].plot([1, 2, 3], [1, 2, 3])
axes[0, 0].set_title('Линейный')

axes[0, 1].scatter([1, 2, 3], [1, 4, 2])
axes[0, 1].set_title('Точечный')

axes[1, 0].bar(['A', 'B', 'C'], [3, 7, 2])
axes[1, 0].set_title('Столбчатый')

axes[1, 1].hist([1, 1, 2, 2, 2, 3, 3, 3, 3])
axes[1, 1].set_title('Гистограмма')

plt.tight_layout()  # Автоматически расставляет отступы
plt.show()
```

#### Основные элементы оформления

```
plt.figure(figsize=(10, 6))

plt.plot(x, y,
         color='red',        # цвет линии
         linestyle='--',     # стиль: '-', '--', '-.', ':'
         linewidth=2,        # толщина
         marker='o',         # маркеры: 'o', 's', '^', '*'
         markersize=8,
         label='Линия 1')    # для легенды

plt.title('Заголовок графика', fontsize=16, fontweight='bold')
plt.xlabel('Подпись оси X', fontsize=12)
plt.ylabel('Подпись оси Y', fontsize=12)

plt.xlim(0, 100)    # диапазон по X
plt.ylim(0, 50)     # диапазон по Y

plt.grid(True, linestyle=':', alpha=0.7)  # сетка
plt.legend(loc='upper right')              # легенда

plt.show()
```

### 3.3. Seaborn — статистическая визуализация

`Seaborn` строится поверх matplotlib и делает красивые статистические графики из коробки. Он «понимает» DataFrame pandas: вы передаёте названия столбцов, а не массивы.

```
import seaborn as sns
import matplotlib.pyplot as plt

# Стиль по умолчанию — приятнее, чем чистый matplotlib
sns.set_theme(style='whitegrid')
```

#### Распределения одной переменной

**`histplot` — гистограмма**

Показывает, сколько значений попало в каждый интервал (бин).

```
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='возраст', bins=20, kde=True, color='steelblue')
plt.title('Распределение возраста')
plt.show()
```

Параметры:
- `bins` — количество столбцов. Мало bins -> слишком грубо, много -> шум.
- `kde=True` — наложить плавную кривую плотности (см. ниже).
- `hue='пол'` — разделить по категории и показать разными цветами.

**`kdeplot` — оценка плотности (Kernel Density Estimation)**

Плавная кривая, которая показывает форму распределения без дискретных столбцов.

```
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='доход', hue='город', fill=True, alpha=0.5)
plt.title('Плотность распределения дохода по городам')
plt.show()
```

- `fill=True` — заливка под кривой.
- `alpha=0.5` — прозрачность (чтобы перекрывающиеся кривые были видны).
- `hue` — сравнение групп на одном графике.

**`ecdfplot` — эмпирическая функция распределения**

Показывает, какая доля данных меньше заданного значения.

```
sns.ecdfplot(data=df, x='возраст', hue='пол')
plt.title('Доля людей младше заданного возраста')
plt.show()
```

Интерпретация: если в точке x=30 значение y=0.4, значит 40% людей младше 30 лет.

### 3.4. Категориальные данные

**`countplot` — количество по категориям**

График частот. Показывает, сколько раз встречается каждое значение.

```
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='город', order=df['город'].value_counts().index)
plt.title('Количество наблюдений по городам')
plt.show()
```

- `order` — задаёт порядок столбцов (по убыванию частоты).

**`barplot` — среднее (или другая статистика) по категориям**

Не путать с `countplot`! `barplot` показывает **агрегированное значение** (по умолчанию среднее) с доверительным интервалом.

```
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='город', y='доход', estimator='mean', ci=95)
plt.title('Средний доход по городам')
plt.show()
```

- `estimator='mean'` — что считать (можно `np.median`, `np.sum` и др.)
- `ci=95` — доверительный интервал 95% (чёрточки на столбцах). Если он не перекрывается у двух городов — разница статистически значима.

**`boxplot` — ящик с усами**

Самый важный график для поиска выбросов и сравнения распределений.

```
        │                    <- выброс (точка за усами)
    ────┼────  верхний ус (Q3 + 1.5×IQR)
        │
   ┌────┴────┐  верхняя граница ящика (Q3)
   │    │    │  ─── медиана (линия внутри ящика)
   │    │    │
   └────┬────┘  нижняя граница ящика (Q1)
        │
    ────┼────  нижний ус (Q1 - 1.5×IQR)
```

```
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='город', y='доход')
plt.title('Распределение дохода по городам')
plt.show()
```

- Ящик содержит средние 50% данных (от Q1 до Q3).
- Линия внутри — медиана.
- Усы — границы для «нормальных» значений.
- Точки за усами — выбросы.

**`violinplot` — скрипка**

Комбинация boxplot и kdeplot. Показывает форму распределения.

```
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='город', y='доход', inner='box')
plt.title('Форма распределения дохода по городам')
plt.show()
```

- `inner='box'` — добавляет boxplot внутри скрипки.
- Ширина «скрипки» в каждой точке пропорциональна плотности данных.

**Сравнение boxplot и violinplot:**

| График | Что показывает | Когда использовать |
|--------|---------------|-------------------|
| **boxplot** | Медиана, квартили, выбросы | Быстрая проверка на выбросы, сравнение групп |
| **violinplot** | Форма всего распределения | Когда важна бимодальность или скошенность |

### 3.5. Связи между признаками

**`scatterplot` — точечная диаграмма**

Показывает связь между двумя числовыми переменными. Каждая точка — одно наблюдение.

```
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='возраст', y='доход', hue='пол', size='стаж', alpha=0.7)
plt.title('Связь возраста и дохода')
plt.show()
```

- `hue='пол'` — раскрасить точки по категории.
- `size='стаж'` — размер точки по третьей переменной.
- `alpha=0.7` — прозрачность (чтобы видеть наложение точек).

**Что искать на scatterplot:**
- **Линейная тенденция:** с ростом X растёт Y (или падает).
- **Кластеры:** группы точек, отдельные от основной массы.
- **Выбросы:** точки далеко от общего облака.
- **Нелинейность:** парабола, экспонента, логарифм.

**`lineplot` — линейный график**

Для временных рядов или когда важна последовательность.

```
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='месяц', y='продажи', hue='регион', marker='o')
plt.title('Динамика продаж по месяцам')
plt.show()
```

- Автоматически считает среднее и доверительный интервал, если в одной точке X несколько значений Y.
- Чтобы показать все линии (не усредняя), добавьте `estimator=None`.

**`pairplot` — матрица диаграмм рассеяния**

Если у вас много числовых признаков и нужно увидеть все связи сразу.

```
# Выберем только числовые столбцы
num_cols = ['возраст', 'доход', 'стаж', 'кредитный_скоринг']

sns.pairplot(df[num_cols], hue='пол', diag_kind='kde', corner=True)
plt.show()
```

- На диагонали — распределения (`kde` или `hist`).
- Вне диагонали — scatterplot для каждой пары признаков.
- `corner=True` — рисует только нижний треугольник (экономит место).

> **Внимание:** на больших датасетах (>5000 строк) `pairplot` работает медленно. Используйте выборку: `df.sample(1000)`.

### 3.6. Тепловые карты (heatmap)

**`heatmap`** — таблица чисел, раскрашенная в цвета.

**Применение 1: матрица корреляций**

```
# Считаем корреляцию только для числовых столбцов
corr = df.select_dtypes(include=['number']).corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr,
            annot=True,        # подписать числа в ячейках
            fmt='.2f',         # 2 знака после запятой
            cmap='coolwarm',   # цветовая схема
            center=0,          # центр шкалы — 0
            square=True,       # квадратные ячейки
            linewidths=0.5)    # линии между ячейками

plt.title('Матрица корреляций')
plt.show()
```

**Интерпретация цветов:**
- Тёмно-красный: сильная положительная корреляция (+1)
- Тёмно-синий: сильная отрицательная корреляция (−1)
- Белый/светлый: корреляция близка к 0

**Применение 2: таблица пропусков**

```
import missingno as msno

# Или через seaborn:
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Карта пропущенных значений')
plt.show()
```

Жёлтые/светлые полосы — пропуски. Помогает увидеть паттерны: пропуски случайны или системны?

### 3.7. Plotly — интерактивные графики

Plotly создаёт графики, с которыми можно взаимодействать: приближать, наводить курсор для подсказок, выбирать области.

```
import plotly.express as px

# Интерактивный scatterplot
fig = px.scatter(df, x='возраст', y='доход', color='город', size='стаж',
                 hover_data=['имя', 'должность'],
                 title='Интерактивная диаграмма: возраст vs доход')

fig.show()
```

**Что можно делать:**
- **Hover** — навести на точку, увидеть все данные.
- **Zoom** — выделить область лассо или прямоугольником.
- **Pan** — двигать график.
- **Toggle** — кликать на легенду, чтобы скрывать/показывать категории.

**Другие графики в Plotly:**

```
# Интерактивная гистограмма
px.histogram(df, x='доход', nbins=50, color='город', marginal='box')

# Интерактивная тепловая карта
px.density_heatmap(df, x='возраст', y='доход', nbinsx=20, nbinsy=20)

# 3D scatter (если 3 числовых признака)
px.scatter_3d(df, x='возраст', y='доход', z='стаж', color='город')
```

**Когда использовать Plotly вместо Seaborn:**

| Ситуация | Что выбрать |
|----------|-------------|
| Исследование данных для себя | Seaborn (быстрее пишется) |
| Отчёт для заказчика / дашборд | Plotly (интерактивность) |
| Много точек (>10 000) | Plotly (можно зумить) |
| Нужна 3D-визуализация | Plotly |
| Публикация в статье / презентации | Seaborn + matplotlib (векторная графика) |

### 3.8. Как выбрать правильный тип графика

| Что хотим показать | График | Библиотека |
|-------------------|--------|-----------|
| Распределение одной числовой переменной | `histplot`, `kdeplot` | Seaborn |
| Сравнение частот категорий | `countplot` | Seaborn |
| Среднее/медиана по категориям | `barplot` | Seaborn |
| Распределение по категориям + выбросы | `boxplot`, `violinplot` | Seaborn |
| Связь двух числовых переменных | `scatterplot` | Seaborn / Plotly |
| Динамика во времени | `lineplot` | Seaborn / Plotly |
| Все связи между числовыми признаками | `pairplot` | Seaborn |
| Корреляции между признаками | `heatmap` | Seaborn |
| Пропуски в данных | `heatmap(df.isnull())` | Seaborn / missingno |
| Интерактивное исследование | `scatter`, `histogram` | Plotly |

### 3.10. Частые ошибки в визуализации

#### 1. Обрезанные оси (Truncated Axes)

Если ось Y начинается не с нуля, разница выглядит катастрофической.

```
Плохо:                    Хорошо:
доход                     доход
80K |    ██               100K|        
    |    ██                   |    ██  
75K |                        80K|    ██
    |                        75K|    ██
    └──────                  0  └──────
```

```
# Плохо — ось начинается с 70000
plt.ylim(70000, 85000)

# Хорошо — ось начинается с 0 (для сравнения величин)
plt.ylim(0, 100000)
```

**Правило:** для сравнения величин (barplot) ось должна начинаться с 0. Для линейных графиков (тренды) можно обрезать, но с подписью.

#### 2. Ложная корреляция (Spurious Correlation)

На графике две переменные растут вместе, но причинно не связаны.

Классические примеры:
- Продажи мороженого и утоньшения (оба зависят от жары).
- Число пиратов и глобальное потепление.

**Как избежать:** не делайте выводов только по scatterplot. Проверяйте логику и используйте частные корреляции.

#### 3. Чартджанк (Chartjunk)

Всё лишнее, что отвлекает от данных: 3D-эффекты, тени, лишние линии сетки, яркие фоны.

```
# Плохо — лишние элементы
plt.grid(True, color='red', linewidth=2)
plt.title('!!! ДОХОДЫ !!!', fontsize=24, color='magenta')

# Хорошо — минимализм
sns.set_theme(style='white')
sns.despine()  # убрать верхнюю и правую границы
```

#### 4. Перегрузка графика

Не пытайтесь показать всё на одном графике. 10 линий разных цветов — это хаос.

**Решение:** фасетирование (разбивка на подграфики).

```
# Вместо 10 линий на одном графике:
sns.relplot(data=df, x='возраст', y='доход', kind='line',
            col='город', col_wrap=3, height=3)
# Создаст отдельный мини-график для каждого города
```

#### 5. Неправильный тип графика

- Для распределения использовать `pie chart` (круговая диаграмма) — плохо: человек плохо сравнивает углы. Лучше `barplot` или `countplot`.
- Для временных рядов использовать `barplot` — плохо: теряется последовательность. Лучше `lineplot`.

### 3.11. Практический пример: полный цикл визуализации

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Настройка стиля
sns.set_theme(style='whitegrid', palette='husl')

# Демо-данные
np.random.seed(42)
df = pd.DataFrame({
    'возраст': np.random.normal(35, 12, 1000).clip(18, 70).astype(int),
    'доход': np.random.lognormal(11, 0.6, 1000),
    'стаж': np.random.poisson(8, 1000),
    'город': np.random.choice(['Москва', 'СПб', 'Казань', 'Новосибирск'], 1000),
    'пол': np.random.choice(['М', 'Ж'], 1000)
})

# Добавим связь: доход зависит от возраста и стажа
df['доход'] = df['доход'] * (1 + (df['возраст'] - 35) * 0.02 + df['стаж'] * 0.05)

# === ПОЛНЫЙ НАБОР ГРАФИКОВ ДЛЯ EDA ===

fig = plt.figure(figsize=(16, 12))

# 1. Распределение возраста
ax1 = plt.subplot(2, 3, 1)
sns.histplot(df, x='возраст', kde=True, ax=ax1)
ax1.set_title('1. Распределение возраста')

# 2. Доход по городам (boxplot)
ax2 = plt.subplot(2, 3, 2)
sns.boxplot(df, x='город', y='доход', ax=ax2)
ax2.set_title('2. Доход по городам')
plt.setp(ax2.xaxis.get_majorticklabels(), rotation=45)

# 3. Scatter: возраст vs доход
ax3 = plt.subplot(2, 3, 3)
sns.scatterplot(df, x='возраст', y='доход', hue='пол', alpha=0.6, ax=ax3)
ax3.set_title('3. Возраст vs Доход')

# 4. Средний доход по полу и городу
ax4 = plt.subplot(2, 3, 4)
sns.barplot(df, x='город', y='доход', hue='пол', ax=ax4)
ax4.set_title('4. Средний доход: пол и город')
plt.setp(ax4.xaxis.get_majorticklabels(), rotation=45)

# 5. Корреляционная матрица
ax5 = plt.subplot(2, 3, 5)
corr = df[['возраст', 'доход', 'стаж']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, ax=ax5)
ax5.set_title('5. Корреляции')

# 6. Плотность дохода по полу
ax6 = plt.subplot(2, 3, 6)
sns.kdeplot(df, x='доход', hue='пол', fill=True, alpha=0.5, ax=ax6)
ax6.set_title('6. Плотность дохода по полу')

plt.suptitle('EDA Dashboard', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
