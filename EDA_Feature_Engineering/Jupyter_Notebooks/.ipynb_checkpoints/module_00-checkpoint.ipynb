## Подробный конспект

### 0.1. Что такое данные в контексте ML

#### Простая аналогия

Представьте, что вы — врач, и к вам приходит пациент. Вы записываете:
- возраст (35 лет)
- рост (175 см)
- вес (70 кг)
- есть ли аллергия (да/нет)
- давление (120/80)

Каждый пациент — это **одна строка** в таблице. Каждый показатель (возраст, рост, вес...) — это **столбец**. В ML эти столбцы называются **признаками** (features), а строки — **наблюдениями** (samples / observations).

#### Ключевые термины

| Термин | Что это | Пример |
|--------|---------|--------|
| **Наблюдение (sample)** | Одна строка таблицы = один объект | Один пациент, одна квартира, одна транзакция |
| **Признак (feature)** | Один столбец таблицы = одно свойство объекта | Возраст, цена, цвет, дата |
| **Целевая переменная (target)** | То, что мы хотим предсказать | Болен/здоров, цена квартиры, уйдёт ли клиент |
| **Датасет (dataset)** | Вся таблица целиком | Все пациенты со всеми их данными |

#### Типы признаков

**Числовые (Numerical)**
- *Непрерывные*: могут принимать любое значение в диапазоне. Примеры: рост, вес, температура, цена.
- *Дискретные*: только целые значения (обычно). Примеры: количество детей, этаж дома, число покупок.

**Категориальные (Categorical)**
- *Номинальные*: порядка нет, просто группы. Примеры: цвет глаз (голубые, карие, зелёные), город проживания, марка автомобиля.
- *Порядковые*: есть естественный порядок. Примеры: уровень образования (начальное -> среднее -> высшее), категория отеля (1 -> 5).

**Дата и время (Datetime)**
- Содержат календарную информацию. Примеры: дата рождения, время транзакции, дата регистрации.
- Важны для извлечения признаков: день недели, час, сезон, время с момента события.

**Текст (Text)**
- Свободные строки: отзывы, описания товаров, комментарии.
- Требуют специальной обработки (будет в Модуле 10).

**Геоданные (Geospatial)**
- Координаты (широта, долгота), адреса, города.
- Используются для создания признаков расстояния, плотности, региона.

### 0.2. Почему «мусор на входе = мусор на выходе»

#### Что это значит
Модель машинного обучения — это не волшебник. Она **учится на тех данных, которые вы ей дадите**. Если в данных ошибки, пропуски, выбросы или нерелевантная информация — модель выучит именно это.

#### Примеры проблем

| Проблема в данных | Что произойдёт с моделью |
|-------------------|--------------------------|
| В столбце «возраст» есть значение 250 лет | Модель подумает, что 250 — нормальный возраст и даст неверный прогноз |
| 30% значений «доход» пропущены | Модель не сможет использовать важный признак или выучит ложную закономерность |
| Цена в разных валютах (рубли, доллары, евро), но без указания валюты | Модель увидит «100» и «5000» и не поймёт, что 100 долларов > 5000 рублей |
| Дублированные строки | Модель переобучится на повторяющихся примерах |

#### Почему EDA и FE — это половина успеха

В реальных проектах **60–80% времени** уходит именно на работу с данными:
1. Понять, что за данные (EDA)
2. Очистить их (Data Cleaning)
3. Преобразовать в удобный вид (Feature Engineering)
4. Проверить качество

Только потом — выбор и обучение модели. Хорошие признаки + простая модель часто побеждают плохие признаки + сложную модель.

### 0.3. Жизненный цикл ML-проекта

```
┌─────────────────┐
│  1. Постановка  │  -> Что решаем? Какая бизнес-метрика?
│     задачи      │
└────────┬────────┘
         ▼
┌─────────────────┐
│  2. Сбор данных │  -> Откуда берём? Достаточно ли?
│                 │
└────────┬────────┘
         ▼
┌─────────────────────────────────────────┐
│  3. EDA (Exploratory Data Analysis)     │  <- МЫ ЗДЕСЬ
│     - Смотрим данные                    │
│     - Ищем аномалии                     │
│     - Строим графики                    │
│     - Формулируем гипотезы              │
└────────┬────────────────────────────────┘
         ▼
┌─────────────────────────────────────────┐
│  4. Data Cleaning + Feature Engineering │  <- И ЗДЕСЬ
│     - Убираем пропуски                  │
│     - Обрабатываем выбросы              │
│     - Создаём новые признаки            │
│     - Кодируем категории                │
│     - Масштабируем числа                │
└────────┬────────────────────────────────┘
         ▼
┌─────────────────┐
│  5. Моделирование│  -> Выбор алгоритма, обучение, валидация
│                 │
└────────┬────────┘
         ▼
┌─────────────────┐
│  6. Оценка      │  -> Проверяем метрики, анализируем ошибки
│                 │
└────────┬────────┘
         ▼
┌─────────────────┐
│  7. Деплой      │  -> Запускаем в продакшен, мониторим
│                 │
└─────────────────┘
```

**Где находятся EDA и FE:** на этапах 3 и 4 — между сбором данных и моделированием. Это фундамент, на котором строится всё остальное.

### 0.6. Первая практика: загрузка данных

#### Пример 1: Создаём данные вручную

```
import pandas as pd

# Создаём таблицу из словаря
data = {
    'имя': ['Анна', 'Борис', 'Виктор', 'Галина'],
    'возраст': [25, 30, 35, 28],
    'город': ['Москва', 'СПб', 'Москва', 'Казань'],
    'доход': [50000, 70000, None, 45000]  # None = пропущенное значение
}

df = pd.DataFrame(data)
print(df)
```

Вывод:

```
      имя  возраст    город   доход
0    Анна       25   Москва  50000.0
1   Борис       30      СПб  70000.0
2  Виктор       35   Москва      NaN
3  Галина       28   Казань  45000.0
```

### 0.7. Как читать документацию и искать ответы

#### Документация библиотек
- **pandas:** [pandas.pydata.org/docs](https://pandas.pydata.org/docs/)
- **numpy:** [numpy.org/doc](https://numpy.org/doc/)
- **scikit-learn:** [scikit-learn.org/stable/documentation](https://scikit-learn.org/stable/documentation.html)
- **matplotlib:** [matplotlib.org/stable](https://matplotlib.org/stable/)

**Как пользоваться:** в поиске документации введите название функции, например, `pandas read_csv`. Откроется страница со всеми параметрами и примерами.

#### Где искать ответы на ошибки

| Проблема | Куда идти |
|----------|-----------|
| Не понятно, как работает функция | Документация + примеры на странице функции |
| Ошибка в коде | Скопируйте текст ошибки в Google, добавьте `python` или `pandas` |
| Не понятно, какой график выбрать | Галерея seaborn / matplotlib с примерами |
| Сложная задача, нужен пример | Stack Overflow (stackoverflow.com) — поиск по тегу `[pandas]` |
| Хочу разобраться в теории | Книги: «Hands-On Machine Learning» (A. Géron), «Python for Data Analysis» (W. McKinney) |

#### Как читать сообщения об ошибках
Python сообщает об ошибках сверху вниз. **Смотрите на последнюю строку** — там написано, что пошло не так.

Пример:

```
df = pd.read_csv('nesuschestvuyuschiy_fayl.csv')
```

Ошибка:

```
FileNotFoundError: [Errno 2] No such file or directory: 'nesuschestvuyuschiy_fayl.csv'
```

**Что делать:** проверьте, что файл лежит в той же папке, что и ноутбук, или укажите полный путь.
