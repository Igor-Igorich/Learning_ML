## 2.1 Разреженные матрицы

### 2.1.1 Почему нельзя просто хранить обычный (dense) массив

Вернёмся к примеру из Модуля 1.4.2: 100 000 пользователей × 10 000 товаров, из них заполнено 10 миллионов ячеек (sparsity 99%).

**Расчёт памяти для плотного (dense) хранения:**

```python
100 000 × 10 000 ячеек × 8 байт (float64) = 8 000 000 000 байт ≈ 8 ГБ
```

Это 8 ГБ **только на одну матрицу взаимодействий**, из которых 99% байт — это хранение нулей, не несущих никакой информации. При работе с реальными промышленными масштабами (десятки миллионов пользователей, миллионы товаров) такой подход становится физически невозможным — ни одна оперативная память не выдержит.

**Решение:** хранить только координаты и значения **ненулевых** ячеек. Это и есть идея разреженных (sparse) матриц.

### 2.1.2 Формат COO (Coordinate List) — самый простой для понимания

**Идея:** хранить три параллельных массива одинаковой длины — `row` (индексы строк), `col` (индексы столбцов), `data` (значения) — только для ненулевых элементов.

**Пример.** Возьмём небольшую матрицу 4 пользователя × 5 товаров:

```python
       I1   I2   I3   I4   I5
U1  [   0    5    0    3    0  ]
U2  [   4    0    0    0    2  ]
U3  [   0    0    0    0    0  ]
U4  [   0    1    4    0    0  ]
```

Здесь только 6 ненулевых значений. COO-представление:

```python
row  = [0, 0, 1, 1, 3, 3]
col  = [1, 3, 0, 4, 1, 2]
data = [5, 3, 4, 2, 1, 4]
```

Читается построчно: элемент `row[0]=0, col[0]=1, data[0]=5` означает «в строке 0 (U1), столбце 1 (I2) стоит значение 5» — что соответствует `U1, I2 = 5` в исходной матрице.

**Плюс COO:** предельно просто конструировать (например, при построчном чтении лога взаимодействий из базы данных — как раз ваш сценарий: читаете таблицу `interactions` и сразу добавляете тройки `(user_id, item_id, rating)`).
**Минус COO:** неэффективен для арифметических операций и произвольного доступа — чтобы найти все элементы конкретной строки, нужно перебрать весь массив `row` целиком (`O(nnz)`, где `nnz` — число ненулевых элементов).

### 2.1.3 Формат CSR (Compressed Sparse Row) — рабочая лошадка для user-based операций

**Идея:** сжать массив `row` из COO в **указатели на начало каждой строки**, вместо того чтобы явно повторять номер строки для каждого элемента.

Три массива:
- `data` — значения (те же, что в COO, но упорядоченные по строкам).
- `indices` — индексы столбцов (аналог `col` из COO).
- `indptr` (index pointer) — массив длиной `n_строк + 1`, где `indptr[i]` и `indptr[i+1]` задают диапазон в `data`/`indices`, соответствующий строке `i`.

**Построим CSR для той же матрицы:**

```python
data    = [5, 3, 4, 2, 1, 4]
indices = [1, 3, 0, 4, 1, 2]
indptr  = [0, 2, 4, 4, 6]
```

**Как это читать (пошагово):**
- `indptr[0]=0, indptr[1]=2` -> строка U1 занимает `data[0:2] = [5, 3]` с `indices[0:2] = [1, 3]` -> U1: I2=5, I4=3.  совпадает с исходной матрицей.
- `indptr[1]=2, indptr[2]=4` -> строка U2 занимает `data[2:4] = [4, 2]` с `indices[2:4] = [0, 4]` -> U2: I1=4, I5=2. 
- `indptr[2]=4, indptr[3]=4` -> строка U3 занимает `data[4:4]` — **пустой диапазон**, потому что `indptr[2] == indptr[3] == 4`. Это корректно отражает, что U3 не оценил ни одного товара.
- `indptr[3]=4, indptr[4]=6` -> строка U4 занимает `data[4:6] = [1, 4]` с `indices[4:6] = [1, 2]` -> U4: I2=1, I3=4.

**Почему это быстро:** чтобы получить все оценки конкретного пользователя (то есть всю строку), не нужно сканировать весь массив — достаточно двух чисел `indptr[i]` и `indptr[i+1]`, чтобы сразу «вырезать» нужный диапазон. Это ключевое свойство, которое будет активно использоваться в User-Based CF (Модуль 3): для целевого пользователя нужно быстро достать список его оценённых товаров.

### 2.1.4 Формат CSC (Compressed Sparse Column) — зеркальный аналог для item-based операций

Та же идея, что CSR, но сжимаются **столбцы**, а не строки: `indptr` теперь задаёт диапазоны по товарам, а не по пользователям.

**Когда нужен именно CSC:** в Item-Based CF (Модуль 4) типичная операция — «получить всех пользователей, оценивших конкретный товар» (то есть весь столбец). Если бы матрица хранилась только в CSR, эта операция требовала бы прохода по всем строкам — `O(n_users)`. В CSC та же операция выполняется за время, пропорциональное только числу пользователей, реально оценивших этот товар.

**Практическое правило:** CSR — когда основная операция «дай мне всю строку» (по пользователю). CSC — когда основная операция «дай мне весь столбец» (по товару). Выбор формата — это не техническая деталь для галочки, а прямая оптимизация под то, какой алгоритм вы реализуете.

### 2.1.5 Сравнение затрат памяти: dense vs sparse (CSR) на реалистичном примере

Продолжим пример из 2.1.1: 100 000 × 10 000, 10 миллионов ненулевых элементов.

**Dense:** `100 000 × 10 000 × 8 байт = 8 000 000 000 байт ≈ 7.45 ГБ`

**CSR:**

```python
data:    10 000 000 элементов × 8 байт  = 80 000 000 байт
indices: 10 000 000 элементов × 4 байта = 40 000 000 байт
indptr:  100 001 элемент × 4 байта      =     400 004 байта
─────────────────────────────────────────────────────────
Итого:                                  ≈ 120 400 004 байта ≈ 114.8 МБ
```

**Итог:** разреженное хранение в этом примере занимает **~115 МБ вместо ~7.45 ГБ** — сжатие примерно в **65 раз**. Это не абстрактная оптимизация «на будущее» — без такого сжатия работа с реальными данными RecSys в оперативной памяти была бы попросту невозможна на обычном оборудовании.

### 2.1.6 Практический код: работа со scipy.sparse

```python
import numpy as np
import scipy.sparse as sp

# Данные из таблицы interactions (Модуль 13.3): user_id, item_id, rating
row = np.array([0, 0, 1, 1, 3, 3])
col = np.array([1, 3, 0, 4, 1, 2])
data = np.array([5, 3, 4, 2, 1, 4])

# Построение в формате COO (естественный формат для "сырых" данных)
coo_matrix = sp.coo_matrix((data, (row, col)), shape=(4, 5))

# Конвертация в CSR для быстрого построчного доступа (User-Based CF)
csr_matrix = coo_matrix.tocsr()

# Конвертация в CSC для быстрого постолбцового доступа (Item-Based CF)
csc_matrix = coo_matrix.tocsc()

# Получить все оценки пользователя U2 (индекс 1) — быстро благодаря CSR
user_2_ratings = csr_matrix.getrow(1)
print(user_2_ratings.toarray())  # [[4. 0. 0. 0. 2.]]

# Получить всех пользователей, оценивших товар I2 (индекс 1) — быстро благодаря CSC
item_2_ratings = csc_matrix.getcol(1)
print(item_2_ratings.toarray())  # [[5.] [0.] [0.] [1.]]

# Реальный объём занимаемой памяти
print(f"Dense-эквивалент: {coo_matrix.toarray().nbytes} байт")
print(f"CSR: {csr_matrix.data.nbytes + csr_matrix.indices.nbytes + csr_matrix.indptr.nbytes} байт")
```

## 2.2 Меры сходства (Similarity Metrics)

### 2.2.0 Сквозной пример для всех метрик

Чтобы напрямую сравнить, как разные метрики оценивают одну и ту же пару пользователей, используем один пример на протяжении всего раздела:

- **Пользователь A** (широко использует шкалу оценок): смотрел 3 фильма, оценки — `[5, 3, 1]`.
- **Пользователь E** (использует только верхнюю часть шкалы, но с тем же относительным порядком предпочтений): те же 3 фильма, оценки — `[5, 4, 3]`.

Обратите внимание: оба поставили **фильму 1 — максимум**, **фильму 2 — среднее**, **фильму 3 — минимум из своей шкалы**. То есть их относительные предпочтения полностью совпадают, но абсолютные значения и «щедрость» шкалы — разные. Это специально сконструированный пример, чтобы показать, как по-разному метрики реагируют на это различие.

### 2.2.1 Косинусное сходство (Cosine Similarity)

**Формула:**

```python
cos(A, B) = (A · B) / (||A|| × ||B||)
```

где `A · B` — скалярное произведение, `||A||` — евклидова норма (длина) вектора.

**Расчёт на нашем примере:**

```python
A · E = 5×5 + 3×4 + 1×3 = 25 + 12 + 3 = 40
||A|| = √(5² + 3² + 1²) = √35 ≈ 5.916
||E|| = √(5² + 4² + 3²) = √50 ≈ 7.071

cos(A, E) = 40 / (5.916 × 7.071) = 40 / 41.833 ≈ 0.956
```

**Интерпретация:** значение от -1 до 1 (для рейтингов, которые всегда положительны, обычно от 0 до 1). 1 означает «векторы направлены абсолютно одинаково» (пропорциональны друг другу), 0 — «полностью независимы» (ортогональны), отрицательные значения — «противоположные предпочтения» (возможны только если допустить отрицательные значения, например, после центрирования — см. 2.3).

**Ключевое свойство:** косинус инвариантен к **масштабированию** вектора (если умножить все оценки E на константу, косинус не изменится), но **не инвариантен к сдвигу** (если прибавить константу ко всем оценкам). Именно поэтому в нашем примере, где E = A + сдвиг с сжатием диапазона, косинус получился высоким (0.956), но не идеальным — часть рассогласования вызвана именно тем, что косинус не полностью «прощает» разницу в стиле выставления оценок.

**Когда используется:** база для Item-Based CF (Модуль 4) — сравнение векторов-столбцов (оценок разных пользователей по конкретному товару).

**Код:**

```python
import numpy as np

def cosine_similarity(a: np.ndarray, b: np.ndarray) -> float:
    dot = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 0.0  # защита от деления на ноль (пользователь без оценок)
    return dot / (norm_a * norm_b)

A = np.array([5, 3, 1])
E = np.array([5, 4, 3])
print(cosine_similarity(A, E))  # 0.9558...
```

### 2.2.2 Корреляция Пирсона (Pearson Correlation)

**Формула:** косинусное сходство, применённое к **центрированным** (mean-centered) векторам — то есть после вычитания среднего значения каждого вектора из самого себя (подробный механизм центрирования — раздел 2.3.1):

```python
Pearson(A, B) = Σ((A_i - Ā)(B_i - B̄)) / (√Σ(A_i - Ā)² × √Σ(B_i - B̄)²)
```

**Расчёт на нашем примере:**

```python
Ā = (5+3+1)/3 = 3        ->  A - Ā = [2, 0, -2]
Ē = (5+4+3)/3 = 4        ->  E - Ē = [1, 0, -1]

числитель = 2×1 + 0×0 + (-2)×(-1) = 2 + 0 + 2 = 4
||A-Ā|| = √(4+0+4) = √8 ≈ 2.828
||E-Ē|| = √(1+0+1) = √2 ≈ 1.414

Pearson(A, E) = 4 / (2.828 × 1.414) = 4 / 4.0 = 1.0
```

**Это принципиальный результат.** После центрирования по среднему значению каждого пользователя, корреляция Пирсона показала **идеальное совпадение (1.0)** — потому что относительный паттерн предпочтений A и E абсолютно одинаков (оба ставят: максимум -> среднее -> минимум, с одинаковым шагом относительно своего среднего). Косинус же (2.2.1) дал 0.956 — «почти, но не идеально», потому что не учёл разницу в «щедрости» шкалы каждого пользователя.

**Вывод, критичный для User-Based CF (Модуль 3):** Пирсон **специально спроектирован** для того, чтобы отделять «различия во вкусах» от «различий в стиле выставления оценок» (один пользователь щедрый и ставит всем высокие баллы, другой строгий и ставит низкие — но относительные предпочтения могут полностью совпадать). Это и есть причина, по которой User-Based CF почти всегда использует именно Пирсона, а не «сырой» косинус.

**Ограничение:** корреляция Пирсона нестабильна, когда у двух пользователей мало общих оценённых товаров (например, всего 1-2 совпадения) — на таком малом числе точек коэффициент корреляции может принимать экстремальные значения (близкие к ±1) просто случайно, не отражая реальной закономерности. На практике вводят порог минимального числа общих оценок, ниже которого сходство не учитывается или искусственно занижается (шринкидж/shrinkage).

**Код:**

```python
def pearson_similarity(a: np.ndarray, b: np.ndarray) -> float:
    a_centered = a - a.mean()
    b_centered = b - b.mean()
    return cosine_similarity(a_centered, b_centered)

print(pearson_similarity(A, E))  # 1.0
```

### 2.2.3 Евклидово расстояние (Euclidean Distance)

**Формула:**

```python
dist(A, B) = √Σ(A_i - B_i)²
```

Это расстояние, а не сходство — чем **меньше** значение, тем ближе объекты. Чтобы превратить его в сходство (0 = далеко, 1 = идентично), обычно применяют преобразование:

```python
similarity = 1 / (1 + dist(A, B))
```

**Расчёт на нашем примере:**

```python
dist(A, E) = √((5-5)² + (3-4)² + (1-3)²) = √(0 + 1 + 4) = √5 ≈ 2.236
similarity = 1 / (1 + 2.236) = 1 / 3.236 ≈ 0.309
```

**Важнейшее наблюдение:** несмотря на то, что A и E имеют **абсолютно идентичный относительный паттерн предпочтений** (мы это уже подтвердили Пирсоном = 1.0), евклидово расстояние даёт низкую оценку сходства (0.309) — потому что оно чувствительно к **абсолютной** разнице значений, а не к паттерну. Это ключевое ограничение метрики: она путает «разные вкусы» с «разный стиль оценивания».

**Проклятие размерности (curse of dimensionality) — почему Евклидово расстояние плохо работает в RecSys напрямую:**
При росте числа измерений (в user-item матрице это тысячи и миллионы товаров) все точки в пространстве постепенно становятся «примерно одинаково далеко» друг от друга — разница между расстоянием до ближайшего и до самого дальнего соседа стремится к нулю в относительном выражении. Интуитивно: в 2D-пространстве легко увидеть, кто «явно ближе» — в пространстве из 10 000 измерений (10 000 товаров) почти все векторы почти ортогональны друг другу просто в силу геометрии высокой размерности, независимо от реального сходства вкусов. Это одна из причин, по которой Евклидово расстояние на практике используется в RecSys реже, чем косинус или Пирсон, а также одна из мотиваций перехода к компактным эмбеддингам (Модуль 5, 10) — там размерность искусственно снижается до 50-300, где Евклидово расстояние снова становится осмысленным.

**Код:**

```python
def euclidean_similarity(a: np.ndarray, b: np.ndarray) -> float:
    distance = np.linalg.norm(a - b)
    return 1 / (1 + distance)

print(euclidean_similarity(A, E))  # 0.3090...
```

### 2.2.4 Сходство Жаккара (Jaccard Similarity)

**Формула:**

```python
Jaccard(A, B) = |A ∩ B| / |A ∪ B|
```

Работает не с числовыми векторами оценок, а с **множествами** — то есть с бинарными данными (взаимодействовал / не взаимодействовал). Это прямая связь с Модулем 1.3.2 (implicit feedback): когда у нас нет рейтинга, а есть только факт «купил/посмотрел», Жаккар — естественный выбор.

**Расчёт на новом примере (специально для бинарного случая):**
- Пользователь A посмотрел фильмы: `{1, 2, 3, 4, 5}`
- Пользователь B посмотрел фильмы: `{2, 3, 4, 6, 7}`

```python
Пересечение: {2, 3, 4}          -> |A ∩ B| = 3
Объединение: {1, 2, 3, 4, 5, 6, 7} -> |A ∪ B| = 7

Jaccard(A, B) = 3 / 7 ≈ 0.4286
```

**Интерпретация:** 42.86% от всех фильмов, которые смотрел хотя бы один из двух пользователей, — общие для обоих. Заметьте: Жаккар **полностью игнорирует силу взаимодействия** (сколько раз посмотрел, какой рейтинг поставил) — учитывается только сам факт. Это одновременно и простота (не нужно нормализовывать шкалы, п. 2.3), и ограничение (теряется информация о степени предпочтения).

**Код:**

```python
def jaccard_similarity(set_a: set, set_b: set) -> float:
    intersection = len(set_a & set_b)
    union = len(set_a | set_b)
    if union == 0:
        return 0.0
    return intersection / union

watched_A = {1, 2, 3, 4, 5}
watched_B = {2, 3, 4, 6, 7}
print(jaccard_similarity(watched_A, watched_B))  # 0.4286
```

### 2.2.5 Сводная таблица: когда какую метрику использовать

| Метрика | Тип данных | Учитывает силу сигнала | Учитывает «щедрость» пользователя | Типичное применение |
|:---|:---|:---:|:---:|:---|
| Cosine | Числовые рейтинги | Да | Нет | Item-Based CF (Модуль 4) |
| Pearson | Числовые рейтинги | Да | Да (за счёт центрирования) | User-Based CF (Модуль 3) |
| Euclidean | Числовые рейтинги | Да | Нет | Редко в чистом виде; чаще в компактных эмбеддингах (Модули 5, 10) |
| Jaccard | Бинарные множества | Нет (только факт) | Не применимо | Implicit feedback без явных оценок |

### 2.2.6 Итоговый результат сравнения на сквозном примере

| Метрика | Значение для (A, E) | Что это означает |
|:---|:---:|:---|
| Cosine | 0.956 | Высокое, но не идеальное сходство — не до конца «прощает» разницу шкал |
| Pearson | 1.000 | Идеальное сходство — относительные предпочтения полностью совпадают |
| Euclidean similarity | 0.309 | Низкое сходство — метрика «путает» разный стиль оценивания с разными вкусами |

Этот результат — не случайность конкретного примера, а системное свойство метрик, которое нужно прочно усвоить перед Модулем 3.

## 2.3 Нормализация рейтингов

### 2.3.1 Mean-centering (центрирование по среднему)

**Проблема, которую решает:** два пользователя с идентичными вкусами могут выглядеть «непохожими» просто из-за разной щедрости в выставлении оценок — щедрый пользователь ставит 4-5 всему, что нейтрально или чуть выше среднего, строгий пользователь ставит 2-3 тому же самому. Без нормализации это различие в стиле маскирует реальное совпадение вкусов.

**Формула:**

```python
r'_ui = r_ui - r̄_u
```

где `r̄_u` — средний рейтинг, поставленный пользователем `u` по всем товарам, которые он оценил.

Мы уже применили это на практике в разделе 2.2.2 при выводе формулы Пирсона — она и есть по сути «косинус после mean-centering».

**Важный нюанс — по какой оси центрировать:**
- Центрирование **по пользователю** (вычитаем средний рейтинг каждого пользователя) — используется, когда сравниваем пользователей между собой (User-Based CF).
- Центрирование **по товару** (вычитаем средний рейтинг, полученный каждым товаром) — применяется в другом сценарии: когда некоторые товары систематически получают завышенные/заниженные оценки независимо от пользователя (например, культовый фильм, который принято хвалить).
- В Item-Based CF (Модуль 4) часто используется **adjusted cosine similarity** — косинус между столбцами (товарами), но с вычитанием среднего рейтинга **пользователя** (а не товара) из каждой оценки перед расчётом. Это тонкий, но важный момент: даже сравнивая товары, мы всё равно устраняем именно пользовательское смещение, потому что источник «шума щедрости» — это стиль конкретного человека, а не свойство конкретного товара.

### 2.3.2 Z-score нормализация

**Проблема, которую дополнительно решает (сверх mean-centering):** mean-centering устраняет разницу в **среднем уровне** оценок, но не устраняет разницу в **разбросе** (variance). Представим пользователя, который ставит только крайние оценки (1 или 5, никогда середину) — «поляризованный» тип — и пользователя, который ставит более сдержанные оценки в диапазоне 2-4. После простого центрирования оба всё ещё будут иметь разный масштаб отклонений от своего среднего, и это может исказить сравнение.

**Формула:**

```python
z_ui = (r_ui - r̄_u) / σ_u
```

где `σ_u` — стандартное отклонение оценок пользователя `u`.

**Расчёт на нашем сквозном примере (A и E):**

```python
A = [5, 3, 1], Ā = 3
σ_A = √(((5-3)² + (3-3)² + (1-3)²) / 3) = √((4+0+4)/3) = √2.667 ≈ 1.633
Z_A = [(5-3)/1.633, (3-3)/1.633, (1-3)/1.633] = [1.225, 0, -1.225]

E = [5, 4, 3], Ē = 4
σ_E = √(((5-4)² + (4-4)² + (3-4)²) / 3) = √((1+0+1)/3) = √0.667 ≈ 0.816
Z_E = [(5-4)/0.816, (4-4)/0.816, (3-4)/0.816] = [1.225, 0, -1.225]
```

**Наблюдение:** после Z-score нормализации векторы A и E стали **абсолютно идентичными** — `[1.225, 0, -1.225]`. Это иллюстрирует важный математический факт: **косинусное сходство Z-score-нормализованных векторов математически эквивалентно корреляции Пирсона** (при использовании одной и той же — популяционной — формулы стандартного отклонения для обоих векторов). Именно поэтому Пирсон уже дал 1.0 в разделе 2.2.2 — Z-score-нормализация не открывает здесь принципиально новой информации *в этом конкретном примере*, потому что у A и E совпадает не только направление отклонений, но и относительный масштаб разброса.

**Тогда зачем вообще нужен Z-score отдельно от Пирсона, если для расчёта сходства они по сути эквивалентны?**
1. Z-score применяется не только на этапе расчёта сходства между пользователями, но и как **самостоятельный шаг предобработки самих рейтингов** перед их использованием в других частях пайплайна — например, перед матричной факторизацией (Модуль 5), где важно, чтобы входные значения были в сопоставимом масштабе для всех пользователей.
2. Z-score отдельно учитывает случай, когда у двух пользователей одинаковое **направление** предпочтений, но разный **разброс** (интенсивность) — в примерах, отличных от нашего специально сконструированного (где разброс тоже совпал пропорционально), Пирсон и Z-score-based сходство могут немного расходиться в деталях реализации (например, если использовать выборочное, а не популяционное стандартное отклонение — тогда появляется поправочный коэффициент `n/(n-1)`, который не влияет на итоговое косинусное сходство, но влияет на абсолютные значения нормализованных рейтингов, используемых дальше в предсказании).

### 2.3.3 Когда нормализовать, а когда — нет: практическое правило

| Сценарий | Нормализация |
|:---|:---|
| User-Based CF (Модуль 3), сравнение пользователей | Mean-centering или полный Z-score (то есть фактически использование Пирсона вместо косинуса) |
| Item-Based CF (Модуль 4), сравнение товаров | Adjusted cosine (центрирование по пользователю, но сравнение по столбцам) |
| Implicit feedback (Жаккар, бинарные данные) | Нормализация не применяется — данные уже бинарны по своей природе |
| Matrix Factorization (Модуль 5) | Часто используют собственные bias-термы (`b_u`, `b_i`) внутри самой модели вместо ручной предварительной нормализации — модель обучается компенсировать смещения автоматически |

## 2.4 Практика

### 2.4.1 Реализация всех метрик векторизованно (без циклов) на NumPy

Циклы по элементам вектора медленны и не соответствуют вашему уже отработанному навыку векторизации (Pandas/NumPy, общий план, Недели 1-2). Задача — переписать функции из разделов 2.2 так, чтобы они считали сходство сразу для **матрицы** пользователей, а не по одной паре за раз.

```python
import numpy as np

def cosine_similarity_matrix(matrix: np.ndarray) -> np.ndarray:
    """Векторизованный расчёт косинусного сходства между всеми парами строк матрицы."""
    norms = np.linalg.norm(matrix, axis=1, keepdims=True)
    norms[norms == 0] = 1e-10  # защита от деления на ноль
    normalized = matrix / norms
    return normalized @ normalized.T

def pearson_similarity_matrix(matrix: np.ndarray) -> np.ndarray:
    """Корреляция Пирсона между всеми парами строк = косинус после центрирования по строкам."""
    row_means = matrix.mean(axis=1, keepdims=True)
    centered = matrix - row_means
    return cosine_similarity_matrix(centered)

def euclidean_similarity_matrix(matrix: np.ndarray) -> np.ndarray:
    """1 / (1 + евклидово расстояние) между всеми парами строк."""
    # ||a-b||^2 = ||a||^2 + ||b||^2 - 2*a·b — классический трюк для векторизации
    sq_norms = np.sum(matrix**2, axis=1)
    dist_sq = sq_norms[:, np.newaxis] + sq_norms[np.newaxis, :] - 2 * (matrix @ matrix.T)
    dist_sq = np.maximum(dist_sq, 0)  # защита от отрицательных чисел из-за погрешности вычислений (позлементное сравнение элемента с нулём)
    dist = np.sqrt(dist_sq)
    return 1 / (1 + dist)
```

### 2.4.2 Поиск 5 ближайших соседей по косинусу и по Пирсону — сравнение результатов

```python
import pandas as pd

# Предполагается, что user_item_matrix уже построена (Модуль 1.6, но полная, не разреженная,
# для учебных целей — на подвыборке из нескольких сотен активных пользователей)
target_user_idx = 0

cos_sim = cosine_similarity_matrix(user_item_matrix.values)
pearson_sim = pearson_similarity_matrix(user_item_matrix.values)

top5_cosine = np.argsort(-cos_sim[target_user_idx])[1:6]     # [1:6], т.к. индекс 0 - сам пользователь
top5_pearson = np.argsort(-pearson_sim[target_user_idx])[1:6]

print("Топ-5 по косинусу:", user_item_matrix.index[top5_cosine].tolist())
print("Топ-5 по Пирсону:", user_item_matrix.index[top5_pearson].tolist())

# Ключевой вопрос для анализа: пересекаются ли эти два списка?
overlap = set(top5_cosine) & set(top5_pearson)
print(f"Пересечение: {len(overlap)} из 5")
```

**На что обратить внимание в результате:** если списки соседей заметно отличаются — это прямое практическое подтверждение того, что показал наш аналитический пример в 2.2.6: косинус и Пирсон систематически по-разному оценивают сходство, когда в данных присутствуют пользователи с разной «щедростью» шкалы (а в реальных данных они присутствуют почти всегда).

| Pearson Correlation | Косинус после центрирования по среднему; устраняет «щедрость» пользователя |
| Euclidean Distance | Геометрическое расстояние; страдает от проклятия размерности |
| Jaccard Similarity | Отношение пересечения к объединению множеств; для бинарных данных |
| Mean-centering | Вычитание среднего значения пользователя/товара из его оценок |
| Z-score | Центрирование, дополнительно делённое на стандартное отклонение |
| Adjusted Cosine | Косинус между товарами с центрированием оценок по пользователю |
| Curse of Dimensionality | Эффект, при котором расстояния теряют информативность в высоких размерностях |
