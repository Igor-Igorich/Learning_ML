**Модуль 2. Docker с нуля: контейнер как единица доставки**

### 2.1. Зачем мы здесь: от первого контейнера к пониманию

В прошлом модуле вы запустили `docker run hello-world` и увидели приветственное сообщение. Теперь нам нужно разобраться, *что именно* произошло, и научиться управлять контейнерами уверенно — как водителю, который не просто завёл машину, а понимает, как работают педали, зеркала и передачи.

К концу этого модуля вы будете:
- Чётко различать **образ**, **контейнер** и **реестр**.
- Запускать, останавливать и подключаться к контейнерам.
- Понимать, почему программа внутри контейнера «не видит» ваш компьютер.
- Запускать Python внутри Docker и выполнять свои скрипты.

### 2.2. Главная аналогия: контейнер — это герметичный ящик

Представьте, что вам нужно отправить хрустальную вазу из Москвы во Владивосток. Вы кладёте вазу в **жёсткий пластиковый ящик** с пенопластом внутри. Ящик герметичен: внутри него свой микроклимат — влажность, температура, давление. Внешние толчки и перепады температур не достают до вазы, потому что ящик изолирован от мира.

**Docker-контейнер** — это такой же ящик, только для программ. Внутри него живут:
- Ваш код.
- Python (или другой язык).
- Все библиотеки (`numpy`, `pandas`, `scikit-learn`).
- Даже часть операционной системы — всё, что нужно программе для работы.

Программа внутри ящика **не знает**, какая операционная система на вашем компьютере. Она видит только то, что вы ей положили в ящик. Это и есть **изоляция**.

### 2.3. Три ключевых понятия: Образ, Контейнер, Реестр

Это самое важное различие в Docker. Путаница здесь — главная причина, почему новички застревают.

#### 2.3.1. Образ (Image) — чертёж ящика

**Образ** — это неизменяемый шаблон, инструкция для создания контейнера. Он хранится как файл (точнее, набор слоёв) на диске.

**Аналогия:** Образ — это **рецепт торта** или **3D-модель детали для 3D-принтера**. По одному рецепту можно испечь десять тортов. По одной 3D-модели можно напечатать десять деталей. Каждый торт/деталь — это отдельный экземпляр, но все они созданы по одному шаблону.

**Свойства образа:**
- **Неизменяем (read-only).** Вы не можете «отредактировать» образ напрямую. Если хотите изменить — создаёте новый образ.
- **Многослойный.** Образ состоит из слоёв (подробно в Модуле 3).
- **Имеет имя и тег.** Например, `python:3.11-slim`. Тег `3.11-slim` уточняет версию. Если тег не указан, Docker использует `latest`.

#### 2.3.2. Контейнер (Container) — запущенный ящик

**Контейнер** — это **работающий экземпляр** образа. Это процесс (или группа процессов), который живёт в изолированной среде, созданной на основе образа.

**Аналогия:** Если образ — это рецепт торта, то контейнер — это **готовый торт на столе**. Вы можете съесть его (выполнить задачу), а затем выбросить тарелку. Рецепт при этом останется нетронутым.

**Свойства контейнера:**
- **Изменяем.** Внутри контейнера программа может создавать файлы, писать в базу данных, менять настройки. Но когда контейнер удаляется — все эти изменения исчезают (если не используются volumes, см. Модуль 5).
- **Временный по умолчанию.** Контейнер создаётся, работает, завершается. Это нормально.
- **Уникальный ID.** У каждого контейнера есть свой идентификатор, например `a1b2c3d4e5f6`.

#### 2.3.3. Реестр (Registry) — магазин чертежей

**Реестр** — это удалённое хранилище образов. Самый известный — **Docker Hub** (hub.docker.com). Туда разработчики загружают образы, чтобы другие могли их скачать.

**Аналогия:** Реестр — это **App Store или Google Play**, но не для приложений, а для образов. Вы заходите, ищете «Python», видите официальный образ от разработчиков Python, нажимаете «скачать» (`docker pull`).

**Ключевые моменты:**
- Образ можно скачать из реестра (`docker pull`).
- Образ можно загрузить в реестр (`docker push`) — например, чтобы коллега или сервер в облаке его использовал.
- Реестр может быть публичным (Docker Hub) или приватным (внутри компании).

### 2.4. Изоляция: почему программа внутри «не видит» ваш компьютер

Это самая мощная идея Docker. Чтобы понять её, не нужно быть системным программистом. Достаточно представить квартиру в многоэтажном доме.

#### 2.4.1. Аналогия: квартира в многоквартирном доме

Представьте, что ваш компьютер — это многоквартирный дом. Каждый контейнер — это отдельная квартира в этом доме.

- У квартиры **свои стены** (изоляция процессов): жилец квартиры А не может заглянуть в окно квартиры Б.
- У квартиры **своя ванная и кухня** (изоляция файловой системы): жилец А не видит посуду жильца Б.
- У квартиры **свой почтовый ящик** (изоляция сети): письмо, адресованное квартире А, не попадёт к квартире Б, если не переслать специально.
- Но все квартиры стоят **на одном фундаменте** (общее ядро ОС). Это отличает контейнер от виртуальной машины, где под каждой квартирой — отдельный фундамент с отдельной землёй.

#### 2.4.2. Три уровня изоляции

**1. Изоляция процессов**

На вашем компьютере работают десятки программ: браузер, музыкальный плеер, терминал. Операционная система ведёт список всех процессов. Если вы откроете Диспетчер задач (Windows) или выполните `ps aux` (Linux/macOS), увидите все процессы.

Внутри контейнера программа видит **только свои процессы**. Она не знает, что у вас открыт браузер. Более того, процесс внутри контейнера думает, что он работает в своей собственной операционной системе — у него свой PID 1 (главный процесс), свой список процессов.

**2. Изоляция файловой системы**

Внутри контейнера своя корневая папка `/`. Она не совпадает с корневой папкой вашего компьютера. Если в контейнере вы создадите файл `/tmp/myfile.txt`, он появится только внутри контейнера. На вашем хосте в папке `/tmp` этого файла не будет.

Это как в квартире: у вас есть шкаф `/tmp`, но он находится внутри вашей квартиры, а не в подъезде дома.

**3. Изоляция сети**

По умолчанию контейнер получает свой собственный IP-адрес (внутренний, вида `172.17.0.2`). Он может выходить в интернет (через хост), но извне, с вашего компьютера, вы не можете просто так подключиться к программе внутри контейнера. Нужно специально «пробросить» порт — мы разберём это ниже.

### 2.5. Жизненный цикл контейнера

Контейнер, как живое существо, проходит через состояния:

1. **Created** — создан, но не запущен. Редко используется вручную.
2. **Running** — работает. Внутри выполняется главный процесс.
3. **Paused** — приостановлен (заморожен). Процессы не выполняются, но не убиты.
4. **Stopped** — остановлен. Главный процесс завершился (успешно или с ошибкой). Контейнер существует на диске, но не работает.
5. **Deleted** — удалён. Все данные контейнера стёрты.

**Важно:** Контейнер живёт, пока живёт его **главный процесс**. Если главный процесс завершился — контейнер остановился. Например, если вы запустите контейнер с Python-скриптом, который печатает «Hello» и завершается — контейнер тоже завершится. Это не баг, это фича.

### 2.6. Команды Docker: полный разбор

Теперь разберём команды, которые вы будете использовать каждый день. Мы пойдём от простого к сложному.

#### 2.6.1. `docker run` — создать и запустить контейнер

Это самая частая команда. Она делает две вещи: **создаёт** контейнер из образа и **запускает** его.

```
docker run имя_образа
```

**Что происходит под капотом:**
1. Docker проверяет, есть ли образ локально.
2. Если нет — скачивает из реестра (Docker Hub).
3. Создаёт новый контейнер из образа.
4. Запускает главный процесс контейнера.
5. Подключается к выводу процесса (по умолчанию).

**Флаги, которые нужно знать:**

| Флаг | Значение | Пример |
|------|----------|--------|
| `-it` | Интерактивный режим + псевдотерминал. Позволяет вводить команды внутри контейнера. | `docker run -it python:3.11 bash` |
| `-d` | Detached (фоновый режим). Контейнер работает в фоне, не блокируя терминал. | `docker run -d nginx` |
| `--name` | Дать контейнеру понятное имя вместо случайного ID. | `docker run --name my_api python:3.11` |
| `-p` | Проброс порта. `хост:контейнер`. | `docker run -p 8080:80 nginx` (ваш порт 8080 -> порт 80 в контейнере) |
| `-v` | Монтирование тома (volume). Связать папку на хосте с папкой в контейнере. | `docker run -v /home/user/data:/app/data python:3.11` |
| `-e` | Передать переменную окружения. | `docker run -e SECRET_KEY=abc123 python:3.11` |
| `--rm` | Автоматически удалить контейнер после остановки. Полезно для одноразовых задач. | `docker run --rm python:3.11 python -c "print(1+1)"` |

**Примеры:**

```
# Простой запуск: скачает образ python:3.11, запустит, выполнит python --version, остановится
docker run python:3.11 python --version

# Интерактивный запуск: вы попадёте внутрь контейнера, в оболочку bash
docker run -it python:3.11 bash

# Фоновый запуск с именем
docker run -d --name my_redis redis:7

# Одноразовый запуск с удалением после выполнения
docker run --rm python:3.11 python -c "import numpy; print(numpy.__version__)"
```

#### 2.6.2. `docker ps` — посмотреть список контейнеров

```
docker ps          # Только работающие (Running)
docker ps -a       # Все: работающие, остановленные, созданные
docker ps -q       # Только ID (удобно для скриптов)
docker ps -l       # Последний созданный контейнер
```

Вывод показывает:
- `CONTAINER ID` — короткий ID.
- `IMAGE` — из какого образа создан.
- `COMMAND` — какая команда выполняется.
- `CREATED` — сколько времени назад создан.
- `STATUS` — состояние (Up 5 minutes, Exited (0) 2 minutes ago).
- `PORTS` — проброшенные порты.
- `NAMES` — имя контейнера.

#### 2.6.3. `docker stop`, `docker start`, `docker restart` — управление жизнью контейнера

```
docker stop my_container    # Корректно остановить (послать SIGTERM, подождать 10 сек, потом SIGKILL)
docker start my_container   # Запустить остановленный контейнер
docker restart my_container # Перезапустить
docker kill my_container    # Жёстко убить (SIGKILL, мгновенно)
```

**Разница между `stop` и `kill`:**  
`stop` — это как выключить компьютер через «Завершение работы». Система успевает сохранить файлы, закрыть программы.  
`kill` — это как выдернуть шнур питания. Мгновенно, но можно потерять данные.

#### 2.6.4. `docker exec` — выполнить команду внутри работающего контейнера

Это команда для «заглянуть внутрь работающего ящика».

```
# Выполнить одну команду внутри контейнера
docker exec my_container ls /app

# Открыть интерактивную оболочку внутри работающего контейнера
docker exec -it my_container bash
```

**Важно:** Контейнер должен быть в статусе `Up` (работать). Нельзя зайти в остановленный контейнер через `exec`.

#### 2.6.5. `docker logs` — посмотреть логи контейнера

Когда контейнер работает в фоне (`-d`), вы не видите его вывод. Логи помогают понять, что происходит.

```
docker logs my_container           # Показать все логи
docker logs -f my_container        # Follow: следить за логами в реальном времени (как tail -f)
docker logs --tail 50 my_container # Последние 50 строк
```

#### 2.6.6. `docker rm` и `docker rmi` — уборка

```
docker rm my_container      # Удалить остановленный контейнер
docker rm -f my_container   # Удалить работающий контейнер (форсированно)
docker rm $(docker ps -aq)  # Удалить ВСЕ контейнеры (осторожно!)

docker rmi python:3.11      # Удалить образ (если от него не зависят контейнеры)
docker image prune          # Удалить все образы без тегов (висячие)
docker system prune         # Удалить всё ненужное: остановленные контейнеры, висячие образы, неиспользуемые сети
```

### 2.7. Практика: запускаем Python внутри Docker

Это практическое занятие растянуто на несколько шагов. Выполняйте их последовательно. Перед началом убедитесь, что Docker Desktop запущен.

#### Шаг 0. Подготовка рабочей папки

Создайте на вашем компьютере папку `docker-practice` и перейдите в неё через терминал:

```
mkdir ~/docker-practice
cd ~/docker-practice
```

Все дальнейшие команды выполняйте из этой папки.

#### Шаг 1. Скачиваем образ Python

```
docker pull python:3.11-slim
```

**Что происходит:** Docker скачивает образ `python:3.11-slim` с Docker Hub. Тег `slim` означает облегчённую версию — внутри нет лишних инструментов, только минимум для работы Python. Это экономит место.

Проверьте, что образ скачался:

```
docker images
```

Вы увидите таблицу с образами. Там должна быть строка `python` с тегом `3.11-slim`.

#### Шаг 2. Интерактивный запуск: знакомство с внутренним миром

```
docker run -it python:3.11-slim bash
```

**Разбор команды:**
- `docker run` — создать и запустить.
- `-it` — интерактивно, с терминалом.
- `python:3.11-slim` — образ.
- `bash` — команда, которую выполнить внутри контейнера (открыть оболочку bash).

**Что вы увидите:**  
Приглашение командной строки изменится. Например:

```
root@a1b2c3d4e5f6:/#
```

Это означает, что вы **внутри** контейнера. Вы — пользователь `root`, ID контейнера `a1b2c3d4e5f6`, текущая папка `/` (корень).

**Проведите эксперименты внутри контейнера:**

```
# Узнать версию Python
python --version

# Узнать, какая операционная система внутри
cat /etc/os-release

# Посмотреть список процессов
ps aux

# Посмотреть, сколько свободного места
df -h

# Создать файл
echo "Hello from container" > /tmp/test.txt
cat /tmp/test.txt
```

**Важное наблюдение:** Выполните `exit` или нажмите `Ctrl+D`. Вы вышли из контейнера. Теперь на вашем хосте выполните:

```
cat /tmp/test.txt
```

Файла нет. Он существовал только внутри контейнера. Контейнер остановился, и все его изменения исчезли.

#### Шаг 3. Выполняем Python-скрипт без входа в контейнер

Создайте на хосте файл `script.py`:

```
# script.py
import sys
import platform

print(f"Python version: {platform.python_version()}")
print(f"Platform: {platform.system()}")
print(f"Arguments: {sys.argv}")

data = [1, 2, 3, 4, 5]
print(f"Sum: {sum(data)}")
```

Теперь запустим его внутри контейнера, не заходя в него интерактивно:

```
docker run --rm -v $(pwd):/app python:3.11-slim python /app/script.py hello 123
```

**Разбор команды:**
- `docker run` — создать и запустить.
- `--rm` — удалить контейнер после выполнения (чтобы не мусорить).
- `-v $(pwd):/app` — **монтирование тома**. `$(pwd)` — текущая папка на вашем компьютере. `/app` — папка внутри контейнера. Мы говорим Docker: «Сделай так, чтобы папка `/app` внутри контейнера видела содержимое моей текущей папки на хосте».
- `python:3.11-slim` — образ.
- `python /app/script.py hello 123` — команда, выполняемая внутри контейнера.

**Что произошло:**  
Контейнер запустился, увидел ваш файл `script.py` в папке `/app`, выполнил его на Python 3.11, напечатал результат, остановился и удалился.

**Проверьте:** `docker ps -a` — контейнера нет в списке (благодаря `--rm`).

#### Шаг 4. Запуск в фоновом режиме и подключение к работающему контейнеру

Запустим Python в фоне в режиме «бесконечного сна», чтобы контейнер не завершился:

```
docker run -d --name my_python_box python:3.11-slim sleep 3600
```

**Разбор:**
- `-d` — фоновый режим.
- `--name my_python_box` — дали имя, чтобы не запоминать ID.
- `sleep 3600` — главный процесс контейнера будет спать 3600 секунд (1 час). Контейнер останется работать всё это время.

Проверьте, что он работает:

```
docker ps
```

Должен появиться контейнер `my_python_box` со статусом `Up`.

Теперь подключимся к нему:

```
docker exec -it my_python_box bash
```

Вы снова внутри! Попробуйте:

```
python -c "import math; print(math.pi)"
exit
```

Контейнер продолжает работать, потому что вы вышли из `bash`, который запустился через `exec`, а не из главного процесса `sleep`.

Остановите и удалите контейнер:

```
docker stop my_python_box
docker rm my_python_box
```

#### Шаг 5. Проброс портов (подготовка к API)

Хотя полноценный API мы будем делать позже, важно понять, как вытащить наружу то, что работает внутри контейнера.

Создайте файл `simple_server.py`:

```
# simple_server.py
from http.server import HTTPServer, BaseHTTPRequestHandler

class Handler(BaseHTTPRequestHandler):
    def do_GET(self):
        self.send_response(200)
        self.send_header('Content-type', 'text/plain')
        self.end_headers()
        self.wfile.write(b"Hello from Docker container!\n")

if __name__ == "__main__":
    server = HTTPServer(("0.0.0.0", 8000), Handler)
    print("Server running on port 8000...")
    server.serve_forever()
```

Запустите контейнер с пробросом порта:

```
docker run --rm -p 8080:8000 -v $(pwd):/app python:3.11-slim python /app/simple_server.py
```

**Разбор `-p 8080:8000`:**
- `8080` — порт на вашем компьютере (хосте).
- `8000` — порт внутри контейнера, на котором слушает Python-сервер.
- Docker создаёт «трубу»: запрос на `localhost:8080` вашего компьютера попадает на порт `8000` внутри контейнера.

**Проверка:**  
Откройте браузер и перейдите по адресу `http://localhost:8080`. Вы увидите текст `Hello from Docker container!`. При этом сервер работает внутри контейнера, изолированно от вашей основной системы.

Остановите сервер: в терминале, где запущен контейнер, нажмите `Ctrl+C`. Контейнер удалится (`--rm`).

### 2.8. Что происходит под капотом (упрощённо, но точно)

Вы можете спросить: «Как Docker делает изоляцию? Это же не настоящая виртуальная машина!»

Правильный ответ: Docker использует **возможности ядра Linux**, которые называются:

1. **Namespaces (Пространства имён)** — создают иллюзию изоляции. Контейнер думает, что у него свой список процессов, своя сеть, своя файловая система. На самом деле это «фильтры» в ядре, которые ограничивают, что процесс может видеть.

2. **Control Groups (cgroups)** — ограничивают ресурсы. Можно сказать: «Этому контейнеру — максимум 1 ГБ оперативной памяти и 50% процессора». Если контейнер попытается съесть больше — ядро его ограничит.

3. **Union File System (OverlayFS)** — позволяет создавать многослойную файловую систему образа. Каждый слой образа read-only, а изменения контейнера пишутся в отдельный верхний слой. Это делает контейнеры лёгкими и быстрыми.

**Важно для новичка:** Вам не нужно глубоко понимать эти механизмы, чтобы пользоваться Docker. Но полезно знать, что «магия» — это не магия, а проверенные технологии ядра Linux. На Windows и macOS Docker Desktop использует лёгкую виртуальную машину с Linux, внутри которой и работают контейнеры.

### 2.9. Итоги модуля: чек-лист того, что вы теперь умеете

- [ ] Объяснить разницу между **образом**, **контейнером** и **реестром**.
- [ ] Запустить контейнер в **интерактивном** (`-it`) и **фоновом** (`-d`) режимах.
- [ ] Дать контейнеру **имя** (`--name`) и удалить его после работы (`--rm`).
- [ ] Посмотреть список работающих и остановленных контейнеров (`docker ps`, `docker ps -a`).
- [ ] Остановить, запустить и перезапустить контейнер (`docker stop/start/restart`).
- [ ] Выполнить команду внутри работающего контейнера (`docker exec -it`).
- [ ] Посмотреть логи контейнера (`docker logs -f`).
- [ ] **Смонтировать папку** с хоста в контейнер (`-v`) и выполнить свой Python-скрипт.
- [ ] **Пробросить порт** (`-p`) и получить доступ к сервису внутри контейнера из браузера.
- [ ] Понимать, что **изоляция** означает: свои процессы, своя файловая система, своя сеть.
- [ ] Удалять контейнеры и образы (`docker rm`, `docker rmi`, `docker system prune`).

**В следующем модуле** мы научимся создавать свои собственные образы. Вы напишете первый `Dockerfile`, поймёте, как устроены слои, и научитесь оптимизировать сборку, чтобы она занимала секунды, а не минуты.
