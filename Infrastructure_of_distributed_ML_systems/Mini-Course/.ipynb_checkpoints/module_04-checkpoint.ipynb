**Модуль 4. Multi-stage builds: радикальное похудение образа**

### 4.1. Проблема: почему ML-контейнеры толстеют

#### 4.1.1. Что такое «сборочные зависимости»

В Модуле 3 мы собирали простое FastAPI-приложение. Его зависимости (`fastapi`, `uvicorn`) — чистый Python. Они устанавливаются быстро и не требуют ничего, кроме самого Python.

Но в мире машинного обучения всё иначе. Библиотеки вроде **`numpy`**, **`pandas`**, **`scikit-learn`**, **`tensorflow`**, **`psycopg2-binary`** (драйвер PostgreSQL) содержат внутри себя **код на языке C** (или C++, Fortran). Этот код нужно **скомпилировать** — превратить из человекочитаемого текста в машинные инструкции, которые процессор понимает напрямую.

**Аналогия: мебель из IKEA vs мебель на заказ**

Представьте, что вы купили шкаф в IKEA. В коробке лежат уже готовые детали — вам только скрутить. Это «чистый Python» — всё готово, просто собери.

А теперь представьте, что вы заказали мебель у плотника. Он привозит в вашу квартиру доски, пилу, молоток, стамеску, шуруповёрт, лобзик, столярный клей, тиски и мешок опилок. Он работает у вас дома 3 часа, пилит, сверлит, строгает. В итоге у вас красивый шкаф. Но в квартире остаётся: пила, молоток, стамеска, шуруповёрт, тиски, мешок опилок, обрезки досок, пустые банки от лака.

**Сборочные зависимости** — это инструменты плотника: `gcc`, `g++`, `python3-dev`, `build-essential`, `libpq-dev`. Они нужны только на этапе установки пакетов. В готовом, работающем приложении они **бесполезны**. Но если вы оставите их в контейнере — образ раздуется в 3-5 раз.

#### 4.1.2. Конкретные цифры: насколько раздувается образ?

Давайте посчитаем на примере.

**Базовый образ `python:3.11-slim`:** ~120 МБ.  
**После установки `gcc`, `build-essential`, `python3-dev`:** +200-300 МБ.  
**После установки `numpy`, `pandas`, `scikit-learn`:** +150-200 МБ.  
**Кэш pip, логи компиляции, временные файлы:** +50-100 МБ.

**Итого:** Вместо 120 МБ вы получаете 600-800 МБ. А если добавить `tensorflow` — легко перевалите за 2-3 ГБ.

**Почему это плохо?**
1. **Медленная загрузка.** Образ весит 2 ГБ — его 10 минут качает сервер в облаке.
2. **Медленный запуск.** Контейнеру нужно распаковать эти 2 ГБ в память.
3. **Уязвимости.** Каждый лишний пакет (`gcc`, `make`, `wget`) — это потенциальная дверь для хакера. Чем меньше в образе, тем безопаснее.
4. **Деньги.** В облаке хранение и передача больших образов стоит дороже.

### 4.2. Идея Multi-stage builds: две мастерские в одном рецепте

#### 4.2.1. Аналогия: плотник и квартира

Вернёмся к плотнику. Представьте, что у вас есть **две отдельные комнаты**:

- **Комната сборки (builder):** Здесь плотник распиливает доски, сверлит, мажет клеем. Здесь грязно, здесь валяются инструменты, опилки, обрезки. Здесь он работает 3 часа.
- **Комната эксплуатации (production):** Сюда он заносит только готовый шкаф. Здесь чисто, уютно, никаких инструментов. Здесь живут люди.

**Multi-stage build** — это именно такой подход. В одном Dockerfile мы описываем две (или больше) «комнаты»:

1. **Этап сборки (builder stage):** Берём толстый образ со всеми инструментами. Устанавливаем зависимости. Компилируем всё, что нужно.
2. **Этап выполнения (runtime stage):** Берём лёгкий образ. Копируем из «комнаты сборки» только готовые артефакты — установленные пакеты, скомпилированные бинарники, код. Никаких `gcc`, никаких опилок.

#### 4.2.2. Как это выглядит в Dockerfile

Ключевое слово — **`AS`**. Оно даёт имя этапу:

```
# ЭТАП 1: СБОРКА (builder)
FROM python:3.11 AS builder

# ... здесь устанавливаем всё, что нужно для компиляции ...

# ЭТАП 2: ВЫПОЛНЕНИЕ (runtime)
FROM python:3.11-slim AS runtime

# ... здесь копируем только результаты из builder ...
```

Docker собирает оба этапа. Но в финальный образ попадает только последний `FROM` и всё, что описано после него. Предыдущие этапы остаются промежуточными — они используются как «источник», но не попадают в итоговый образ.

### 4.3. Практика: собираем ML-образ без мусора

#### Шаг 0. Подготовка проекта

Создайте папку:

```
mkdir ~/docker-module4
cd ~/docker-module4
```

Создайте файлы:

**`requirements.txt`:**

```
numpy==1.26.4
pandas==2.2.2
scikit-learn==1.5.0
fastapi==0.111.0
uvicorn[standard]==0.30.0
```

**`main.py` (ML API):**

```
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from fastapi import FastAPI

app = FastAPI(title="ML Model API")

# Обучаем простую модель при старте
model = LinearRegression()
X_train = np.array([[1], [2], [3], [4], [5]])
y_train = np.array([2.1, 4.0, 6.1, 7.8, 10.2])
model.fit(X_train, y_train)

@app.get("/")
def root():
    return {"message": "ML API is running", "model": "LinearRegression"}

@app.get("/predict")
def predict(x: float):
    prediction = model.predict(np.array([[x]]))[0]
    return {
        "input": x,
        "prediction": round(float(prediction), 4),
        "model_type": "scikit-learn"
    }

@app.get("/stats")
def stats():
    # Демонстрация работы pandas
    df = pd.DataFrame({
        "feature": [1, 2, 3, 4, 5],
        "target": [2.1, 4.0, 6.1, 7.8, 10.2]
    })
    return {
        "mean": float(df["target"].mean()),
        "std": float(df["target"].std())
    }
```

#### Шаг 1. Плохой Dockerfile (без multi-stage)

Создайте `Dockerfile.single`:

```
FROM python:3.11

WORKDIR /app

COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

COPY . .

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

Соберите и замерьте:

```
docker build -f Dockerfile.single -t ml_app_single .
docker images ml_app_single
```

Запишите размер. Скорее всего, это **1.0–1.2 ГБ**.

#### Шаг 2. Хороший Dockerfile (с multi-stage)

Создайте `Dockerfile`:

```
# ============================================================
# ЭТАП 1: BUILDER (комната плотника)
# ============================================================
FROM python:3.11 AS builder

# Устанавливаем рабочую директорию
WORKDIR /app

# Устанавливаем системные зависимости, нужные для компиляции
# gcc — компилятор C
# g++ — компилятор C++
# build-essential — набор инструментов для сборки
# python3-dev — заголовочные файлы Python для компиляции расширений
RUN apt-get update && apt-get install -y --no-install-recommends \
    gcc \
    g++ \
    build-essential \
    python3-dev \
    && rm -rf /var/lib/apt/lists/*

# Создаём виртуальное окружение в папке /opt/venv
# Виртуальное окружение — это изолированная папка со своим Python и пакетами
RUN python -m venv /opt/venv

# Активируем его (добавляем в PATH)
ENV PATH="/opt/venv/bin:$PATH"

# Копируем requirements и устанавливаем зависимости
# Они установятся ВНУТРИ /opt/venv
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# ============================================================
# ЭТАП 2: RUNTIME (чистая квартира)
# ============================================================
FROM python:3.11-slim AS runtime

# Отключаем .pyc и буферизацию
ENV PYTHONDONTWRITEBYTECODE=1
ENV PYTHONUNBUFFERED=1

# Устанавливаем рабочую директорию
WORKDIR /app

# Копируем виртуальное окружение целиком из builder
# Здесь уже есть numpy, pandas, scikit-learn — скомпилированные и готовые
COPY --from=builder /opt/venv /opt/venv

# Делаем так, чтобы при запуске использовался Python из виртуального окружения
ENV PATH="/opt/venv/bin:$PATH"

# Копируем только код приложения (без зависимостей, они уже в venv)
COPY . .

# Документируем порт
EXPOSE 8000

# Запускаем
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

**Разбор ключевых моментов:**

**`FROM python:3.11 AS builder`**  
Мы берём **полный** образ Python (не `slim`), потому что в нём уже есть много инструментов. К тому же мы доустанавливаем `gcc`, `g++`, `build-essential`. Этот этап будет тяжёлым, но он **не попадёт в финальный образ**.

**`RUN python -m venv /opt/venv`**  
Мы создаём виртуальное окружение. Зачем? Потому что так проще перенести все установленные пакеты целиком. Мы копируем папку `/opt/venv` из builder в runtime — и там оказывается Python + все библиотеки.

**`COPY --from=builder /opt/venv /opt/venv`**  
Это **сердце** multi-stage build. `--from=builder` говорит Docker: «Возьми файл из этапа с именем `builder`». Мы копируем всё виртуальное окружение — там уже установлены `numpy`, `pandas`, `scikit-learn`, и они **уже скомпилированы**. Но в финальном образе нет ни `gcc`, ни исходников — только готовые `.so` файлы (скомпилированные библиотеки).

**`FROM python:3.11-slim AS runtime`**  
Финальный образ — лёгкий. В нём нет компиляторов. Только Python и то, что мы скопировали из builder.

#### Шаг 3. Собираем и сравниваем

```
# Собираем многоэтапный образ
docker build -t ml_app_multi .

# Смотрим размеры
docker images | grep ml_app
```

**Ожидаемый результат:**

| Образ | Размер |
|-------|--------|
| `ml_app_single` | ~1000–1200 МБ |
| `ml_app_multi` | ~250–350 МБ |

**Экономия:** 3-4 раза меньше места. При этом функциональность абсолютно идентичная.

```
docker run -d --name ml_api -p 8000:8000 ml_app_multi
```

Проверьте:

### 4.4. .dockerignore: не тащим мусор в контекст сборки

#### 4.4.1. Что такое «контекст сборки»?

Когда вы выполняете `docker build -t my_app .` — точка в конце означает **контекст сборки**. Это папка, которую Docker передаёт демону (серверной части Docker) для сборки.

Docker упаковывает ВСЮ эту папку (все файлы, подпапки) и отправляет на сборку. Даже если в Dockerfile вы копируете только `requirements.txt` — Docker всё равно передаёт всё остальное «на всякий случай».

**Аналогия:** Вы несёте портфель на собеседование. В портфеле — резюме, которое вам нужно. Но вы также положили туда: старые чеки из супермаркета, зарядку от телефона 2015 года, полпачки печенья, носки на всякий случай, 50 распечатанных фотографий кота. Рекрутеру нужно только резюме, но он вынужден таскать весь портфель.

#### 4.4.2. Последствия «тяжёлого» контекста

- **Медленная сборка.** Docker упаковывает и передаёт гигабайты мусора.
- **Ломается кэш.** Если в папке есть файлы, которые меняются каждую секунду (логи, временные файлы), Docker видит, что «контекст изменился», и пересобирает слои.
- **Утечка данных.** Можно случайно скопировать в образ `.env` с паролями, `.git` с историей коммитов, личные фотографии.

#### 4.4.3. Синтаксис .dockerignore

Файл `.dockerignore` (с точкой в начале, без расширения) лежит рядом с Dockerfile. В нём перечислены файлы и папки, которые Docker должен **игнорировать** при сборке.

Синтаксис похож на `.gitignore`:

```
# Комментарий
*.pyc
__pycache__/
.env
.git/
.gitignore
*.md
!README.md          # Исключение из правила: README.md НЕ игнорировать
.venv/
venv/
node_modules/
.pytest_cache/
.mypy_cache/
.DS_Store
Thumbs.db
*.log
logs/
tmp/
data/               # Если data — большие датасеты, не нужные в образе
```

#### 4.4.4. Практика: создаём .dockerignore

В папке `~/docker-module4` создайте файл `.dockerignore`:

```
# Python
__pycache__/
*.py[cod]
*$py.class
*.so
.venv/
venv/
env/
ENV/
.ENV

# Git
.git/
.gitignore

# IDE
.idea/
.vscode/
*.swp
*.swo

# Логи и временные файлы
*.log
logs/
tmp/
*.tmp

# Тесты и документация (если не нужны в продакшене)
tests/
test/
*.md
!README.md

# Данные (если они монтируются через volume, а не копируются)
data/
datasets/

# Docker
Dockerfile*
.dockerignore

# Системные файлы
.DS_Store
Thumbs.db
```

Теперь проверьте размер контекста:

```
docker build -t ml_app_multi .
```

Docker напишет что-то вроде:

```
[+] Building 0.5s (15/15) FINISHED
 => [internal] load build context  0.1s
 => => transferring context: 5.12kB
```

Если бы `.dockerignore` не был — контекст мог бы быть 500 МБ (если рядом лежит датасет).

### 4.5. Продвинутая практика: ещё более лёгкий образ с `python:3.11-alpine`

**Примечание для новичка:** Этот раздел необязателен для первого прохождения, но полезен для понимания. `alpine` — это минималистичный дистрибутив Linux весом ~5 МБ. Но с ним сложнее работать, потому что там нет привычных инструментов (`bash`, `glibc`). Для ML он часто не подходит из-за проблем совместимости с `numpy`. Но знать о нём стоит.

Если вы хотите максимально ужать образ:

```
# ЭТАП СБОРКИ
FROM python:3.11-alpine AS builder

RUN apk add --no-cache gcc musl-dev python3-dev libffi-dev
RUN python -m venv /opt/venv
ENV PATH="/opt/venv/bin:$PATH"
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# ЭТАП ВЫПОЛНЕНИЯ
FROM python:3.11-alpine AS runtime

ENV PYTHONDONTWRITEBYTECODE=1
ENV PYTHONUNBUFFERED=1
WORKDIR /app
COPY --from=builder /opt/venv /opt/venv
ENV PATH="/opt/venv/bin:$PATH"
COPY . .
EXPOSE 8000
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
```

**Результат:** Образ может ужаться до **150–200 МБ**. Но `alpine` требует `apk` вместо `apt-get` и иногда ломает компиляцию сложных пакетов. Для начала используйте `python:3.11-slim` — это золотая середина.

### 4.6. Сравнительная таблица подходов

| Подход | Размер | Сложность | Когда использовать |
|--------|--------|-----------|-------------------|
| **Один этап, полный образ** | 1000+ МБ | Просто | Никогда в продакшене |
| **Один этап, slim** | 400–600 МБ | Просто | Быстрые прототипы без C-зависимостей |
| **Multi-stage + slim** | 200–400 МБ | Средне | **Стандарт для ML и backend** |
| **Multi-stage + alpine** | 100–200 МБ | Сложно | Микросервисы, где критичен каждый мегабайт |
