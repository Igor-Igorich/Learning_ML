Kimi: ## Модуль 1. Фундамент: процессы, изоляция и сетевая модель

### 1.1. Процессы, потоки и адресное пространство в Linux

#### 1.1.1. Что такое процесс

**Процесс** — это экземпляр компьютерной программы во время её выполнения. Но это определение слишком абстрактное. На уровне операционной системы процесс — это **контейнер ресурсов**: выделенное адресное пространство, открытые файлы, сокеты, сигналы и хотя бы один поток исполнения.

В ядре Linux каждый процесс представлен структурой `task_struct` (определена в `<linux/sched.h>`). Эта структура содержит сотни полей, но ключевые для понимания:
- **PID** (Process ID): уникальный идентификатор в пределах пространства имён PID.
- **state**: состояние процесса (`TASK_RUNNING`, `TASK_INTERRUPTIBLE`, `TASK_UNINTERRUPTIBLE`, `TASK_STOPPED`, `EXIT_ZOMBIE`).
- **mm_struct**: указатель на дескриптор памяти (виртуальное адресное пространство).
- **files_struct**: таблица открытых файловых дескрипторов.
- **nsproxy**: указатель на пространства имён (namespaces), к которым принадлежит процесс.
- **signal_struct**: обработчики сигналов.

**Жизненный цикл процесса:**
1. **Создание.** Родительский процесс вызывает `fork()` (или `clone()`), ядро копирует `task_struct` и связывает новый процесс с родителем.
2. **Выполнение.** Планировщик выделяет процессу квант времени CPU. Процесс находится в состоянии `TASK_RUNNING`.
3. **Ожидание.** Процесс блокируется на I/O (`TASK_INTERRUPTIBLE` — можно разбудить сигналом; `TASK_UNINTERRUPTIBLE` — ожидание диска, игнорирует сигналы).
4. **Завершение.** Процесс вызывает `exit()`. Ядро освобождает большинство ресурсов, но `task_struct` остаётся до тех пор, пока родитель не вызовет `wait()` — это состояние **зомби** (`EXIT_ZOMBIE`).
5. **Сбор.** Родитель получает код возврата, ядро удаляет `task_struct`.

**Почему зомби важны для контейнеров?** Если PID 1 в контейнере (инициализирующий процесс) не обрабатывает сигналы `SIGCHLD` и не вызывает `wait()`, зомби-процессы накапливаются, исчерпывая таблицу PID. Именно поэтому в Docker-контейнерах часто используют специальный init-процесс (`tini`, `dumb-init`).

#### 1.1.2. Виртуальное адресное пространство

**Проблема физической памяти.**
В ранних ОС программы работали напрямую с физической памятью. Это порождало три катастрофические проблемы:
1. **Фрагментация.** Программе нужен непрерывный блок памяти, но после множества аллокаций и освобождений свободная память представляет собой «решётку» мелких кусков.
2. **Безопасность.** Процесс A может прочитать или перезаписать память процесса B.
3. **Переносимость.** Программа привязана к конкретным физическим адресам и не может быть загружена по другому адресу.

**Виртуальная память как абстракция.**
Каждый процесс видит **собственное** непрерывное адресное пространство, начинающееся с нуля. Это пространство отображается на физическую память (RAM) или страничный файл (swap) через механизм **страничной адресации**.

**Структура виртуального адресного пространства** (типичная для x86_64 Linux):

```
0x0000000000000000  +------------------+
                    |       text       |  Сегмент кода (read-only, executable)
                    |   (программа)    |
0x0000000000400000  +------------------+
                    |       data       |  Инициализированные глобальные переменные
                    |                  |
                    +------------------+
                    |       bss        |  Неинициализированные глобальные переменные (zero-initialized)
                    |                  |
                    +------------------+
                    |       heap       |  Динамическая память (malloc), растёт вверх
                    |        ↓         |
                    |                  |
                    +------------------+
                    |   memory mapping |  Файлы, библиотеки (mmap)
                    |     region       |
                    +------------------+
                    |                  |
                    |       ↓          |
                    |      stack       |  Локальные переменные, вызовы функций, растёт вниз
                    |                  |
0x00007fffffffffff  +------------------+
```

**Страничная адресация.**
Виртуальное адресное пространство разбито на **страницы** (pages), обычно по 4 KiB (4096 байт). Физическая память разбита на **фреймы** (frames) того же размера. Отображение виртуальных страниц на физические фреймы осуществляется через **таблицы страниц** (page tables).

**Многоуровневые таблицы страниц (x86_64).**
В 64-битном режиме используется 4-уровневая таблица:
- Виртуальный адрес разбивается на индексы: **PML4** (9 бит) -> **PDPT** (9 бит) -> **PD** (9 бит) -> **PT** (9 бит) -> **Offset** (12 бит).
- $2^{12} = 4096$ байт — размер страницы.
- $2^9 = 512$ записей в каждой таблице.
- $512^4 \approx 2^{48}$ адресуемых страниц.

**TLB (Translation Lookaside Buffer).**
Проход по 4 уровням таблиц страниц при каждом обращении к памяти — дорогостоящая операция (4 обращения к RAM). CPU кэширует последние использованные отображения в специальном ассоциативном кэше — TLB. Промах TLB (TLB miss) стоит десятки тактов.

**Математическая подоплёка.**
Виртуальный адрес $V$ можно представить как кортеж $(l_4, l_3, l_2, l_1, o)$, где $l_i$ — индексы на уровнях таблиц, $o$ — смещение внутри страницы. Физический адрес $P$ находится как:

$$P = \text{PFN}(l_4, l_3, l_2, l_1) \times \text{PAGE\_SIZE} + o$$

где PFN (Page Frame Number) — номер физического фрейма, извлекаемый из последней таблицы. Это композиция частичных функций:

$$\text{translate}: V \xrightarrow{PT_4} PDPT \xrightarrow{PT_3} PD \xrightarrow{PT_2} PT \xrightarrow{PT_1} \text{Physical}$$

#### 1.1.3. Таблица файловых дескрипторов

**Файловый дескриптор (fd)** — это неотрицательное целое число, абстрагирующее доступ к ресурсу. В Linux «всё есть файл»: обычные файлы, директории, сокеты, pipe-ы, терминалы, устройства.

**Три уровня абстракции:**
1. **Файловый дескриптор процесса** — индекс в массиве `fd_array` структуры `files_struct`.
2. **File table entry** (`struct file`) — описывает открытый файл: текущая позиция (offset), режим доступа (read/write), флаги (O_APPEND, O_NONBLOCK), указатель на dentry.
3. **Inode** (`struct inode`) — метаданные файла в файловой системе.

```
fd (0, 1, 2, 3...) 
  -> files_struct.fd_array[fd] 
    -> struct file (offset, mode, f_op)
      -> struct dentry (имя файла)
        -> struct inode (метаданные, блоки на диске)
```

**Стандартные дескрипторы:**
- **0 — stdin** (standard input). По умолчанию связан с терминалом.
- **1 — stdout** (standard output).
- **2 — stderr** (standard error).

**Наследование при fork() и exec().**
При `fork()` таблица дескрипторов копируется. Родитель и потомок делят `struct file` (счётчик ссылок увеличивается). Это значит, что если оба пишут в fd=1, они пишут в один и тот же файл с общим offset-ом. При `exec()` дескрипторы сохраняются, если не установлен флаг `O_CLOEXEC` (close on exec).

**Для Docker это критично:** контейнер наследует дескрипторы от dockerd. Если dockerd не закрыл лишние fd, они «утекают» в контейнер — потенциальная угроза безопасности.

#### 1.1.4. Системные вызовы fork, exec, clone

**fork() — создание нового процесса.**
`fork()` создаёт почти точную копию вызывающего процесса. Потомок получает:
- Копию виртуального адресного пространства (но через Copy-on-Write, см. 1.3.3).
- Копию таблицы дескрипторов (разделяемые `struct file`).
- Копию контекста (регистры, счётчик команд).

Возвращаемое значение:
- В родителе: PID потомка (> 0).
- В потомке: 0.
- При ошибке: -1.

**exec() — замещение образа.**
Семейство функций `execve()` и т.д. загружает новый бинарный файл в текущее адресное пространство, полностью заменяя код, данные, стек и heap. PID не меняется. Это стандартный паттерн: `fork()` -> `exec()`.

**wait() и waitpid() — сбор зомби.**
Родитель блокируется (или проверяет неблокирующе) до завершения потомка. Ядро передаёт код возврата (exit status) и освобождает `task_struct`.

**clone() — обобщённый механизм.**
В Linux нет отдельного системного вызова для создания потоков. `pthread_create()` в glibc внутри вызывает `clone()` с определёнными флагами:
- `CLONE_VM` — разделять адресное пространство (потоки внутри процесса).
- `CLONE_FS` — разделять информацию о файловой системе (cwd, umask).
- `CLONE_FILES` — разделять таблицу дескрипторов.
- `CLONE_SIGHAND` — разделять обработчики сигналов.
- `CLONE_THREAD` — поместить в ту же группу потоков (один PID для всех с точки зрения пространства имён).

Контейнеры создаются через `clone()` с флагами пространств имён: `CLONE_NEWPID`, `CLONE_NEWNET`, `CLONE_NEWNS` и т.д.

#### 1.1.5. Потоки ядра и пользовательские потоки

**Поток (thread)** — это единица планирования (scheduling entity) внутри процесса. В Linux потоки реализованы как **лёгкие процессы** (Light Weight Processes, LWP): каждый поток имеет собственный `task_struct` и PID в ядре, но делит адресное пространство с другими потоками через `CLONE_VM`.

**Модели потоков:**
1. **1:1 (Kernel-level threads).** Каждый пользовательский поток соответствует одному потоку ядра. Linux, Windows. Преимущества: true parallelism, блокировка одного потока не блокирует процесс. Недостатки: дорогие переключения контекста.
2. **N:M (Hybrid threads).** N пользовательских потоков мультиплексируются на M потоков ядра. Solaris, старые реализации. Сложность реализации, редко используется сейчас.
3. **Green threads (N:1).** N пользовательских потоков на одном потоке ядра. Ранние JVM. Блокировка одного потока блокирует все.

**Thread Local Storage (TLS).**
Потоки делят память, но иногда нужны переменные, уникальные для каждого потока. Регистр `FS` (x86_64) указывает на TLS-сегмент. В glibc `errno` реализована через TLS, чтобы каждый поток видел свой код ошибки.

**Futex (Fast Userspace muTEX).**
Большинство операций блокировки не требуют системных вызовов. Если мьютекс свободен, поток захватывает его атомарной операцией в userspace. Если занят — только тогда вызывается `futex()` для блокировки в ядре. Это снижает накладные расходы на синхронизацию.

#### 1.1.6. Планировщик ОС: CFS

**Зачем нужен планировщик?**
В системе обычно больше готовых к выполнению процессов, чем физических ядер CPU. Планировщик решает, **какой процесс** и **на какое время** получит CPU.

**Completely Fair Scheduler (CFS)** — планировщик по умолчанию в Linux с версии 2.6.23. Вместо квантов времени CFS использует концепцию **виртуального времени** (vruntime).

**Красно-чёрное дерево (rbtree).**
Все runnable-процессы хранятся в самобалансирующемся двоичном дереве поиска — красно-чёрном дереве. Ключом является `vruntime`. Левыйmost узел — процесс с наименьшим vruntime, который и получит CPU.

**Виртуальное время.**
Физическое время, нормализованное по «весу» процесса:

$$\text{vruntime}_i = \text{vruntime}_i + \frac{\text{weight}_0}{\text{weight}_i} \times \Delta t$$

где:
- $\Delta t$ — физическое время, которое процесс провёл на CPU.
- $\text{weight}_0 = 1024$ — базовый вес (nice = 0).
- $\text{weight}_i$ — вес процесса с nice-значением $i$.

Nice values: от -20 (самый высокий приоритет) до +19 (самый низкий). Веса вычисляются по формуле:

$$\text{weight}(n) = \frac{1024}{1.25^n}$$

где $n$ — nice-значение. Это гарантирует, что процесс с nice=-20 получит примерно в 3 раза больше CPU, чем с nice=0, а с nice=+19 — примерно в 3 раза меньше.

**Свойства CFS:**
- **Пропорциональная справедливость.** Если два процесса с одинаковым nice работают бесконечно долго, их vruntime стремится к равенству.
- **O(log N)** на вставку/удаление из rbtree.
- **Минимальная гранулярность:** `sched_min_granularity_ns` (обычно 0.75–4 мс). Процесс не будет вытесняться чаще этого интервала.

**Контекстное переключение (context switch).**
Когда планировщик решает переключить CPU на другой процесс:
1. Сохранение регистров текущего процесса (в `task_struct` или на стек ядра).
2. Сохранение счётчика команд (RIP/EIP) и указателя стека (RSP/ESP).
3. Обновление таблицы страниц (если процессы разные — flush TLB).
4. Загрузка регистров нового процесса.
5. Переход в пространство пользователя.

Накладные расходы: 1–10 мкс для потоков, до 1–10 мс для процессов (из-за инвалидации TLB).

### 1.2. Механизмы изоляции ОС

#### 1.2.1. Namespaces: разбиение глобальных ресурсов

**Проблема.**
В классической Unix-системе многие ресурсы глобальны: все процессы видят один и тот же список PID, одни и те же сетевые интерфейсы, одну и ту же файловую систему, один hostname. Это означает, что процесс A может послать сигнал процессу B (`kill(pid, SIGTERM)`), прослушивать сетевой трафик процесса C (`tcpdump`), или смонтировать файловую систему, видимую всем.

**Namespaces** — механизм ядра Linux, позволяющий разделить эти глобальные ресурсы между группами процессов. Каждый процесс принадлежит одному namespace каждого типа. Процессы в разных namespaces не видят друг друга (с точки зрения данного ресурса).

**Семь типов namespaces:**

| Namespace | Флаг clone() | Изолирует | Для чего нужен в Docker |
|---|---|---|---|
| **PID** | `CLONE_NEWPID` | Идентификаторы процессов | Процесс в контейнере видит PID 1, хотя в хосте у него PID 3421 |
| **NET** | `CLONE_NEWNET` | Сетевые интерфейсы, таблицы маршрутизации, iptables, сокеты | Контейнер имеет свой lo, свой eth0 (veth pair), свои порты |
| **MNT** | `CLONE_NEWNS` | Точки монтирования | Контейнер видит только свой rootfs, не видет /host |
| **UTS** | `CLONE_NEWUTS` | Hostname и NIS domain name | `hostname` внутри контейнера не меняет хост |
| **IPC** | `CLONE_NEWIPC` | System V IPC и POSIX message queues | Изоляция shared memory между контейнерами |
| **USER** | `CLONE_NEWUSER` | UID/GID | root в контейнере (UID 0) может быть UID 100000 на хосте |
| **CGROUP** | `CLONE_NEWCGROUP` | Представление о cgroups | Процесс в контейнере видит только свою cgroup |

**Как создаются namespaces:**
1. **`clone()`** с соответствующим флагом — новый процесс создаётся в новом namespace.
2. **`unshare()`** — вызывающий процесс перемещается в новый namespace (используется Docker при `docker exec`).
3. **`setns()`** — процесс присоединяется к существующему namespace по файловому дескриптору (`/proc/[pid]/ns/*`).

**PID namespace — подробнее.**
PID namespace образуют иерархию. Процесс в дочернем namespace может иметь PID 1 (инициализирующий процесс). При выходе PID 1 все остальные процессы этого namespace получают `SIGKILL`. Это объясняет, почему в Docker-контейнере завершение PID 1 убивает весь контейнер.

**User namespace — подробнее.**
Это самый мощный namespace с точки зрения безопасности. Внутри контейнера процесс может иметь UID 0 (root) и все capabilities, но с точки зрения хоста его UID отображается на непривилегированного пользователя (например, UID 100000). Файлы, созданные внутри контейнера, имеют UID снаружи, что предотвращает escape из контейнера.

**Математическая подоплёка.**
Пусть $R$ — универсальное множество ресурсов ОС (процессы, сетевые интерфейсы, точки монтирования). Namespace типа $T$ определяет разбиение $R_T$ на непересекающиеся подмножества:

$$\bigcup_{i} N_{T,i} = R_T, \quad N_{T,i} \cap N_{T,j} = \emptyset \text{ для } i \neq j$$

Процесс $p$ принадлежит кортежу namespaces $(n_{\text{PID}}, n_{\text{NET}}, ..., n_{\text{CGROUP}})$. Два процесса изолированы по типу $T$ тогда и только тогда, когда их $n_T$ различны. Это формально определяет отношение эквивалентности на множестве процессов: $p \sim_T q \iff n_T(p) = n_T(q)$.

#### 1.2.2. Cgroups v1 и v2: управление ресурсами

**Концепция.**
Namespaces изолируют *видимость* ресурсов, но не ограничивают *потребление*. Если процесс в контейнере запустит `fork bomb` (бесконечное создание процессов), он исчерпает PID хоста. Если он выделит всю RAM, хост начнёт своппинг. **Cgroups (Control Groups)** решают эту проблему: они группируют процессы и применяют к ним политики ограничения ресурсов.

**Cgroups v1 (legacy).**
В cgroups v1 каждый подсистемный контроллер (cpu, memory, blkio, pids, devices, net_cls, net_prio, freezer, cpuset) монтируется в отдельную иерархию файловой системы:

```
/sys/fs/cgroup/cpu/docker/...    # CPU ограничения
/sys/fs/cgroup/memory/docker/... # Memory ограничения
/sys/fs/cgroup/pids/docker/...   # Ограничение числа процессов
```

Процесс может принадлежать разным группам в разных иерархиях. Это приводит к сложностям: например, `freezer` (заморозка процесса) и `cpuset` (привязка к ядрам) могут конфликтовать.

**Cgroups v2 (unified hierarchy).**
Единая иерархия, все контроллеры управляются из одного дерева. Ключевые улучшения:
- **Threaded mode:** потоки внутри процесса могут принадлежать разным cgroup.
- **Delegation model:** безопасная передача управления cgroup непривилегированному процессу.
- **Root cgroup:** все процессы изначально в корне, перемещаются в дочерние группы.
- **Pressure Stall Information (PSI):** метрики давления на CPU, memory, I/O.

**CPU контроллер:**
- **`cpu.shares`** (v1) / **`cpu.weight`** (v2): относительное взвешивание. Если контейнер A имеет weight=512, а B — 1024, при конфликте за CPU B получит в 2 раза больше. Это *мягкое* ограничение: если CPU свободен, контейнер может использовать всё.
- **`cpu.cfs_quota_us`** и **`cpu.cfs_period_us`** (v1) / **`cpu.max`** (v2): жёсткое ограничение. Например, `cpu.max = "50000 100000"` означает: 50 мс CPU на каждые 100 мс реального времени = 0.5 CPU.
- **`cpuset.cpus`**: привязка к конкретным физическим ядрам (NUMA-оптимизация).

**Memory контроллер:**
- **`memory.limit_in_bytes`** (v1) / **`memory.max`** (v2): жёсткий лимит RAM. При превышении процесс получает `OOM Killer` (Out-Of-Memory Killer) — ядро принудительно завершает процесс.
- **`memory.swap.max`**: лимит swap.
- **`memory.high`**: «мягкий» лимит — ядро начинает активно реклаймить память (shrink page cache, вызвать `kswapd`) до жёсткого ограничения.
- **`memory.oom.group`** (v2): если один процесс в cgroup погибает от OOM, убиваются все процессы группы. Критично для контейнеров: если PID 1 погибает, весь контейнер должен упасть.

**Blkio контроллер:**
- **`io.max`** (v2): ограничение IOPS и bandwidth для блочных устройств. Например, `wbps=10485760` (10 МБ/с запись).

**Pids контроллер:**
- **`pids.max`**: максимальное количество процессов и потоков в cgroup. Защита от fork bomb.

**Devices контроллер:**
- **Белый список** (v2) или **чёрный список** (v1) доступа к `/dev/*`. По умолчанию контейнер не имеет доступа к `/dev/sda`, но может читать `/dev/null`, `/dev/zero`, `/dev/random`. Для GPU-контейнеров NVIDIA Docker добавляет `/dev/nvidia*`.

**Математическая подоплёка.**
Cgroup определяет функцию ограничения:

$$L: \mathcal{P}(\text{processes}) \times \text{resource} \to \mathbb{R}_{\geq 0} \cup \{\infty\}$$

где $L(G, r)$ — лимит ресурса $r$ для группы процессов $G$. Для CPU shares это отношение:

$$\frac{\text{CPU}_A}{\text{CPU}_B} = \frac{\text{weight}_A}{\text{weight}_B} \quad \text{при условии } \sum \text{CPU}_i = \text{total\_CPU}$$

Для hard limit это индикаторная функция:

$$\text{throttle}(t) = \begin{cases} 1 & \text{if } \int_{t-T}^{t} \text{CPU}(s)\,ds < \text{quota} \\ 0 & \text{otherwise} \end{cases}$$

где $T$ — период, quota — квота внутри периода.

#### 1.2.3. Capabilities: дизайнерские права root

**Проблема привилегий Unix.**
В классической модели Unix есть два режима: root (UID 0) и обычный пользователь. Root может делать всё: монтировать файловые системы, менять сетевую конфигурацию, читать любую память, отправлять сигналы любому процессу. Это «всё или ничего» — небезопасно.

**Capabilities** (введены в Linux 2.2) разбивают привилегии root на 41 тонкое полномочие. Некоторые примеры:
- `CAP_NET_ADMIN`: конфигурация сетевых интерфейсов, маршрутизация, iptables.
- `CAP_SYS_PTRACE`: трассировка произвольных процессов (`strace`, `gdb`).
- `CAP_SYS_ADMIN`: «новый root» — монтирование, pivot_root, многие административные операции.
- `CAP_DAC_OVERRIDE`: обход проверок прав доступа на чтение/запись/исполнение.
- `CAP_SETUID`: изменение UID процесса.
- `CAP_KILL`: отправка сигналов любому процессу.

**Наборы capabilities.**
Каждый процесс имеет пять наборов:
1. **Permitted:** максимальный набор, который процесс может запросить.
2. **Effective:** текущие активные полномочия (проверяются при системных вызовах).
3. **Inheritable:** наследуется при `exec()`.
4. **Bounding:** верхняя граница для всего дерева процессов (может только уменьшаться).
5. **Ambient:** наследуется при `exec()` непривилегированными программами (добавлено в Linux 4.3).

**Docker и capabilities.**
По умолчанию Docker запускает контейнер с ограниченным набором capabilities (14 штук), исключая опасные (`CAP_SYS_ADMIN`, `CAP_SYS_PTRACE`, `CAP_SYS_MODULE`). Флаг `--privileged` снимает все ограничения, добавляя все capabilities и отключая многие механизмы безопасности — это антипаттерн для production.

**Математическая подоплёка.**
Пусть $\mathcal{C}$ — множество всех capabilities. Процесс $p$ имеет наборы $P(p), E(p), I(p), B(p), A(p) \subseteq \mathcal{C}$. При `fork()` потомок наследует наборы родителя. При `exec()`:

$$E_{\text{new}} = (E_{\text{old}} \cap B_{\text{new}}) \cup A_{\text{new}}$$

где $B_{\text{new}}$ — bounding set нового бинарника. Это гарантирует, что программа не может получить больше прав, чем позволено bounding set.

#### 1.2.4. Дополнительные механизмы безопасности

**Seccomp (Secure Computing Mode).**
Фильтрация системных вызовов через BPF (Berkeley Packet Filter). Процесс может перейти в режим, где разрешён только минимальный набор syscall (`read`, `write`, `exit`, `sigreturn`). Docker по умолчанию использует профиль `seccomp-default`, блокирующий ~44 опасных системных вызовов (например, `mount`, `pivot_root`, `swapon`).

**AppArmor и SELinux — MAC (Mandatory Access Control).**
- **AppArmor** (Ubuntu/Debian): path-based. Правила вида «программа `/usr/bin/nginx` может читать `/var/www/*`».
- **SELinux** (RHEL/CentOS/Fedora): label-based. Каждый объект (файл, процесс, сокет) имеет метку безопасности (context). Правила определяют, какие метки могут взаимодействовать.

Docker поддерживает оба: `--security-opt apparmor=...` и `--security-opt seccomp=...`.

### 1.3. Файловые системы и Copy-on-Write

#### 1.3.1. От VFS до ext4: архитектура файловой системы Linux

**VFS (Virtual File System).**
Linux поддерживает десятки файловых систем (ext4, XFS, btrfs, tmpfs, NFS, overlayfs). Чтобы не писать отдельный код для каждой в ядре, существует абстракция VFS — единый интерфейс. VFS определяет четыре ключевые структуры:
1. **`struct superblock`** — метаданные файловой системы целиком (размер, свободное место, блоки).
2. **`struct inode`** — метаданные файла (права, владелец, размер, timestamps, указатели на блоки данных). Уникален в пределах файловой системы (inode number).
3. **`struct dentry`** (directory entry) — кэш соответствия между именем файла и inode. Ускоряет повторный `open()`.
4. **`struct file`** — представление открытого файла (позиция, режим доступа).

**Inode — сердце файловой системы.**
Inode не содержит имени файла. Имя хранится в директории как пара `(name, inode_number)`. Один inode может иметь несколько имён (hard links). Inode содержит:
- Mode (тип и права: regular file, directory, symlink, socket, pipe).
- UID/GID владельца.
- Размер в байтах.
- Timestamps (atime, mtime, ctime).
- Количество hard links.
- Указатели на блоки данных (direct blocks, indirect blocks, double indirect, triple indirect — в ext2/3; extents в ext4).

**Ext4: extent-based файловая система.**
В отличие от ext3, где блоки файла описывались цепочкой индексных блоков (inefficient для больших файлов), ext4 использует **extents** — непрерывные диапазоны блоков. Один extent описывает «файл занимает блоки от 1000 до 5000». Это радикально ускоряет работу с большими файлами (образы Docker!) и уменьшает фрагментацию.

**Журналирование (Journaling).**
При сбое питания файловая система может оказаться в неконсистентном состоянии (например, inode обновлён, но данные не записаны). Журнал — это кольцевой буфер операций:
- **Journal mode (data=journal):** журналируются и метаданные, и данные. Максимальная надёжность, минимальная производительность.
- **Ordered mode (data=ordered, default):** журналируются только метаданные, но данные гарантированно записаны до метаданных. Компромисс.
- **Writeback mode:** журналируются только метаданные, данные могут быть записаны позже. Максимальная скорость, риск потери данных при сбое.

#### 1.3.2. Файловые дескрипторы, pipe-ы и IPC

**Pipe (анонимный канал).**
Создаётся системным вызовом `pipe(int fd[2])`. Возвращает два дескриптора: `fd[0]` для чтения, `fd[1]` для записи. Pipe работает только между родственными процессами (обычно parent-child). В ядре реализован как кольцевой буфер фиксированного размера (обычно 64 KiB в современных ядрах).

**FIFO (named pipe).**
Создаётся `mkfifo()`. Имеет имя в файловой системе, но не содержит данных на диске — данные хранятся в памяти. Позволяет общаться неродственным процессам.

**Socketpair.**
`socketpair()` создаёт два связанных сокета (Unix domain sockets), работающих в режиме SOCK_STREAM. Двунаправленный канал, более гибкий, чем pipe.

**Математическая подоплёка.**
Pipe — это классическая задача **bounded buffer** (ограниченный буфер) из теории параллельных вычислений:
- Производитель (producer) пишет в `fd[1]`.
- Потребитель (consumer) читает из `fd[0]`.
- Если буфер полон, производитель блокируется (`write()` возвращает -1 с `EAGAIN` в неблокирующем режиме).
- Если буфер пуст, потребитель блокируется.

Это семафор Дейкстры в действии: ядро использует внутренние семафоры для синхронизации доступа к кольцевому буферу.

#### 1.3.3. Copy-on-Write (COW) на уровне памяти

**Идея.**
При `fork()` ядро не копирует всё адресное пространство родителя. Вместо этого оно помечает все страницы памяти обоих процессов как **read-only** и увеличивает счётчик ссылок (reference count) на каждую страницу. Оба процесса «делят» одни и те же физические страницы.

**Page fault при записи.**
Когда любой из процессов пытается записать в read-only страницу, CPU генерирует **page fault** (исключение #PF). Обработчик page fault в ядре проверяет:
1. Это COW-страница (флаг `VM_SHARED` не установлен, счётчик ссылок > 1).
2. Ядро выделяет новую физическую страницу.
3. Копирует содержимое старой страницы в новую.
4. Обновляет PTE (Page Table Entry) процесса-писателя: новая страница, права read-write.
5. Уменьшает счётчик ссылок старой страницы.
6. Возвращает управление в userspace, инструкция записи выполняется повторно.

**Почему fork() быстрый.**
Сложность `fork()` без COW: $O(\text{size\_of\_memory})$ — копирование всех страниц. С COW: $O(1)$ — только копирование таблиц страниц (page tables), что значительно меньше. Фактическое копирование происходит лениво (lazy), только для изменённых страниц.

**Математическая подоплёка.**
Пусть $S$ — множество страниц адресного пространства, $|S| = n$. При `fork()` создаётся копия отображения $M': S \to \text{PhysicalPages}$, где $M'(p) = M(p)$ для всех $p \in S$. Стоимость: $O(n)$ для копирования PTE, $O(1)$ для физических страниц.

При записи в страницу $p \in S$ процессом $i$:

$$\text{if } \text{refcount}(M(p)) > 1: \quad M_i(p) \leftarrow \text{alloc\_page}(); \quad \text{memcpy}(M_i(p), M(p)); \quad \text{refcount}(M(p)) \mathbin{{-}{=}} 1$$

Амортизированная стоимость $k$ операций записи после fork: $O(k)$, а не $O(n)$. Если $k \ll n$, выигрыш колоссален.

#### 1.3.4. OverlayFS: слоистая файловая система

**Проблема, которую решает OverlayFS.**
Docker-образ состоит из слоёв (layers). Каждый слой — это read-only набор файлов. Контейнер добавляет writable layer поверх. Нужна файловая система, которая:
1. Показывает объединённое представление (merged view) нескольких директорий.
2. Позволяет читать из нижних слоёв, но писать только в верхний.
3. При удалении файла из нижнего слоя не модифицирует нижний слой.

**Архитектура OverlayFS.**
OverlayFS монтирует объединение директорий:
- **`lowerdir`** — один или несколько read-only слоёв (базовые образы). Читаются снизу вверх: если файл есть в нескольких lowerdir, используется верхний.
- **`upperdir`** — read-write слой (изменения контейнера).
- **`workdir`** — рабочая директория для внутренних операций OverlayFS (нужна для атомарных операций).
- **`merged`** — точка монтирования, видимая процессу. Это виртуальное объединение.

**Как работает чтение:**
1. Файл существует в `upperdir`? -> Читаем оттуда.
2. Нет в `upperdir`, но есть в `lowerdir`? -> Читаем из верхнего `lowerdir`, где он найден.
3. Нигде нет? -> `ENOENT` (файл не найден).

**Как работает запись:**
1. Файл существует в `upperdir`? -> Пишем напрямую.
2. Файл есть только в `lowerdir`? -> **Copy-up:** OverlayFS копирует файл из `lowerdir` в `upperdir` (вместе с родительскими директориями, если нужно), затем пишет в `upperdir`. Нижний слой остаётся нетронутым.

**Как работает удаление:**
1. Файл в `upperdir`? -> Удаляем.
2. Файл только в `lowerdir`? -> Создаётся **whiteout** — специальный файл (char device с мажор/минор 0:0) или расширенный атрибут `trusted.overlay.opaque`, который «маскирует» файл в lowerdir. При чтении merged файла whiteout означает «этого файла нет».

**Математическая подоплёка.**
Пусть $L = (l_1, l_2, ..., l_n)$ — упорядоченный кортеж lowerdir (снизу вверх), $U$ — upperdir. Определим функцию разрешения пути:

$$\text{resolve}(path) = \begin{cases} U/path & \text{if exists}(U/path) \\ l_i/path & \text{if } \exists i: \text{exists}(l_i/path) \land \nexists j > i: \text{exists}(l_j/path) \\ \text{ENOENT} & \text{otherwise} \end{cases}$$

Для операции записи:

$$\text{write}(path, data) = \begin{cases} \text{write}(U/path, data) & \text{if exists}(U/path) \\ \text{copy\_up}(l_i/path \to U/path); \text{write}(U/path, data) & \text{otherwise} \end{cases}$$

Это функциональная композиция, где $U$ имеет приоритет над всеми $l_i$, а $l_{i+1}$ имеет приоритет над $l_i$. Иммутабельность $l_i$ гарантируется тем, что $\text{write}$ никогда не модифицирует $l_i$.

**Связь с Docker.**
- Docker image layer = lowerdir (read-only, content-addressable).
- Container layer = upperdir (read-write, удаляется при `docker rm`).
- При `docker commit` upperdir становится новым lowerdir (новый слой образа).

### 1.4. Сетевая модель и транспорт

#### 1.4.1. Модель OSI и TCP/IP: инкапсуляция и уровни

**Почему нужны уровни?**
Сетевое взаимодействие сложно: физические сигналы, адресация, маршрутизация, надёжность, шифрование, форматы данных. Уровни абстракции позволяют каждому слою решать свою задачу, не зная деталей других.

**Модель OSI (Open Systems Interconnection) — 7 уровней:**

| Уровень | Название | Примеры протоколов | Единица данных |
|---|---|---|---|
| 7 | Прикладной (Application) | HTTP, FTP, SMTP, DNS | Message |
| 6 | Представления (Presentation) | TLS/SSL (частично), ASCII, JPEG, MPEG | Message |
| 5 | Сеансовый (Session) | NetBIOS, RPC, PPTP | Message |
| 4 | Транспортный (Transport) | TCP, UDP, SCTP | Segment / Datagram |
| 3 | Сетевой (Network) | IP, ICMP, IGMP, IPSec | Packet |
| 2 | Канальный (Data Link) | Ethernet, Wi-Fi (802.11), PPP, ARP | Frame |
| 1 | Физический (Physical) | USB, Ethernet (кабель), оптоволокно, радио | Bit |

**Стек TCP/IP — 4 уровня (практическая модель):**

| TCP/IP | Соответствие OSI | Протоколы |
|---|---|---|
| Application (Прикладной) | 5–7 | HTTP, DNS, SSH, TLS |
| Transport (Транспортный) | 4 | TCP, UDP |
| Internet (Межсетевой) | 3 | IP, ICMP |
| Link (Канальный) | 1–2 | Ethernet, Wi-Fi, ARP |

**Инкапсуляция.**
Каждый уровень добавляет свой заголовок (header) к данным верхнего уровня:

```
[HTTP Data]
  -> + TCP Header = TCP Segment
    -> + IP Header = IP Packet
      -> + Ethernet Header + Ethernet Trailer = Ethernet Frame
```

**Математическая подоплёка.**
Инкапсуляция — это композиция функций. Пусть $D$ — полезная нагрузка (payload). Тогда:

$$\text{Frame} = H_{\text{link}} \circ H_{\text{net}} \circ H_{\text{trans}} \circ H_{\text{app}}(D)$$

где $H_i(x) = \text{header}_i \parallel x$ — конкатенация заголовка и данных. Декапсуляция (приёме) — обратная операция: извлечение заголовка и передача остатка вышестоящему уровню.

#### 1.4.2. TCP: надёжность, потоковость и конечный автомат

**TCP Segment.**
Заголовок TCP минимум 20 байт:
- Source Port (16 бит), Destination Port (16 бит).
- Sequence Number (32 бита) — номер первого байта данных в сегменте.
- Acknowledgment Number (32 бита) — номер следующего ожидаемого байта.
- Data Offset, Flags (SYN, ACK, FIN, RST, PSH, URG).
- Window Size (16 бит) — размер приёмного окна (flow control).
- Checksum, Urgent Pointer.
- Options (SACK, timestamps, window scale).

**Three-way handshake (установление соединения).**

```
Клиент                    Сервер
  | ---- SYN, seq=x ----> |    (Клиент выбирает начальный sequence number)
  | <--- SYN-ACK, seq=y, ack=x+1 --- |  (Сервер выбирает свой ISN, подтверждает)
  | ---- ACK, ack=y+1 ---> |   (Клиент подтверждает, соединение установлено)
```

- SYN (synchronize) — запрос на установление соединения.
- ACK (acknowledge) — подтверждение.
- ISN (Initial Sequence Number) выбирается случайно для защиты от hijacking.

**Надёжная доставка.**
Каждый отправленный байт имеет sequence number. Получатель подтверждает (ACK) номер следующего ожидаемого байта. Если отправитель не получает ACK в течение RTO (Retransmission Timeout), он повторно отправляет сегмент.

**Скользящее окно (Sliding Window) — управление потоком.**
Получатель объявляет размер своего буфера (Window Size). Отправитель может послать столько данных, сколько поместится в окно, не дожидаясь ACK на каждый сегмент. По мере поступления ACK окно «скользит» вперёд.

**Контроль перегрузки (Congestion Control).**
TCP не только управляет потоком между двумя хостами, но и адаптируется к состоянию сети (маршрутизаторы, пропускная способность каналов).

**Алгоритм AIMD (Additive Increase / Multiplicative Decrease):**
- **Slow Start:** начинаем с $cwnd = 1$ MSS (Maximum Segment Size). При каждом ACK увеличиваем $cwnd$ на 1 MSS. Экспоненциальный рост: $cwnd = 2^n$ за $n$ RTT.
- **Congestion Avoidance:** когда $cwnd$ достигает `ssthresh` (slow start threshold), переходим к аддитивному увеличению: $cwnd \leftarrow cwnd + \frac{1}{cwnd}$ за каждый ACK (приблизительно +1 MSS за RTT).
- **Fast Retransmit:** если получено 3 дублирующих ACK (dup ACK), считаем, что сегмент потерян, немедленно повторяем его, не дожидаясь таймаута.
- **Fast Recovery:** вместо сброса в Slow Start, устанавливаем $cwnd = ssthresh = \frac{cwnd}{2}$ и продолжаем линейный рост.

**Математическая подоплёка.**
AIMD — это итеративный процесс оптимизации:

$$w_{n+1} = \begin{cases} w_n + \alpha & \text{(additive increase, no loss)} \\ \beta \cdot w_n & \text{(multiplicative decrease, loss detected)} \end{cases}$$

где $\alpha = 1$ MSS, $\beta = 0.5$. Этот процесс сходится к **равновесной точке** (equilibrium), где пропускная способность колеблется около оптимума. Доказано, что AIMD максимизирует использование канала при справедливом разделении между потоками.

**Конечный автомат TCP.**
TCP-соединение проходит через дискретные состояния:

$$\text{CLOSED} \xrightarrow{\text{active open}} \text{SYN\_SENT} \xrightarrow{\text{SYN-ACK}} \text{ESTABLISHED}$$

$$\text{ESTABLISHED} \xrightarrow{\text{close()}} \text{FIN\_WAIT\_1} \xrightarrow{\text{ACK}} \text{FIN\_WAIT\_2} \xrightarrow{\text{FIN}} \text{TIME\_WAIT} \xrightarrow{2 \times \text{MSL}} \text{CLOSED}$$

Состояние **TIME_WAIT** (длительностью $2 \times \text{MSL}$, обычно 60 секунд) гарантирует, что последний ACK дошёл до пассивного закрывающего, и «задержанные» пакеты от предыдущего соединения не попадут в новое с теми же портами.

#### 1.4.3. UDP: stateless, быстрота и применимость

**UDP (User Datagram Protocol)** — минималистичный транспортный протокол без установления соединения. Заголовок UDP всего 8 байт:
- Source Port (16 бит), Destination Port (16 бит).
- Length (16 бит) — длина датаграммы (заголовок + данные).
- Checksum (16 бит) — контрольная сумма (опциональная в IPv4, обязательная в IPv6).

**Отличия от TCP:**
- **Нет установления соединения.** Нет handshake, нет состояний. Отправитель просто посылает датаграмму.
- **Нет гарантии доставки.** Потерянные пакеты не переотправляются. Нет ACK, нет sequence numbers, нет retransmission.
- **Нет упорядоченности.** Датаграммы могут прийти в произвольном порядке или дублироваться.
- **Нет контроля перегрузки.** UDP не адаптирует скорость. Это позволяет достичь максимальной пропускной способности, но может привести к коллапсу сети (bufferbloat).

**Когда UDP предпочтительнее TCP:**
- **DNS-запросы.** Один запрос — один ответ. Накладные расходы TCP (handshake + tear-down = 3 RTT минимум) неоправданны для ~100 байт данных.
- **Метрики и логи.** StatsD, Collectd, syslog-over-UDP. Потеря 1% метрик некритична, важна скорость и низкая задержка.
- **Видео/аудио потоки.** RTP (Real-time Transport Protocol) поверх UDP. Потерянный кадр лучше пропустить, чем ждать retransmission и получить jitter.
- **Квorum/heartbeat в распределённых системах.** Gossip-протоколы, membership protocols (SWIM). Частые маленькие сообщения, где потеря отдельного heartbeat не фатальна.

**Математическая подоплёка.**
UDP можно моделировать как **ненадёжный канал** в теории информации. Пропускная способность канала Шеннона:

$$C = B \cdot \log_2\left(1 + \frac{S}{N}\right)$$

где $B$ — полоса пропускания, $S/N$ — отношение сигнал/шум. UDP стремится максимально приблизиться к $C$, не тратя полосу на служебный трафик (ACK, retransmission). TCP же жертвует частью $C$ на надёжность и справедливость (fairness).

#### 1.4.4. Сокеты, порты, NAT и DNS-резолвинг

**Сокет (socket) — абстракция конечной точки коммуникации.**
В Unix «всё есть файл», и сокет — это тоже файл (тип `SOCK_STREAM` для TCP, `SOCK_DGRAM` для UDP). Системные вызовы:
- `socket(domain, type, protocol)` — создание сокета.
- `bind(sockfd, addr, addrlen)` — привязка к локальному адресу и порту.
- `listen(sockfd, backlog)` — перевод в режим ожидания входящих соединений (только TCP).
- `accept(sockfd, addr, addrlen)` — извлечение установленного соединения из очереди.
- `connect(sockfd, addr, addrlen)` — установление соединения (TCP) или задание адреса по умолчанию (UDP).
- `send()` / `recv()` — передача данных.

**Порты.**
TCP и UDP используют 16-битные номера портов (0–65535):
- **0–1023:** Well-known ports (привилегированные, требуют root/CAP_NET_BIND_SERVICE). HTTP (80), HTTPS (443), SSH (22), SMTP (25).
- **1024–49151:** Registered ports. PostgreSQL (5432), Redis (6379), Kafka (9092).
- **49152–65535:** Dynamic / private / ephemeral ports. Используются клиентами как source port при исходящих соединениях.

**NAT (Network Address Translation).**
Проблема: IPv4-адресов недостаточно (~4.3 млрд, из которых значительная часть зарезервирована). NAT позволяет множеству устройств в частной сети выходить в интернет через один публичный IP.

**Типы NAT:**
- **SNAT (Source NAT / Masquerade):** замена source IP:port частного хоста на публичный IP:port маршрутизатора. Таблица NAT хранит отображение: `(внутренний_IP, внутренний_port, внешний_IP, внешний_port) <-> (публичный_IP, публичный_port)`.
- **DNAT (Destination NAT / Port Forwarding):** перенаправление входящего соединения на публичном порте во внутренний хост.

**NAT и TCP-соединения.**
NAT-маршрутизатор отслеживает состояние TCP-соединений (stateful). Если соединение долго неактивно, запись в таблице NAT истекает (timeout обычно 5–15 минут). Это означает, что **keepalive** (TCP_KEEPALIVE) или application-level heartbeat необходимы для долгоживущих соединений через NAT.

**DNS (Domain Name System).**
Иерархическая распределённая база данных, отображающая доменные имена на IP-адреса (и наоборот — PTR-записи).

**Типы DNS-записей:**
- **A:** IPv4-адрес.
- **AAAA:** IPv6-адрес.
- **CNAME:** каноническое имя (alias).
- **MX:** почтовый сервер.
- **TXT:** текстовые данные (SPF, DKIM).
- **SRV:** сервисные записи (хост + порт).
- **PTR:** обратное отображение IP -> домен.

**Резолвинг в Docker.**
Docker создаёт собственный DNS-сервер (`dockerd` встраивает resolver). Контейнеры в одной пользовательской bridge-сети могут обращаться друг к другу по имени сервиса (из `docker-compose.yml`). Это работает через:
- Встроенный DNS в Docker Engine (слушает 127.0.0.11:53 внутри контейнера).
- Таблицу `/etc/hosts` как fallback.
- Пользовательский DNS (`--dns`).

**Математическая подоплёка.**
NAT — это **частичная биекция** (partial bijection) между множеством внутренних пар `(IP_{in}, port_{in})$ и множеством внешних пар $(IP_{out}, port_{out})$:

$$\text{NAT}: (IP_{in}, port_{in}) \times (IP_{dst}, port_{dst}) \to (IP_{out}, port_{out})$$

Ограничение: $|\text{Range}(\text{NAT})| \leq 2^{16} \times N_{pub\_ips}$. При большом количестве соединений (например, тысячи WebSocket от одного хоста) публичные порты могут исчерпаться.

### 1.5. Итог модуля: от процесса к контейнеру

**Что мы построили:**

| Концепция | Зачем нужна | Как Docker это использует |
|---|---|---|
| **Процесс** (`task_struct`) | Единица выполнения в ОС | Контейнер = процесс(ы) в изолированных namespaces |
| **Виртуальная память** (page tables, TLB) | Изоляция и безопасность | Каждый контейнер имеет собственное адресное пространство |
| **Файловые дескрипторы** | Абстракция ресурсов | Контейнер наследует/ограничивает fd; stdout/stderr логируются |
| **`fork()` + COW** | Быстрое создание процессов | `docker exec` использует clone(); образы используют COW для слоёв |
| **Namespaces** | Изоляция видимости ресурсов | PID, NET, MNT, UTS, IPC, USER, CGROUP — всё это Docker |
| **Cgroups** | Ограничение потребления ресурсов | `--memory`, `--cpus`, `--pids-limit` |
| **Capabilities** | Тонкое разграничение прав | `--cap-drop`, `--cap-add`; root в контейнере безопасен |
| **OverlayFS** | Слоистая файловая система | Docker image layers + container writable layer |
| **TCP** | Надёжный потоковый транспорт | Docker API, registry, межсервисное взаимодействие |
| **UDP** | Быстрый ненадёжный транспорт | DNS-запросы, метрики, syslog |
| **NAT** | Мультиплексирование IP | Docker bridge, port mapping (`-p 8080:80`) |
| **DNS** | Разрешение имён | Встроенный DNS для service discovery в Compose |

**Ключевой вывод:** контейнер — это не «лёгкая виртуальная машина». Контейнер — это **процесс (или группа процессов)**, к которому применены механизмы изоляции Linux (namespaces) и ограничения ресурсов (cgroups), работающий поверх общего ядра хоста. Все эти механизмы существовали в Linux задолго до Docker (namespaces — с 2002, cgroups — с 2007). Docker — это **оркестратор** и **интерфейс** над этими примитивами.

**Переход к следующему модулю.**
Теперь, когда мы понимаем, как ядро изолирует и ограничивает процессы, мы можем изучить, как Docker упаковывает эти механизмы в удобный инструмент: Dockerfile, слои, кэширование, multi-stage builds, сети и безопасность. Это Модуль 2.
