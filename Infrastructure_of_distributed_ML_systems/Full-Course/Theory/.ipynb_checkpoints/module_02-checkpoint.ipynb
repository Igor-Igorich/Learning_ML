## Модуль 2. Docker: анатомия контейнера и образа

### 2.1. Архитектура Docker и экосистема OCI

#### 2.1.1. Эволюция Docker: от монолита к слоистой архитектуре

**Ранний Docker (2013–2015).**
Первоначально Docker представлял собой монолитный демон, который делал всё: парсил Dockerfile, управлял образами, создавал контейнеры, настраивал сети и тома. Внутри он использовал LXC (Linux Containers) — набор userspace-инструментов для работы с namespaces и cgroups. Это было просто, но создавало проблемы:
- Привязка к конкретной версии LXC.
- Невозможность использовать Docker-контейнеры без Docker (vendor lock-in).
- Сложность интеграции с оркестраторами (Kubernetes, Mesos).

**Декомпозиция (2016–н.в.).**
Docker разделили на независимые компоненты, каждый из которых отвечает за свой слой абстракции:

```
┌─────────────────────────────────────────┐
│         Docker Client (CLI)             │
│           docker build, run             │
└─────────────────┬───────────────────────┘
                  │ REST API (Unix socket / TCP)
┌─────────────────▼───────────────────────┐
│         Docker Daemon (dockerd)         │
│  API-шлюз, билдер, сети, volumes,       │
│  плагины, swarm-режим                   │
└─────────────────┬───────────────────────┘
                  │ gRPC (containerd socket)
┌─────────────────▼───────────────────────┐
│            containerd                   │
│  Управление жизненным циклом: pull,     │
│  push, snapshots, runtime, tasks        │
│  (CRI-совместимый runtime для K8s)      │
└─────────────────┬───────────────────────┘
                  │ OCI Runtime Spec (JSON + rootfs)
┌─────────────────▼───────────────────────┐
│              runc                       │
│  Создание namespaces, cgroups,          │
│  capabilities, seccomp, pivot_root      │
│  (чистый Go, без зависимостей)          │
└─────────────────────────────────────────┘
```

**Зачем такая многоуровневость?**

1. **containerd** (донated в CNCF в 2017) — это «runtime для runtimes». Он управляет:
   - **Images:** pull/push, распаковка слоёв, content store (адресация по digest).
   - **Snapshots:** подготовка rootfs через snapshotter (overlayfs, btrfs, zfs, native).
   - **Runtime:** вызов runc (или другого OCI-совместимого runtime: crun, kata, gVisor).
   - **Tasks:** процессы контейнеров, stdin/stdout/stderr (через FIFO), сигналы.

Kubernetes напрямую общается с containerd через CRI (Container Runtime Interface), минуя dockerd. Это позволяет Kubernetes работать с containerd, CRI-O или любым другим CRI-совместимым runtime.

2. **runc** — это «последняя миля». Он получает от containerd:
   - OCI Runtime Spec (JSON-файл с конфигурацией: namespaces, cgroups, mounts, env, args, capabilities).
   - Rootfs (распакованные слои образа).
   - Создаёт процесс через `clone()` с нужными флагами namespaces, настраивает cgroups, применяет seccomp, делает `pivot_root()` в rootfs, и запускает `user-defined command`.

runc не демонизируется. Он запускается, настраивает окружение, запускает процесс контейнера, и ждёт его завершения. Если процесс контейнера — PID 1 — завершается, runc тоже завершается.

#### 2.1.2. Client-server модель и коммуникация

**Docker Daemon** слушает Unix-сокет `/var/run/docker.sock` (по умолчанию). Это доменный сокет (Unix domain socket), работающий через файловую систему, а не через сетевой стек. Он предоставляет REST API.

**Ключевые endpoint-ы API:**
- `POST /build` — сборка образа из Dockerfile и build context.
- `POST /containers/create` — создание контейнера из образа.
- `POST /containers/{id}/start` — запуск.
- `GET /containers/{id}/json` — инспекция.
- `GET /images/{name}/json` — инспекция образа.

**Build context.**
Когда вы выполняете `docker build .`, Docker CLI:
1. Архивирует текущую директорию (build context) в tar.
2. Передаёт её демону через REST API.
3. Демон распаковывает context во временную директорию.
4. Все инструкции `COPY` и `ADD` в Dockerfile работают с этой временной копией, а не с исходной файловой системой клиента.

Это объясняет, почему `COPY ../file.txt .` не работает: build context ограничен директорией, переданной демону. Также это объясняет важность `.dockerignore`: каждый лишний файл в context увеличивает время передачи и размер временной директории.

#### 2.1.3. OCI: стандартизация контейнеров

**Open Container Initiative (OCI)** — проект под эгидой Linux Foundation, созданный в 2015 году Docker, CoreOS и другими. Два ключевых стандарта:

1. **OCI Image Specification.**
   Описывает формат образа: manifest, configuration, layers.
   - **Manifest** (JSON): список слоёв (по digest), конфигурация контейнера (env, entrypoint, cmd, exposed ports, labels).
   - **Config** (JSON): детали runtime: пользователь, working dir, env, cmd, entrypoint, volumes, labels.
   - **Layers:** tar-архивы, сжатые gzip. Каждый слой — дельта изменений относительно предыдущего.

2. **OCI Runtime Specification.**
   Описывает, как runc (или аналог) должен создавать контейнер:
   - `config.json` (OCI Runtime Spec): версия, процесс (args, env, cwd), root (path, readonly), hostname, mounts, linux (namespaces, uidMappings, gidMappings, sysctl, resources — cgroups, capabilities, rlimits, rootfsPropagation, seccomp, apparmorProfile, selinuxLabel, mounts).

**Content-addressable storage.**
Все объекты в Docker (слои, конфиги, манифесты) адресуются не по имени, а по **digest** — SHA-256 хэшу содержимого.

$$\text{digest} = \text{sha256}(\text{content})$$

Если два образа содержат идентичный слой (например, базовый `python:3.11-slim`), их digest совпадает, и слой хранится один раз (deduplication). Если вы изменили один байт в слое, digest меняется полностью (лавинный эффект хэш-функций).

**Математическая подоплёка.**
SHA-256 — криптографическая хэш-функция:

$$H: \{0,1\}^* \to \{0,1\}^{256}$$

Свойства:
- **Детерминизм:** $H(x) = H(x)$ всегда.
- **Лавинный эффект:** $H(x)$ и $H(x \oplus e_i)$ (флип одного бита) коррелированы как случайные строки.
- **Стойкость к коллизиям:** вероятность случайной коллизии для $N$ объектов:

$$P(\text{collision}) \approx 1 - e^{-N(N-1)/(2 \cdot 2^{256})} \approx \frac{N^2}{2^{257}}$$

Для $N = 10^{20}$ (невообразимо больше всех существующих образов) вероятность коллизии пренебрежимо мала. Поэтому digest можно использовать как уникальный идентификатор.

**Детерминированность сборки.**
Идеальный Docker build — чистая функция:

$$\text{Image} = F(\text{Dockerfile}, \text{BuildContext}, \text{BaseImage})$$

Если входные параметры идентичны, выход должен быть бит-в-бит идентичен. На практике это нарушается `apt-get update` (время меняется), `RUN npm install` (semver может подтянуть патчи), `ARG` без значений по умолчанию. BuildKit решает часть проблем через reproducible builds, но полная детерминированность требует pinned версий (`requirements.txt` с хэшами, `package-lock.json`, `poetry.lock`).

### 2.2. Анатомия Dockerfile

#### 2.2.1. Инструкции: семантика и механика

**`FROM` — базовый образ.**
Каждый Dockerfile (кроме scratch-образов) начинается с `FROM`. Это создаёт первый слой образа, наследуя все слои базового образа. Многоэтапные сборки используют несколько `FROM` (см. 2.4).

**`RUN` — выполнение команды в новом слое.**
Команда выполняется во временном контейнере поверх текущих слоёв. Результат (изменения файловой системы) фиксируется как новый read-only слой. Например:

```
RUN apt-get update && apt-get install -y gcc
```

Создаёт слой, содержащий все установленные пакеты. Важно: каждая инструкция `RUN` создаёт **отдельный слой** (если не используется BuildKit с heredocs или squash).

**`COPY` — копирование из build context.**

```
COPY source dest
```

Копирует файлы/директории из build context (переданного демону) в текущий слой образа. Создаёт новый слой. Поддерживает `--chown` для установки владельца.

**`ADD` — расширенное копирование.**
Как `COPY`, но с двумя дополнительными возможностями:
1. Если `source` — URL, скачивает файл.
2. Если `source` — локальный tar-архив (gzip, bzip2, xz), автоматически распаковывает его.

Рекомендация: использовать `COPY` везде, кроме случаев, когда явно нужна распаковка tar. `ADD` с URL не поддерживает кэширование (каждый build заново скачивает), что делает сборку недетерминированной.

**`ENV` — переменные окружения.**

```
ENV PYTHONDONTWRITEBYTECODE=1
ENV PATH="/app/.local/bin:${PATH}"
```

Создаёт слой, содержащий переменные окружения. Они доступны как во время сборки (последующие `RUN`), так и во время выполнения контейнера. Каждая `ENV` создаёт слой (хотя и очень маленький).

**`ARG` — аргументы сборки.**

```
ARG PYTHON_VERSION=3.11
FROM python:${PYTHON_VERSION}-slim
```

`ARG` — это переменные, существующие **только во время сборки**. Они не попадают в итоговый образ (если только не используются в `ENV`). Это важно для безопасности: `ARG` можно использовать для передачи секретов на этап сборки, не оставляя их в слоях образа.

Разница критична:
- `ENV` — в образе, видно через `docker inspect` и внутри контейнера.
- `ARG` — только при сборке, не видно в runtime.

**`WORKDIR` — рабочая директория.**

```
WORKDIR /app
```

Создаёт директорию (если не существует) и делает её текущей для последующих инструкций. Эквивалент `cd /app`, но без создания лишних слоёв (инструкция `WORKDIR` сама по себе создаёт минимальный слой метаданных).

**`EXPOSE` — документация портов.**

```
EXPOSE 8000
```

Не открывает порт и не делает его доступным снаружи. Это **декларативная метка**, используемая для документации и автоматического маппинга (`-P` маппит все exposed порты на случайные порты хоста). Фактическое открытие порта происходит через `docker run -p 8000:8000`.

**`CMD` — команда по умолчанию.**

```
CMD ["python", "main.py"]
```

Определяет команду, выполняемую при запуске контейнера. Переопределяется аргументами `docker run`: `docker run myimage python other.py`.

**`ENTRYPOINT` — точка входа.**

```
ENTRYPOINT ["python"]
CMD ["main.py"]
```

`ENTRYPOINT` не переопределяется аргументами `docker run` легко (только через `--entrypoint`). Аргументы `docker run` дописываются к `ENTRYPOINT`, заменяя `CMD`.

#### 2.2.2. Shell-форма vs Exec-форма

**Exec-форма (JSON array):**

```
CMD ["python", "main.py"]
ENTRYPOINT ["python", "main.py"]
```

- Первый элемент — исполняемый файл (без shell).
- Остальные — аргументы.
- PID 1 контейнера = `python` (прямой запуск).
- Сигналы (SIGTERM) доставляются напрямую процессу `python`.

**Shell-форма:**

```
CMD python main.py
ENTRYPOINT python main.py
```

- Docker оборачивает команду в `/bin/sh -c "..."`.
- PID 1 = `/bin/sh`.
- `python` — дочерний процесс shell.
- **Проблема:** shell не передаёт сигналы дочерним процессам по умолчанию. `docker stop` посылает SIGTERM PID 1 (`sh`), но `sh` ждёт завершения `python` и не пересылает сигнал. Контейнер получает SIGKILL через 10 секунд (grace period). Это ломает graceful shutdown.
- **Решение:** использовать `exec` в shell-форме: `CMD exec python main.py` (замещает shell на python).

**Паттерн ENTRYPOINT + CMD:**

```
ENTRYPOINT ["uvicorn", "main:app"]
CMD ["--host", "0.0.0.0", "--port", "8000"]
```

- `docker run myimage` -> `uvicorn main:app --host 0.0.0.0 --port 8000`.
- `docker run myimage --port 9000` -> `uvicorn main:app --port 9000` (CMD заменяется).
- `docker run --entrypoint python myimage script.py` -> `python script.py`.

Это позволяет создавать образы, ведущие себя как команды (аналог Unix-утилит), с настраиваемыми аргументами.

#### 2.2.3. Build context и `.dockerignore`

**Build context** — это содержимое директории, переданной аргументом `docker build`:

```
docker build .          # context = текущая директория
docker build -f Dockerfile ../  # context = родительская директория
```

CLI архивирует context в tar и передаёт демону. Все файлы, включая `.git`, `node_modules`, `__pycache__`, `.env`, логи, временные файлы, попадают в архив. Это:
- Увеличивает время передачи (особенно по сети к удалённому демону).
- Увеличивает потребление RAM демоном.
- Может привести к случайному копированию секретов (`COPY . .` скопирует `.env` с паролями).

**`.dockerignore`** — файл в корне context, синтаксис аналогичен `.gitignore`:

```
# Игнорировать
.git
__pycache__
*.pyc
.env
node_modules
*.log
Dockerfile
.dockerignore

# Но не игнорировать
!requirements.txt
!src/
```

**Математическая подоплёка.**
Build context можно моделировать как множество файлов $F = \{f_1, f_2, ..., f_n\}$ с весами $w(f_i)$ (размер). `.dockerignore` определяет фильтрующую функцию:

$$\text{context} = \{f \in F \mid \neg \text{ignored}(f)\}$$

Стоимость передачи:

$$T = \sum_{f \in \text{context}} w(f) \cdot t_{network} + t_{archive} + t_{extract}$$

Минимизация $T$ — оптимизационная задача. На практике `.dockerignore` может уменьшить context с 500 МБ (`node_modules`, `.git`) до 5 МБ.

#### 2.2.4. Порядок инструкций и кэширование

**Ключевой принцип:** Docker кэширует результат выполнения каждой инструкции. Если инструкция и её входные данные не изменились, Docker использует кэшированный слой, пропуская выполнение.

**Механизм:**
1. Для каждой инструкции Docker вычисляет **cache key** — хэш от:
   - Текста инструкции (например, `RUN pip install -r requirements.txt`).
   - Хэшей всех файлов, участвующих в инструкции (для `COPY` — хэш копируемых файлов из build context).
   - Хэша предыдущего слоя (parent layer).

2. Если cache key совпадает с существующим слоём в локальном кэше, слой берётся из кэша.

3. Если cache key не совпадает (инструкция изменилась или файлы изменились), слой пересобирается. **Все последующие инструкции также пересобираются**, так как их parent layer изменился.

**Классическая антипаттерн:**

```
FROM python:3.11-slim
COPY . /app          # Копируем ВЕСЬ проект
WORKDIR /app
RUN pip install -r requirements.txt  # Устанавливаем зависимости
CMD ["python", "main.py"]
```

Проблема: при изменении любого файла в проекте (например, `main.py`) инвалидация кэша происходит на `COPY . /app`. Слой `RUN pip install` пересобирается заново, хотя `requirements.txt` не менялся. Установка зависимостей занимает минуты.

**Оптимизированный порядок:**

```
FROM python:3.11-slim
WORKDIR /app

# Сначала копируем ТОЛЬКО файл с зависимостями
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Теперь копируем остальной код (меняется чаще)
COPY . .

CMD ["python", "main.py"]
```

Почему это работает:
- `requirements.txt` меняется редко (только при добавлении/обновлении зависимостей).
- `RUN pip install` кэшируется в 99% сборок.
- `COPY . .` инвалидирует только последний слой (копирование кода), который выполняется за миллисекунды.

**Обобщённая стратегия:**
Упорядочивайте инструкции от **наиболее стабильных** (меняются редко) к **наиболее изменчивым** (меняются часто):

1. `FROM` (базовый образ — меняется при обновлении Python).
2. `RUN apt-get install` (системные зависимости — меняются редко).
3. `COPY requirements.txt` + `RUN pip install` (зависимости — меняются умеренно).
4. `COPY . .` (исходный код — меняется при каждом коммите).
5. `CMD` / `ENTRYPOINT` (метаданные — почти не меняются).

**Математическая подоплёка.**
Пусть Dockerfile содержит $n$ инструкций, каждая инструкция $i$ имеет вероятность инвалидации $p_i$ (вероятность, что входные данные изменились между сборками). Ожидаемое время сборки:

$$E[T] = \sum_{i=1}^{n} \left( \prod_{j=1}^{i-1} (1-p_j) \right) \cdot t_i^{cache} + \left(1 - \prod_{j=1}^{i-1} (1-p_j)\right) \cdot t_i^{rebuild}$$

где $t_i^{cache} \approx 0$ (извлечение из кэша), а $t_i^{rebuild}$ — полное время выполнения. Чтобы минимизировать $E[T]$, нужно минимизировать $p_i$ для дорогих инструкций (больших $t_i^{rebuild}$). Это и есть принцип «стабильное сверху».

### 2.3. Кэширование слоёв: механика

#### 2.3.1. Вычисление cache ID

Каждый слой образа идентифицируется **cache ID** (или layer ID в старой терминологии). В современном Docker с BuildKit это называется **digest** слоя.

Для инструкции `RUN` cache key вычисляется как:
$$\text{cache\_key} = \text{Hash}(\text{parent\_layer\_digest}, \text{instruction\_text}, \text{build\_args})$$

Для инструкции `COPY`:
$$\text{cache\_key} = \text{Hash}(\text{parent\_layer\_digest}, \text{file\_metadata}, \text{file\_contents})$$

`file_metadata` включает: пути, размеры, permissions, timestamps, inode numbers (но не всегда — BuildKit может использовать content-based hashing).

**Content-based vs timestamp-based hashing.**
Старый Docker использовал timestamp файлов. Если вы делали `touch file.txt`, кэш инвалидировался, хотя содержимое не менялось. BuildKit использует content-based hashing (хэш содержимого файла), что делает кэширование более агрессивным и правильным.

#### 2.3.2. Immutable слои и writable container layer

Образ Docker — это упорядоченный список read-only слоёв. Контейнер добавляет сверху writable layer:

$$\text{Container FS} = \text{OverlayFS}(\text{layer}_1, \text{layer}_2, ..., \text{layer}_n, \text{container\_layer})$$

При удалении контейнера (`docker rm`) writable layer удаляется. Все изменения, сделанные внутри контейнера (записанные файлы, логи), исчезают, если не используются volumes.

**Иммутабельность слоёв образа.**
После сборки слои образа никогда не меняются. Это даёт:
- **Дедупликацию:** один и тот же базовый слой (`python:3.11-slim`) используется множеством образов, хранится один раз.
- **Кэширование:** слои из registry не скачиваются повторно.
- **Безопасность:** атакующий не может модифицировать слои уже собранного образа.

#### 2.3.3. BuildKit: следующее поколение билдера

BuildKit (включён по умолчанию с Docker 20.10+) — это переписанный с нуля билдер, использующий DAG (Directed Acyclic Graph) для параллельного выполнения независимых инструкций.

**Возможности BuildKit:**
1. **Параллельная сборка.** Независимые `RUN` инструкции выполняются одновременно.
2. **Cache mounts.** Возможность смонтировать кэш-директорию между сборками:

```
   RUN --mount=type=cache,target=/root/.cache/pip \
       pip install -r requirements.txt
```

Кэш `pip` сохраняется между сборками, но не попадает в итоговый образ.
3. **Secret mounts.** Передача секретов на этап сборки без попадания в слои:

```
   RUN --mount=type=secret,id=mysecret,target=/run/secrets/mysecret \
       cat /run/secrets/mysecret
```

Секрет монтируется как tmpfs и не записывается в layer.
4. **SSH mounts.** Передача SSH-ключей для `git clone` приватных репозиториев.
5. **Squash.** Объединение слоёв в один (экспериментально).

**Математическая подоплёка.**
BuildKit представляет Dockerfile как DAG, где вершины — операции (vertex), а рёбра — зависимости (слои). Параллельное выполнение — это топологическая сортировка с максимальным параллелизмом:

$$\text{parallel\_time} = \max_{\text{path } P \in DAG} \sum_{v \in P} t(v)$$

где $t(v)$ — время выполнения вершины. Оптимальное расписание (list scheduling) приближает минимальное время для DAG с неопределённостью.

### 2.4. Многоэтапная сборка (Multi-stage builds)

#### 2.4.1. Проблема «толстых» образов

Python-образ для ML обычно требует:
- `gcc`, `g++`, `build-essential` — для компиляции C-расширений (`numpy`, `pandas`, `scikit-learn`, `psycopg2`).
- `python3-dev`, `libffi-dev`, `libssl-dev` — заголовочные файлы Python.
- `git` — для установки пакетов из репозиториев.
- Все эти инструменты занимают сотни мегабайт, но **не нужны в рантайме**. Модель использует только скомпилированные `.so` файлы и Python-код.

**Пример «толстого» образа:**

```
FROM python:3.11
RUN apt-get update && apt-get install -y gcc g++ build-essential python3-dev
COPY requirements.txt .
RUN pip install -r requirements.txt  # Компилирует numpy, pandas
COPY . .
CMD ["python", "main.py"]
```

Размер: ~1.5 ГБ (базовый образ `python:3.11` ≈ 900 МБ + зависимости + build tools).

#### 2.4.2. Паттерн `AS builder`

Многоэтапная сборка позволяет использовать несколько `FROM` в одном Dockerfile. Каждый `FROM` начинает новый этап (stage). Этапы могут копировать артефакты друг из друга.

```
# Этап 1: Сборка (builder)
FROM python:3.11 AS builder

WORKDIR /build
RUN apt-get update && apt-get install -y gcc g++ build-essential python3-dev

COPY requirements.txt .
RUN pip install --user --no-cache-dir -r requirements.txt

# Этап 2: Рантайм
FROM python:3.11-slim AS runtime

WORKDIR /app

# Копируем установленные пакеты из builder
COPY --from=builder /root/.local /root/.local
ENV PATH=/root/.local/bin:$PATH

# Копируем код приложения
COPY . .

CMD ["python", "main.py"]
```

**Что происходит:**
1. Этап `builder` использует полный образ `python:3.11` с build tools.
2. `pip install --user` устанавливает пакеты в `/root/.local` (вместо системного `/usr/local/lib/python3.11/`).
3. Этап `runtime` начинается с `python:3.11-slim` (~50 МБ вместо ~900 МБ).
4. `COPY --from=builder` копирует **только** результат установки (скомпилированные `.so`, `.py` файлы), не копируя `gcc`, `build-essential`, заголовочные файлы.
5. Итоговый образ содержит только рантайм.

**Результат:** образ уменьшается с 1.5 ГБ до 200–300 МБ.

#### 2.4.3. Базовые образы: сравнительный анализ

| Образ | Размер | glibc | shell | package manager | Когда использовать |
|---|---|---|---|---|---|
| `python:3.11` | ~900 МБ | Да | bash | apt | Builder stage |
| `python:3.11-slim` | ~50 МБ | Да | bash | apt | Runtime, если нужен apt |
| `python:3.11-alpine` | ~20 МБ | musl | ash | apk | Минимальный размер, но musl может ломать C-расширения |
| `gcr.io/distroless/python3` | ~30 МБ | Да | Нет | Нет | Максимальная безопасность, нет shell для escape |

**Alpine Linux и musl libc.**
Alpine использует `musl` вместо `glibc`. Musl легче и безопаснее, но:
- Некоторые Python-пакеты с C-расширениями (особенно связанные с ML: `numpy`, `scipy`, `tensorflow`) требуют перекомпиляции под musl или вообще не работают.
- DNS-разрешение в musl имеет особенности (нет `nsswitch.conf`, игнорирует `/etc/hosts` в некоторых случаях).

**Distroless (Google).**
Образы без shell, без package manager, без `curl`, без `ls`. Только runtime и приложение. Если злоумышленник получит RCE (Remote Code Execution) в контейнере, он не сможет:
- Открыть reverse shell (нет `/bin/sh`).
- Скачать эксплойт (нет `curl`, `wget`).
- Провести lateral movement (нет сетевых утилит).

Но отладка невозможна без `docker debug` или копирования инструментов извне.

#### 2.4.4. Анализ образов

**`docker history`:**

```
docker history myimage
```

Показывает каждый слой с командой, размером и временем создания. Позволяет найти «толстые» слои.

**`dive`:**
Интерактивный инструмент для анализа образа:
- Визуализация слоёв.
- Показ файлов, добавленных/удалённых/изменённых в каждом слое.
- Метрика «efficiency» — отношение полезных данных к общему размеру (учитывает overwritten файлы).

**`trivy`:**
Сканер уязвимостей:

```
trivy image myimage
```

Проверяет установленные пакеты (OS-level и Python-level) по базе CVE (Common Vulnerabilities and Exposures). Критично для production: даже если ваш код безопасен, устаревшая версия `openssl` или `libssl` в базовом образе может содержать известную уязвимость.

### 2.5. Сети и безопасность контейнеров

#### 2.5.1. Сетевые драйверы Docker

**Bridge (default).**
При установке Docker создаёт виртуальный мост `docker0` (или `br-xxxxxxxx` для user-defined networks) на хосте. Каждый контейнер получает:
- Собственный network namespace (изоляция сетевых интерфейсов).
- Пару виртуальных интерфейсов **veth** (virtual ethernet): один конец в namespace контейнера (`eth0`), другой — подключён к bridge на хосте.
- IP-адрес из подсети bridge (обычно `172.17.0.0/16` для default bridge).

**veth pair — механика.**
veth — это виртуальный кабель. Пакет, отправленный на один конец, появляется на другом. Bridge (`docker0`) работает как L2-switch: он изучает MAC-адреса и пересылает фреймы между подключёнными интерфейсами. Если пакет адресован контейнеру в той же подсети, bridge пересылает его локально, не выходя на физический интерфейс хоста.

**Host network.**

```
docker run --network host myimage
```

Контейнер использует сетевой namespace хоста напрямую. Нет изоляции, NAT, port mapping. Контейнер «видит» все интерфейсы хоста. Используется для максимальной производительности сети (нет накладных расходов на bridge/NAT) или когда нужен доступ к специфическим сетевым интерфейсам.

**None network.**

```
docker run --network none myimage
```

Контейнер получает только loopback (`lo`). Нет внешней сети. Используется для изоляции (например, контейнер с секретами, который не должен иметь сетевого доступа).

**Overlay network.**
Используется в Docker Swarm (и Kubernetes через CNI). Позволяет контейнерам на разных физических хостах общаться как в одной L2-сети через VXLAN-туннелирование.

#### 2.5.2. User-defined bridges и DNS

**Default bridge vs user-defined bridge.**
Контейнеры в default bridge (`docker0`) могут общаться по IP, но **не по имени**. DNS не работает.

User-defined bridge (создаётся `docker network create` или в `docker-compose.yml`):
- Встроенный DNS-сервер (встроен в dockerd, слушает `127.0.0.11:53` внутри контейнера).
- Контейнеры могут обращаться друг к другу по имени сервиса: `http://api:8000`, `redis://redis:6379`.
- Изоляция: контейнеры в разных bridge-сетях не видят друг друга.
- Автоматическое подключение/отключение при `docker run`.

**Как работает DNS в Docker:**
1. Контейнер отправляет запрос `api` на `127.0.0.11:53`.
2. Встроенный resolver dockerd проверяет:
   - Есть ли контейнер/сервис с таким именем в той же сети?
   - Если да — возвращает его IP.
   - Если нет — форвардит на внешний DNS (из `/etc/resolv.conf` хоста или `--dns`).

#### 2.5.3. Не-root пользователь и read-only rootfs

**Проблема root в контейнере.**
По умолчанию процессы в контейнере выполняются от `root` (UID 0). Если злоумышленник получит escape из контейнера (через уязвимость ядра, misconfigured volume, или `--privileged`), он окажется root на хосте.

**Решение: не-root пользователь.**

```
RUN useradd -m -u 1000 appuser
USER appuser
```

Теперь PID 1 работает от `appuser` (UID 1000). Даже при escape злоумышленник получит UID 1000 на хосте (если не совпадает с root хоста).

**Read-only rootfs:**

```
docker run --read-only myimage
```

Корневая файловая система монтируется read-only. Контейнер не может:
- Писать в `/etc`, `/usr`, `/app` (если `/app` не volume).
- Модифицировать бинарники.
- Создавать файлы в корне.

Для приложений, которым нужна временная запись (логи, кэш, uploads), используются volumes:

```
docker run --read-only -v /tmp -v /app/logs myimage
```

#### 2.5.4. Capabilities, Seccomp, AppArmor, SELinux

*(Продолжение темы из Модуля 1.2.3–1.2.4)*

**Capability dropping в Docker:**

```
docker run --cap-drop=ALL --cap-add=NET_BIND_SERVICE myimage
```

Удаляем все capabilities, добавляем обратно только необходимые. Это принцип **least privilege**.

**Seccomp:**

```
docker run --security-opt seccomp=default.json myimage
```

Профиль по умолчанию блокирует ~44 системных вызовов. Для ML-контейнеров, которым не нужны `mount`, `pivot_root`, `swapon`, `reboot`, это безопасно.

**AppArmor:**

```
docker run --security-opt apparmor=docker-default myimage
```

Профиль `docker-default` ограничивает доступ к файловой системе, сети, capabilities.

#### 2.5.5. Управление секретами

**Проблема секретов в образах.**

```
ENV DB_PASSWORD=supersecret
```

Секрет попадает в слой образа. Любой, у кого есть доступ к образу, может прочитать его:

```
docker history myimage  # видит ENV инструкции
docker run --rm -it myimage env  # видит переменные окружения
```

**Решения:**

1. **Runtime secrets (не в образе):**

```
   docker run -e DB_PASSWORD_FILE=/run/secrets/db_password myimage
```

Переменные окружения передаются при запуске, не при сборке.

2. **BuildKit secret mounts:**

```
   RUN --mount=type=secret,id=db_password,target=/run/secrets/db_password \
       pip install --index-url https://user:$(cat /run/secrets/db_password)@private.pypi.org/simple mypackage
```

Секрет монтируется как tmpfs во время `RUN` и не сохраняется в слое.

3. **Docker Secrets (Swarm mode) / Kubernetes Secrets:**
   Хранение секретов в orchestrator, монтирование в `/run/secrets/` как файлы.

### 2.6. Итог модуля: от Dockerfile к production-образу

**Что мы построили:**

| Концепция | Зачем нужна | Как применять |
|---|---|---|
| **Docker Engine -> containerd -> runc** | Разделение ответственности | Понимать, где происходит сборка, а где запуск |
| **OCI Spec** | Стандартизация | Образы переносимы между Docker, containerd, podman |
| **Content-addressable storage** | Дедупликация и целостность | Слои адресуются по SHA-256, не по имени |
| **Dockerfile инструкции** | Декларативное описание образа | Знать семантику каждой инструкции |
| **Shell vs Exec форма** | Корректная обработка сигналов | Всегда использовать Exec-форму для PID 1 |
| **ENTRYPOINT + CMD** | Гибкость образа | Образ ведёт себя как команда с аргументами |
| **Build context + .dockerignore** | Минимизация передаваемых данных | Исключать `.git`, `__pycache__`, `.env` |
| **Порядок инструкций** | Максимизация cache hit | Стабильное сверху, изменчивое снизу |
| **BuildKit** | Параллелизм, advanced caching | `--mount=type=cache`, `--mount=type=secret` |
| **Multi-stage builds** | Уменьшение размера образа | `AS builder` + `COPY --from=builder` |
| **Базовые образы** | Баланс размер/совместимость/безопасность | `slim` для Python, `distroless` для максимальной безопасности |
| **Сети** | Коммуникация между контейнерами | User-defined bridge для DNS, `--network none` для изоляции |
| **Безопасность** | Defense in depth | Не-root, `--read-only`, `--cap-drop`, seccomp |

**Ключевой вывод:** Docker-образ — это не просто «упаковка приложения». Это **функционально чистый артефакт** (идентифицируемый по digest), состоящий из иммутабельных слоёв, каждый из которых кэшируется и дедуплицируется. Правильно написанный Dockerfile минимизирует размер образа, максимизирует использование кэша, исключает секреты из слоёв и запускает процесс от непривилегированного пользователя.

**Переход к следующему модулю.**
Теперь, когда мы умеем создавать эффективные и безопасные образы, нам нужно научиться связывать несколько сервисов (API, БД, брокер сообщений) в единую среду, управлять их жизненным циклом, персистентностью данных и порядком запуска. Это Модуль 3 — Docker Compose и оркестрация мультисервисных приложений.
